# Category 2 — Binary Search on Rotated Sorted Array

---

## The Master Observation

A rotated sorted array always has **one sorted half and one unsorted half** at any `mid`. This is the invariant that makes BS possible. Every problem in this category reduces to:

> *"Identify the sorted half. Make your decision based on it."*

---

## Find Minimum — The Anchor Problem

Learn this first. Everything else builds on it.

**Key insight:** The minimum = the rotation point. It's the only place where $arr[i] < arr[i-1]$.

**Decision at `mid`:**

$$arr[mid] > arr[high] \Rightarrow \text{left half is sorted, min is in RIGHT} \Rightarrow lo = mid + 1$$
$$arr[mid] \leq arr[high] \Rightarrow \text{right half is sorted, mid could be min} \Rightarrow hi = mid$$

**Why compare with `arr[high]` not `arr[low]`?**
`arr[low]` can be the minimum itself (no rotation). Comparing with `arr[high]` is unambiguous — if `mid > high`, the break point is in the right half, guaranteed.

**Loop:** `lo < hi`, answer is `arr[lo]` when `lo == hi`.

---

## How Many Times Rotated?

$$\text{rotation count} = \text{index of minimum element}$$

Identical to Find Minimum — just return `lo` (the index) instead of `arr[lo]`.

---

## Search in Rotated Sorted Array (Unique Elements)

**Two-step decision at every `mid`:**

**Step 1:** Which half is sorted?
$$arr[lo] \leq arr[mid] \Rightarrow \text{LEFT half is sorted}$$
$$arr[lo] > arr[mid] \Rightarrow \text{RIGHT half is sorted}$$

**Step 2:** Is `target` in the sorted half? If yes → go there. If no → go to the other half.

| Sorted Half | Target in range? | Action |
|---|---|---|
| Left $[lo, mid]$ | $arr[lo] \leq t < arr[mid]$ | `hi = mid - 1` |
| Left $[lo, mid]$ | otherwise | `lo = mid + 1` |
| Right $[mid, hi]$ | $arr[mid] < t \leq arr[hi]$ | `lo = mid + 1` |
| Right $[mid, hi]$ | otherwise | `hi = mid - 1` |

**The delta vs standard BS:** You never directly compare `arr[mid]` with target first. You first orient yourself (which half is sorted), then decide.

**Loop:** `lo <= hi`, standard. Return `-1` if not found.

---

## Search in Rotated Sorted Array II (Duplicates)

**The breaking case:** When $arr[lo] == arr[mid] == arr[hi]$, you **cannot** determine which half is sorted.

$$[3, 1, 2, 3, 3, 3, 3] \quad arr[lo]=arr[mid]=arr[hi]=3 \Rightarrow \text{completely ambiguous}$$

**The fix:** When this happens, shrink both ends: `lo++; hi--`. This is the **only** change from the unique version.

**Why is this $O(n)$ worst case?**
An array of all identical elements forces you to shrink one step at a time, $n$ times.

---

## The Unified Decision Tree

```
At every mid in a rotated array:

1. Duplicates? → arr[lo]==arr[mid]==arr[hi] → lo++, hi--, continue

2. Which half sorted?
   arr[lo] <= arr[mid]  →  LEFT sorted
   else                 →  RIGHT sorted

3. Is target / min in the sorted half?
   YES → eliminate the other half
   NO  → eliminate the sorted half
```

````c++
#include <iostream>
#include <vector>
#include <algorithm>

using namespace std;
````

````c++
// ─── Find Minimum in Rotated Sorted Array ─────────────────────────────────
// Anchor: compare mid with high, not low
int findMin(vector<int>& arr) {
    int lo = 0, hi = (int)arr.size() - 1;
    while (lo < hi) {
        int mid = lo + (hi - lo) / 2;
        if (arr[mid] > arr[hi]) lo = mid + 1; // min in right half
        else                    hi = mid;       // mid is candidate, go left
    }
    return arr[lo];
}

// ─── How Many Times Rotated ───────────────────────────────────────────────
// Rotation count == index of minimum element
int countRotations(vector<int>& arr) {
    int lo = 0, hi = (int)arr.size() - 1;
    while (lo < hi) {
        int mid = lo + (hi - lo) / 2;
        if (arr[mid] > arr[hi]) lo = mid + 1;
        else                    hi = mid;
    }
    return lo; // index, not value
}

// ─── Search in Rotated Sorted Array (Unique) ──────────────────────────────
// Step 1: orient (which half sorted). Step 2: is target in sorted half?
int searchRotated(vector<int>& arr, int target) {
    int lo = 0, hi = (int)arr.size() - 1;
    while (lo <= hi) {
        int mid = lo + (hi - lo) / 2;
        if (arr[mid] == target) return mid;

        if (arr[lo] <= arr[mid]) {          // LEFT half is sorted
            if (arr[lo] <= target && target < arr[mid])
                hi = mid - 1;               // target in sorted left
            else
                lo = mid + 1;               // target in unsorted right
        } else {                            // RIGHT half is sorted
            if (arr[mid] < target && target <= arr[hi])
                lo = mid + 1;               // target in sorted right
            else
                hi = mid - 1;               // target in unsorted left
        }
    }
    return -1;
}

// ─── Search in Rotated Sorted Array II (Duplicates) ───────────────────────
// Only delta from unique version: handle arr[lo]==arr[mid]==arr[hi] → shrink both ends
bool searchRotatedII(vector<int>& arr, int target) {
    int lo = 0, hi = (int)arr.size() - 1;
    while (lo <= hi) {
        int mid = lo + (hi - lo) / 2;
        if (arr[mid] == target) return true;

        // The one extra case duplicates introduce
        if (arr[lo] == arr[mid] && arr[mid] == arr[hi]) {
            lo++; hi--;                     // can't orient, shrink both ends
            continue;
        }

        if (arr[lo] <= arr[mid]) {          // LEFT sorted
            if (arr[lo] <= target && target < arr[mid])
                hi = mid - 1;
            else
                lo = mid + 1;
        } else {                            // RIGHT sorted
            if (arr[mid] < target && target <= arr[hi])
                lo = mid + 1;
            else
                hi = mid - 1;
        }
    }
    return false;
}
````

````c++
vector<int> a = {4, 5, 6, 7, 0, 1, 2};
    cout << "Min: "          << findMin(a)              << "\n"; // 0
    cout << "Rotations: "    << countRotations(a)       << "\n"; // 4
    cout << "Search 0: "     << searchRotated(a, 0)     << "\n"; // 4
    cout << "Search 3: "     << searchRotated(a, 3)     << "\n"; // -1

    vector<int> b = {2, 5, 6, 0, 0, 1, 2};
    cout << "Search(dup) 0: "<< searchRotatedII(b, 0)   << "\n"; // true
    cout << "Search(dup) 3: "<< searchRotatedII(b, 3)   << "\n"; // false
````
