## 7.1 Topological Sort (DFS)

### State definition
Given a DAG $G=(V,E)$, a **topological order** is a linear ordering of $V$ such that for every
edge $(u,v) \in E$, $u$ appears before $v$ in the ordering. $\text{fin}[v]$ = the DFS finish time
from Subtopic 2.2.

### Invariant
Exactly Subtopic 2.2's disc/fin invariant, with one added consequence specific to DAGs: since a
DAG has **no back edges** (a back edge would mean $\text{color}[v]=G$ when examined, i.e. an active
ancestor — which is precisely a cycle, contradicting acyclicity), every non-tree edge in a DAG's
DFS is a forward or cross edge, both of which point to already-BLACK vertices — meaning **every
edge $(u,v)$ satisfies $\text{fin}[v] < \text{fin}[u]$.**

### Why it works — formal proof that reverse-finish-order is a valid topological order

**Claim:** sorting vertices by *decreasing* $\text{fin}[v]$ produces a valid topological order.

*Proof:* take any edge $(u,v) \in E$. We must show $u$ precedes $v$ in decreasing-finish-time
order, i.e. $\text{fin}[u] > \text{fin}[v]$. Consider the moment edge $(u,v)$ is examined during
$u$'s DFS call (so $\text{color}[u] = G$ at that instant):
- If $\text{color}[v] = W$: $v$ becomes a descendant of $u$ (tree edge), so by the nesting property
  of disc/fin times (Subtopic 2.2, disc/fin intervals of a descendant are contained within the
  ancestor's), $\text{fin}[v] < \text{fin}[u]$.
- If $\text{color}[v] = G$: $v$ is an active ancestor — this is a back edge, meaning a cycle exists,
  contradicting the assumption that $G$ is a DAG. This case cannot occur.
- If $\text{color}[v] = B$: $v$ has already finished, so $\text{fin}[v]$ was already set — since
  time only increases, $\text{fin}[v] < (\text{current time}) \le \text{fin}[u]$ (as $u$ hasn't
  finished yet). So $\text{fin}[v] < \text{fin}[u]$ regardless.

In every possible (non-contradictory) case, $\text{fin}[v] < \text{fin}[u]$. Hence sorting by
decreasing $\text{fin}$ places $u$ before $v$ for every edge $(u,v)$ — exactly the definition of a
valid topological order. $\blacksquare$

### Boundary transitions table

| Decision point | Condition | Action |
|---|---|---|
| DFS finishes vertex $u$ | $\text{color}[u] \to B$ | push $u$ onto a stack (or prepend to a list) — this naturally builds reverse-finish order |
| Reading off the answer | after all DFS calls complete | pop the stack top-to-bottom — gives decreasing-finish-time order directly |
| Graph has a cycle | a back edge is encountered during DFS | topological order does not exist — must detect this separately (see 7.3/7.4) before trusting the DFS output |

### The delta
The non-obvious insight: this algorithm's correctness **assumes** $G$ is acyclic — the proof above
explicitly relies on "the back-edge case cannot occur." If you run this DFS-based topo sort on a
graph that secretly has a cycle, it produces a *plausible-looking but meaningless* ordering with no
error signal, unless you separately layer in cycle detection (Subtopic 5.3's three-color scheme,
which this algorithm already computes as a byproduct — reusing the same GRAY check for free).


````c++
// Topological Sort (DFS): push to a stack on finish, read off in reverse-finish order
#include <iostream>
#include <vector>
#include <stack>
using namespace std;
#define vi vector<int>
#define vvi vector<vector<int>>

enum Color { WHITE, GRAY, BLACK };

// dfsTopo: standard DFS, pushing u onto the stack the instant it finishes (color -> BLACK).
void dfsTopo(int u, vvi& adj, vector<Color>& color, stack<int>& finishOrder) {
    color[u] = GRAY;
    for (int v : adj[u]) {
        if (color[v] == WHITE) dfsTopo(v, adj, color, finishOrder);
        // color[v] == GRAY would mean a cycle -- correctness of THIS function assumes acyclicity (see theory)
    }
    color[u] = BLACK;
    finishOrder.push(u);           // pushed on finish -- stack naturally yields reverse-finish order when popped
}

vi topoSortDFS(int V, vvi& adj) {
    vector<Color> color(V, WHITE);
    stack<int> finishOrder;
    for (int v = 0; v < V; v++) {
        if (color[v] == WHITE) dfsTopo(v, adj, color, finishOrder);
    }
    vi order;
    while (!finishOrder.empty()) { order.push_back(finishOrder.top()); finishOrder.pop(); }
    return order;
}

// isValidTopoOrder: verifies every edge (u,v) has u appearing before v in the given order.
bool isValidTopoOrder(int V, vvi& adj, vi& order) {
    vi position(V);
    for (int i = 0; i < V; i++) position[order[i]] = i;
    for (int u = 0; u < V; u++)
        for (int v : adj[u])
            if (position[u] >= position[v]) return false;
    return true;
}
````

````c++
// Test cell: input -> actual (Expected: X)

// Simple chain: 0->1->2
{
    vvi adj(3); adj[0]={1}; adj[1]={2}; adj[2]={};
    vi order = topoSortDFS(3, adj);
    cout << "topoSortDFS(chain) valid -> " << isValidTopoOrder(3, adj, order) << " (Expected: 1)\n";
}

// Diamond DAG: 0->1, 0->2, 1->3, 2->3
{
    vvi adj(4); adj[0]={1,2}; adj[1]={3}; adj[2]={3}; adj[3]={};
    vi order = topoSortDFS(4, adj);
    cout << "topoSortDFS(diamond) valid -> " << isValidTopoOrder(4, adj, order) << " (Expected: 1)\n";
}

// Single node, no edges
{
    vvi adj(1);
    vi order = topoSortDFS(1, adj);
    cout << "topoSortDFS(single node) size -> " << order.size() << " (Expected: 1)\n";
}

// Disconnected DAG: two independent chains
{
    vvi adj(4); adj[0]={1}; adj[1]={}; adj[2]={3}; adj[3]={};
    vi order = topoSortDFS(4, adj);
    cout << "topoSortDFS(disconnected DAG) valid -> " << isValidTopoOrder(4, adj, order) << " (Expected: 1)\n";
}

// Empty graph
{
    vvi adj(0);
    vi order = topoSortDFS(0, adj);
    cout << "topoSortDFS(empty) size -> " << order.size() << " (Expected: 0)\n";
}

// Wide fan-out DAG: 0 -> {1,2,3,4}, all independent afterward
{
    vvi adj(5); adj[0]={1,2,3,4};
    vi order = topoSortDFS(5, adj);
    cout << "topoSortDFS(fan-out) valid -> " << isValidTopoOrder(5, adj, order) << " (Expected: 1)\n";
}
````

## 7.2 Topological Sort (Kahn's Algorithm, BFS)

### State definition
$\text{indeg}[v]$ = number of incoming edges to $v$ still "unresolved" (from vertices not yet
placed in the output order). A vertex is **ready** iff $\text{indeg}[v] = 0$.

### Invariant
**Every popped node has all its dependencies popped before it.** Formally: at the moment $v$ is
dequeued, every $u$ with $(u,v) \in E$ has already been dequeued — because $v$ only ever becomes
eligible (enters the queue) once its $\text{indeg}$ reaches 0, which happens exactly when the last
such $u$ has been processed and decremented $v$'s count.

### Why it works — formal proof

**Claim:** the order in which vertices are dequeued is a valid topological order, and this
succeeds (dequeues all $V$ vertices) iff $G$ is acyclic.

*Proof (order validity):* by induction on dequeue order. Initially, only in-degree-0 vertices (no
dependencies) are eligible — trivially valid to place first. When $u$ is dequeued and processed,
every edge $(u,v)$ decrements $\text{indeg}[v]$; $v$ becomes eligible only when *all* its
in-edges' sources have been dequeued (that's what $\text{indeg}[v]=0$ means, by construction). So
by the time $v$ is dequeued, every predecessor required by an edge into $v$ has already appeared
earlier in the order — satisfying the topological order definition for every edge into $v$. Since
this holds for every $v$, the full sequence is a valid topological order.

*Proof (completion iff acyclic):* if $G$ is acyclic, every DAG has at least one in-degree-0 vertex
(else, following in-edges backward indefinitely without repeating a vertex would require infinite
vertices, or repeating a vertex implies a cycle — contradiction). Removing it (conceptually,
decrementing neighbors) leaves a smaller DAG, which again has an in-degree-0 vertex, by induction —
so the process continues until all $V$ vertices are removed/dequeued. Conversely, if a cycle
exists, every vertex on the cycle has an in-edge from another cycle vertex; no vertex on the cycle
can ever reach $\text{indeg}=0$ purely from vertices outside the cycle finishing, so the cycle's
vertices are never dequeued — the algorithm halts with fewer than $V$ vertices processed. This is
formalized as the cycle-detection criterion in 7.3.

### Boundary transitions table

| Decision point | Condition | Action |
|---|---|---|
| Initial in-degree-0 vertices | $\text{indeg}[v]=0$ before any processing | seed the queue with all of them (this is a form of multi-source BFS, Subtopic 4) |
| Edge $(u,v)$ processed | $u$ dequeued | decrement $\text{indeg}[v]$ |
| Decremented in-degree reaches 0 | $\text{indeg}[v]=0$ after decrement | $v$ becomes eligible, enqueue it |
| Queue empties, fewer than $V$ dequeued | remaining vertices' indegree never reached 0 | cycle exists — see 7.3 |

### The delta
The non-obvious insight: Kahn's algorithm is a **multi-source BFS variant** (Subtopic 4) where the
"sources" are dynamically discovered as in-degree drains to zero, rather than given upfront — the
core multi-source seeding idea (all currently-eligible vertices enter the frontier simultaneously)
is unchanged, but eligibility itself evolves as the algorithm runs, unlike 4.1/4.2 where the source
set was fixed from the start.


````c++
// Topological Sort (Kahn's Algorithm, BFS): indegree-based, eligibility-driven multi-source BFS
#include <iostream>
#include <vector>
#include <queue>
using namespace std;
#define vi vector<int>
#define vvi vector<vector<int>>

vi topoSortKahn(int V, vvi& adj) {
    vi indeg(V, 0);
    for (int u = 0; u < V; u++) for (int v : adj[u]) indeg[v]++;   // compute initial in-degrees

    queue<int> q;
    for (int v = 0; v < V; v++) if (indeg[v] == 0) q.push(v);       // seed: all initially-eligible vertices

    vi order;
    while (!q.empty()) {
        int u = q.front(); q.pop();
        order.push_back(u);
        for (int v : adj[u]) {
            indeg[v]--;                     // resolve this dependency
            if (indeg[v] == 0) q.push(v);    // newly eligible -- enters the frontier
        }
    }
    return order;    // if order.size() < V, a cycle exists (see 7.3)
}
````

````c++
// Test cell: input -> actual (Expected: X)

// Simple chain
{
    vvi adj(3); adj[0]={1}; adj[1]={2}; adj[2]={};
    vi order = topoSortKahn(3, adj);
    cout << "topoSortKahn(chain) size -> " << order.size() << " (Expected: 3)\n";
}

// Diamond DAG
{
    vvi adj(4); adj[0]={1,2}; adj[1]={3}; adj[2]={3}; adj[3]={};
    vi order = topoSortKahn(4, adj);
    cout << "topoSortKahn(diamond) size -> " << order.size() << " (Expected: 4)\n";
    // verify order[0] must be 0 (only initial in-degree-0 vertex)
    cout << "  first vertex -> " << order[0] << " (Expected: 0)\n";
}

// Single node
{
    vvi adj(1);
    vi order = topoSortKahn(1, adj);
    cout << "topoSortKahn(single node) size -> " << order.size() << " (Expected: 1)\n";
}

// Disconnected DAG: two independent chains
{
    vvi adj(4); adj[0]={1}; adj[1]={}; adj[2]={3}; adj[3]={};
    vi order = topoSortKahn(4, adj);
    cout << "topoSortKahn(disconnected) size -> " << order.size() << " (Expected: 4)\n";
}

// Graph WITH a cycle: Kahn's should stall before processing all vertices
{
    vvi adj(3); adj[0]={1}; adj[1]={2}; adj[2]={0};   // 3-cycle
    vi order = topoSortKahn(3, adj);
    cout << "topoSortKahn(3-cycle) size -> " << order.size() << " (Expected: 0, no vertex ever reaches indeg 0)\n";
}
````

## 7.3 Detect Cycle in Directed Graph (BFS-based, via Kahn's)

### State definition
Reuses Kahn's algorithm from 7.2 exactly, with the query changed from "produce the order" to
"does a valid full order exist."

### Invariant
Restated from 7.2's second proof: a vertex is dequeued **iff** it is not part of any cycle and all
its acyclic dependencies have resolved. Vertices on any cycle (or depending exclusively on a cycle)
never reach $\text{indeg}=0$.

### Why it works — the "fewer than V nodes" proof, stated as the primary claim here

**Claim:** $G$ has a directed cycle **iff** Kahn's algorithm dequeues strictly fewer than $V$
vertices.

*Proof:* ($\Rightarrow$) if a cycle $C$ exists, every vertex in $C$ has an in-edge from another
vertex of $C$. For a vertex $v \in C$ to ever reach $\text{indeg}[v]=0$, *every* in-edge into $v$
— including the one from within $C$ — must have its source already dequeued. But that source is
also in $C$ and faces the identical requirement, circularly. By finite descent (no vertex in $C$
can be first to satisfy this circular requirement without another cycle vertex already having been
dequeued, which is what we're trying to establish in the first place), **no vertex of $C$ is ever
dequeued** — so the total dequeued count is at most $V - |C| < V$.
($\Leftarrow$) contrapositive of 7.2's completion proof: if $G$ is acyclic, Kahn's dequeues all $V$
vertices (already proven). So dequeuing fewer than $V$ implies $G$ is *not* acyclic, i.e. has a
cycle. $\blacksquare$

### Boundary transitions table

| Decision point | Condition | Action |
|---|---|---|
| `order.size() == V` after Kahn's completes | all vertices dequeued | acyclic — no cycle |
| `order.size() < V` after queue empties | some vertices never reached indeg 0 | cycle exists among the undequeued vertices |
| Empty graph ($V=0$) | trivially `order.size() == V == 0` | acyclic (vacuously) |

### The delta
The non-obvious insight: this gives a **second, independent proof technique** for directed cycle
detection beyond Subtopic 5.3's DFS three-color scheme — same conclusion, entirely different
mechanism (in-degree exhaustion vs. recursion-stack membership). Both are $O(V+E)$; the choice
between them is usually about which one composes better with the rest of your solution (e.g.
Course Schedule II, 7.5, needs the topo order anyway, so Kahn's is the natural single-pass choice;
Alien Dictionary, 7.7, is typically done with Kahn's for the same reason).


````c++
// Detect Cycle in Directed Graph (BFS-based via Kahn's): reuse 7.2's algorithm, check output size
#include <iostream>
#include <vector>
#include <queue>
using namespace std;
#define vi vector<int>
#define vvi vector<vector<int>>

bool hasCycleKahn(int V, vvi& adj) {
    vi indeg(V, 0);
    for (int u = 0; u < V; u++) for (int v : adj[u]) indeg[v]++;

    queue<int> q;
    for (int v = 0; v < V; v++) if (indeg[v] == 0) q.push(v);

    int processed = 0;
    while (!q.empty()) {
        int u = q.front(); q.pop();
        processed++;
        for (int v : adj[u]) { indeg[v]--; if (indeg[v] == 0) q.push(v); }
    }
    return processed < V;   // fewer than V processed -> cycle exists (proof above)
}
````

````c++
// Test cell: input -> actual (Expected: X)

// 3-cycle
{
    vvi adj(3); adj[0]={1}; adj[1]={2}; adj[2]={0};
    cout << "hasCycleKahn(3-cycle) -> " << hasCycleKahn(3, adj) << " (Expected: 1)\n";
}

// Acyclic diamond
{
    vvi adj(4); adj[0]={1,2}; adj[1]={3}; adj[2]={3}; adj[3]={};
    cout << "hasCycleKahn(diamond) -> " << hasCycleKahn(4, adj) << " (Expected: 0)\n";
}

// Self-loop
{
    vvi adj(1); adj[0]={0};
    cout << "hasCycleKahn(self-loop) -> " << hasCycleKahn(1, adj) << " (Expected: 1)\n";
}

// Empty graph -- vacuously acyclic
{
    vvi adj(0);
    cout << "hasCycleKahn(empty) -> " << hasCycleKahn(0, adj) << " (Expected: 0)\n";
}

// Cycle plus an acyclic tail hanging off it: 0->1->2->0, 2->3
{
    vvi adj(4); adj[0]={1}; adj[1]={2}; adj[2]={0,3}; adj[3]={};
    cout << "hasCycleKahn(cycle+tail) -> " << hasCycleKahn(4, adj) << " (Expected: 1, vertex 3 alone can't rescue the cycle)\n";
}

// Cross-check against Subtopic 5.3's DFS-based directed cycle detector on the same graph
{
    vvi adj(3); adj[0]={1}; adj[1]={2}; adj[2]={0};
    cout << "hasCycleKahn(3-cycle) again -> " << hasCycleKahn(3, adj) << " (Expected: 1, matches Subtopic 5.3's DFS result)\n";
}
````

## 7.4 Course Schedule I (Cycle Detection Framed as Scheduling Feasibility)

### State definition
Given `numCourses` and a list of prerequisite pairs $[a, b]$ meaning "$b$ must be taken before
$a$" (i.e. a directed edge $b \to a$), determine whether it is possible to finish all courses.

### Invariant
This is **exactly** 7.3's cycle-detection problem wearing a word problem's clothing: "can all
courses be finished" $\iff$ "does the prerequisite graph have a valid topological order" $\iff$
"is the prerequisite graph acyclic" (by 7.2/7.3's proofs).

### Why it works
No new proof required — direct translation. A cycle among prerequisites (course $A$ requires $B$,
$B$ requires $C$, $C$ requires $A$) makes the entire cycle impossible to schedule, since every
course in the cycle transitively depends on itself. This is the real-world meaning of "directed
cycle" that motivated the abstract definition in the first place.

### Boundary transitions table

| Decision point | Condition | Action |
|---|---|---|
| Prerequisite pair $[a,b]$ | build edge $b \to a$ | note the direction: $b$ is a prerequisite FOR $a$, so $b$ must precede $a$ |
| Kahn's processed count | equals `numCourses` | feasible — return true |
| Kahn's processed count | less than `numCourses` | infeasible — a cycle exists, return false |

### The delta
The non-obvious insight: this problem exists in the curriculum specifically to train the pattern
recognition of **"feasibility of a dependency-ordered process" = "acyclicity of its dependency
graph"** — a huge class of real scheduling, build-system, and package-manager problems reduce to
exactly this, and the reduction is always this same one-line translation once the edge direction
is set up correctly.


````c++
// Course Schedule I: cycle detection framed as scheduling feasibility, reusing 7.3's Kahn's-based detector
#include <iostream>
#include <vector>
#include <queue>
using namespace std;
#define vi vector<int>
#define vvi vector<vector<int>>
#define pii pair<int,int>

bool canFinish(int numCourses, vector<pii>& prerequisites) {
    vvi adj(numCourses);
    for (auto& [a, b] : prerequisites) adj[b].push_back(a);   // b is a prerequisite FOR a: edge b -> a

    vi indeg(numCourses, 0);
    for (int u = 0; u < numCourses; u++) for (int v : adj[u]) indeg[v]++;

    queue<int> q;
    for (int v = 0; v < numCourses; v++) if (indeg[v] == 0) q.push(v);

    int processed = 0;
    while (!q.empty()) {
        int u = q.front(); q.pop();
        processed++;
        for (int v : adj[u]) { indeg[v]--; if (indeg[v] == 0) q.push(v); }
    }
    return processed == numCourses;   // all courses schedulable iff no cycle (7.3's proof)
}
````

````c++
// Test cell: input -> actual (Expected: X)

// Simple feasible case: course 1 needs course 0 first
{
    vector<pii> prereqs = {{1,0}};
    cout << "canFinish(1 needs 0) -> " << canFinish(2, prereqs) << " (Expected: 1)\n";
}

// Direct cycle: 0 needs 1, 1 needs 0
{
    vector<pii> prereqs = {{1,0},{0,1}};
    cout << "canFinish(direct cycle) -> " << canFinish(2, prereqs) << " (Expected: 0)\n";
}

// No prerequisites at all
{
    vector<pii> prereqs = {};
    cout << "canFinish(no prereqs) -> " << canFinish(3, prereqs) << " (Expected: 1)\n";
}

// Longer cycle: 0->1->2->0 (via prerequisite semantics)
{
    vector<pii> prereqs = {{1,0},{2,1},{0,2}};
    cout << "canFinish(3-cycle) -> " << canFinish(3, prereqs) << " (Expected: 0)\n";
}

// Diamond dependency, feasible
{
    vector<pii> prereqs = {{1,0},{2,0},{3,1},{3,2}};
    cout << "canFinish(diamond deps) -> " << canFinish(4, prereqs) << " (Expected: 1)\n";
}
````

## 7.5 Course Schedule II (Topo Order as the Schedule)

### State definition
Same setup as 7.4, but instead of a boolean, return **a valid course order** if one exists, or an
empty array if not.

### Invariant
Directly Kahn's algorithm's output order (7.2) — no new invariant. The feasibility check (7.4) is
now a **byproduct** of building the order: if the order's length equals `numCourses`, it's the
answer; otherwise return empty.

### Why it works
Direct application of 7.2's correctness proof: the dequeue order Kahn's produces already satisfies
"every prerequisite appears before the course that needs it," which is exactly the schedule being
asked for. No separate algorithm is needed beyond what 7.2/7.4 already compute.

### Boundary transitions table

| Decision point | Condition | Action |
|---|---|---|
| `order.size() == numCourses` | Kahn's fully drained | return `order` as the valid schedule |
| `order.size() < numCourses` | cycle detected mid-way | return empty array — no valid schedule exists |
| Multiple valid orders exist | more than one vertex eligible at some point | any one produced by Kahn's queue order is acceptable — the problem doesn't require a unique answer |

### The delta
The non-obvious insight: 7.4 and 7.5 are **the same algorithm** — the only difference is whether
you discard the order and return a boolean, or keep and return it. This is worth internalizing
because it means you never need to "choose" between cycle detection and order construction
separately; building the order via Kahn's gives you both for the same $O(V+E)$ cost.


````c++
// Course Schedule II: return the actual schedule (or empty if infeasible), reusing Kahn's directly
#include <iostream>
#include <vector>
#include <queue>
using namespace std;
#define vi vector<int>
#define vvi vector<vector<int>>
#define pii pair<int,int>

vi findOrder(int numCourses, vector<pii>& prerequisites) {
    vvi adj(numCourses);
    for (auto& [a, b] : prerequisites) adj[b].push_back(a);

    vi indeg(numCourses, 0);
    for (int u = 0; u < numCourses; u++) for (int v : adj[u]) indeg[v]++;

    queue<int> q;
    for (int v = 0; v < numCourses; v++) if (indeg[v] == 0) q.push(v);

    vi order;
    while (!q.empty()) {
        int u = q.front(); q.pop();
        order.push_back(u);
        for (int v : adj[u]) { indeg[v]--; if (indeg[v] == 0) q.push(v); }
    }
    if ((int)order.size() != numCourses) return {};   // infeasible -- cycle exists (7.4's exact check)
    return order;
}
````

````c++
// Test cell: input -> actual (Expected: X)

// Simple feasible case
{
    vector<pii> prereqs = {{1,0}};
    vi order = findOrder(2, prereqs);
    cout << "findOrder(1 needs 0) size -> " << order.size() << " (Expected: 2)\n";
    cout << "  order[0] -> " << (order.empty() ? -1 : order[0]) << " (Expected: 0, the prerequisite)\n";
}

// Cycle -> empty result
{
    vector<pii> prereqs = {{1,0},{0,1}};
    vi order = findOrder(2, prereqs);
    cout << "findOrder(cycle) size -> " << order.size() << " (Expected: 0)\n";
}

// No prerequisites
{
    vector<pii> prereqs = {};
    vi order = findOrder(3, prereqs);
    cout << "findOrder(no prereqs) size -> " << order.size() << " (Expected: 3)\n";
}

// Single course, no prerequisites
{
    vector<pii> prereqs = {};
    vi order = findOrder(1, prereqs);
    cout << "findOrder(single course) size -> " << order.size() << " (Expected: 1)\n";
}

// Diamond dependency: verify 3 (needing both 1 and 2) comes after both
{
    vector<pii> prereqs = {{1,0},{2,0},{3,1},{3,2}};
    vi order = findOrder(4, prereqs);
    int pos3=-1, pos1=-1, pos2=-1;
    for (int i=0;i<(int)order.size();i++){ if(order[i]==3)pos3=i; if(order[i]==1)pos1=i; if(order[i]==2)pos2=i; }
    cout << "findOrder(diamond) 3 after 1 and 2 -> " << (pos3>pos1 && pos3>pos2) << " (Expected: 1)\n";
}
````

## 7.6 Find Eventual Safe States (Reverse Graph + Topo Sort Trick)

### State definition
A node is **terminal** if it has no outgoing edges. A node is **safe** if every possible path
starting from it eventually leads to a terminal node (equivalently: the node is not on a cycle,
and cannot reach any cycle). Return all safe nodes in sorted order.

### Invariant — the reversal trick
A node is unsafe iff it lies on a cycle, or has a path to a node that lies on a cycle. Directly
detecting "can reach a cycle" via forward traversal from every node would cost $O(V \cdot (V+E))$
naively. Instead: **reverse every edge**, then a node is safe in the original graph iff, in the
reversed graph, it is reachable via a valid "peeling" process identical to Kahn's — a node becomes
safe once **all** of its (originally outgoing, now incoming after reversal) edges have been
confirmed safe.

$$
G^R = (V, E^R), \quad E^R = \{(v,u) : (u,v) \in E\}
$$

### Why it works — proof that reversed-graph Kahn's peeling identifies exactly the safe set

**Claim:** running Kahn's algorithm on $G^R$, seeded from nodes with **out-degree 0 in the
original graph** (equivalently in-degree 0 in $G^R$... careful, see below), and peeling in the same
in-degree-exhaustion manner, marks exactly the safe nodes.

*Precise construction:* build $G^R$. A node $u$'s "in-degree" in $G^R$ equals $u$'s **out-degree**
in $G$. Terminal nodes in $G$ (out-degree 0) have in-degree 0 in $G^R$ — these are unconditionally
safe (no outgoing edges at all, so trivially every path from them is finite and terminal) and seed
Kahn's queue. When a node $u$ is processed (peeled) in $G^R$'s Kahn's run, every node $w$ with an
edge $w \to u$ in $G^R$ (i.e. $u \to w$ in $G$) has one of its outgoing-edge "obligations"
resolved: **this is only valid to mark $w$ safe if $u$ being safe accounts for one of $w$'s
outgoing edges, and $w$ becomes safe only once ALL of its outgoing edges point to already-confirmed-safe
nodes** — exactly Kahn's in-degree-exhaustion condition, applied on $G^R$. A node never peeled is
one that, in $G$, has some outgoing path that never bottoms out at a terminal node without passing
through a cycle — precisely an unsafe node, by 7.2/7.3's proof that un-peeled vertices in Kahn's
are exactly those "stuck" behind an unresolved cyclic dependency.

### Boundary transitions table

| Decision point | Condition | Action |
|---|---|---|
| Node has out-degree 0 in $G$ | terminal | trivially safe — seed peeling (in-degree 0 in $G^R$) |
| Node's outgoing edges (in $G$) all point to confirmed-safe nodes | all corresponding $G^R$ in-edges resolved | node becomes safe — peel it |
| Node lies on a cycle, or reaches one | some outgoing edge never resolves | never peeled — unsafe |

### The delta
The non-obvious insight: this is the third distinct-looking application of Kahn's
in-degree-exhaustion invariant in this subtopic (after plain topo sort and Course Schedule), and
the pattern to recognize is: **whenever "safety"/"validity" of a node depends on ALL of its
forward-reachable structure eventually terminating cleanly, reverse the graph and let Kahn's
peeling propagate "confirmed-good" status backward from the base case (terminal nodes) — this
converts an expensive forward reachability-to-a-cycle check into a single linear peeling pass.**


````c++
// Find Eventual Safe States: reverse the graph, then Kahn's-style peeling from terminal nodes
#include <iostream>
#include <vector>
#include <queue>
#include <algorithm>
using namespace std;
#define vi vector<int>
#define vvi vector<vector<int>>

vi eventualSafeNodes(int V, vvi& adj) {
    vvi radj(V);           // reversed graph
    vi outdeg(V, 0);       // outdeg[u] in ORIGINAL graph = "obligations" u must resolve before being safe
    for (int u = 0; u < V; u++) {
        for (int v : adj[u]) {
            radj[v].push_back(u);   // reverse the edge: v -> u in radj
            outdeg[u]++;
        }
    }

    queue<int> q;
    vi safe(V, 0);
    for (int v = 0; v < V; v++) if (outdeg[v] == 0) q.push(v);   // terminal nodes: trivially safe, seed peeling

    while (!q.empty()) {
        int u = q.front(); q.pop();
        safe[u] = 1;
        for (int w : radj[u]) {          // w had an original edge w -> u; that obligation is now resolved
            outdeg[w]--;
            if (outdeg[w] == 0) q.push(w);   // ALL of w's outgoing edges now point to confirmed-safe nodes
        }
    }

    vi result;
    for (int v = 0; v < V; v++) if (safe[v]) result.push_back(v);
    sort(result.begin(), result.end());
    return result;
}
````

````c++
// Test cell: input -> actual (Expected: X)

// Classic case: 0->{1,2}, 1->{2,3}, 2->{5}, 3->{0,5}, 4->{5}, 5->{}, 6->{}
{
    vvi adj(7);
    adj[0]={1,2}; adj[1]={2,3}; adj[2]={5}; adj[3]={0,5}; adj[4]={5}; adj[5]={}; adj[6]={};
    vi safe = eventualSafeNodes(7, adj);
    cout << "eventualSafeNodes(classic) -> ";
    for (int v : safe) cout << v << " ";
    cout << "(Expected: 2 4 5 6)\n";
}

// Pure cycle, no terminal nodes at all -> nothing safe
{
    vvi adj(3); adj[0]={1}; adj[1]={2}; adj[2]={0};
    vi safe = eventualSafeNodes(3, adj);
    cout << "eventualSafeNodes(pure cycle) size -> " << safe.size() << " (Expected: 0)\n";
}

// All terminal nodes (no outgoing edges anywhere) -> everything safe
{
    vvi adj(3);
    vi safe = eventualSafeNodes(3, adj);
    cout << "eventualSafeNodes(all terminal) size -> " << safe.size() << " (Expected: 3)\n";
}

// Single node, no edges
{
    vvi adj(1);
    vi safe = eventualSafeNodes(1, adj);
    cout << "eventualSafeNodes(single node) size -> " << safe.size() << " (Expected: 1)\n";
}

// Self-loop: node points to itself -> never safe
{
    vvi adj(2); adj[0]={0}; adj[1]={};
    vi safe = eventualSafeNodes(2, adj);
    cout << "eventualSafeNodes(self-loop) -> ";
    for (int v : safe) cout << v << " ";
    cout << "(Expected: 1, only the terminal node 1)\n";
}
````

## 7.7 Alien Dictionary

### State definition
Given a list of words sorted according to some unknown alien alphabet's ordering, reconstruct a
valid ordering of the alphabet's letters (any valid one, if multiple exist; empty/invalid signal if
none exists).

### Building the precedence graph
For every **adjacent pair** of words $(w_1, w_2)$ in the given list, find the **first position**
at which they differ; if $w_1[i] \neq w_2[i]$, this proves $w_1[i]$ precedes $w_2[i]$ in the alien
alphabet — add a directed edge $w_1[i] \to w_2[i]$. This is exactly Subtopic 1.1's implicit-graph
framing again: the "nodes" are the 26 (or however many) distinct letters, and each adjacent word
pair contributes at most one edge to the precedence graph.

### Why it works
Once the precedence graph is built, "reconstruct a valid alphabet ordering" is *exactly*
topological sort (7.1/7.2) on the letter-precedence graph — any topological order satisfies every
pairwise precedence constraint derived from the word list, which is precisely what "consistent with
the given sorted word list" means.

### Deep dive — edge cases that make this problem genuinely subtle

**Edge case 1 — only the first differing character matters.** Given `w1, w2`, you must stop
comparing at the *first* index where they differ and extract exactly one edge from that index —
continuing to compare further characters would be meaningless (the alien alphabet ordering says
nothing about characters after the first difference has already settled which word is "smaller").

**Edge case 2 — invalid prefix ordering.** If $w_1$ is a proper prefix of $w_2$ (e.g. `"abc"` before
`"ab"`), the *only* valid orderings require the shorter word to come first if all shared characters
are equal — `"abc"` before `"ab"` is **impossible** in any consistent alphabet, since a valid
sort always places prefixes before their extensions when their shared characters compare equal
(this is a length-based tiebreak in the underlying lexicographic definition, not something any
letter-ordering can express). This must be detected as an **immediate contradiction** — return
empty/invalid — since no topological sort can fix an inconsistency that isn't expressible via edges
at all.

**Edge case 3 — cycle in the derived precedence graph.** Even if every individual word pair looks
fine, the combined set of edges across all pairs can still form a cycle (e.g. pair 1 implies
`a` before `b`, pair 2 implies `b` before `c`, pair 3 implies `c` before `a`) — this must be caught
by the topo-sort's own cycle-detection machinery (7.2/7.3's "fewer than $V$ processed" check),
exactly as in Course Schedule.

**Edge case 4 — letters that never appear in any adjacent-pair comparison.** A letter that appears
in the word list but is never directly compared against another (e.g. it only ever appears in
identical relative positions) still needs to be included as an isolated vertex with in-degree 0 in
the final graph — otherwise the output alphabet is missing letters that should appear (in *some*
valid position, since no constraint pins them down).

### Boundary transitions table

| Decision point | Condition | Action |
|---|---|---|
| First differing character found between adjacent words | $w_1[i] \neq w_2[i]$ | add edge $w_1[i] \to w_2[i]$, stop comparing this pair |
| No differing character found, $w_1$ is a proper prefix of $w_2$ | `len(w1) <= len(w2)`, all shared chars equal | valid, no edge needed — move to next pair |
| No differing character found, $w_1$ is LONGER and $w_2$ is its prefix | `len(w1) > len(w2)`, all shared chars equal | **invalid input** — return empty immediately, no topo sort can rescue this |
| Kahn's processed count < total distinct letters | cycle exists in derived precedence graph | invalid ordering — return empty |
| Letter never involved in any edge | isolated vertex | still include with in-degree 0, seeded normally in Kahn's |

### The delta
The non-obvious insight: Alien Dictionary is **not** a new graph algorithm — it is 7.1/7.2's
topological sort, preceded by a genuinely fiddly graph-*construction* step (Subtopic 1.1's "define
$\delta$/edges correctly" lesson, sharpened) where getting the edge-extraction and the
prefix-invalidity check wrong silently produces a plausible-but-wrong graph that topo sort will
happily (and incorrectly) process without any algorithmic error — the bugs here live entirely in
graph construction, not in the topological sort itself.


````c++
// Alien Dictionary: build precedence graph from adjacent word pairs, then Kahn's topo sort
#include <iostream>
#include <vector>
#include <string>
#include <queue>
#include <set>
using namespace std;
#define vi vector<int>
#define vvi vector<vector<int>>

string alienOrder(vector<string>& words) {
    set<char> letterSet;
    for (string& w : words) for (char c : w) letterSet.insert(c);

    vector<char> letters(letterSet.begin(), letterSet.end());
    int n = letters.size();
    vector<vector<int>> childIdx(26);   // adjacency by letter index (0-25 for 'a'-'z')
    vi indeg(26, 0);
    vector<vector<bool>> edgeExists(26, vector<bool>(26, false));  // avoid duplicate edges inflating indeg

    // Build the precedence graph from adjacent word pairs -- this is where the real subtlety lives.
    for (int i = 0; i + 1 < (int)words.size(); i++) {
        string& w1 = words[i];
        string& w2 = words[i+1];
        int minLen = min(w1.size(), w2.size());
        bool foundDiff = false;
        for (int j = 0; j < minLen; j++) {
            if (w1[j] != w2[j]) {
                int a = w1[j] - 'a', b = w2[j] - 'a';
                if (!edgeExists[a][b]) {              // edge case: only count each edge once
                    edgeExists[a][b] = true;
                    childIdx[a].push_back(b);
                    indeg[b]++;
                }
                foundDiff = true;
                break;                                  // edge case 1: stop at the FIRST differing character
            }
        }
        // edge case 2: w1 longer than w2 but w2 is a true prefix of w1 -> invalid, no valid alphabet exists
        if (!foundDiff && w1.size() > w2.size()) return "";
    }

    queue<int> q;
    for (char c : letters) if (indeg[c - 'a'] == 0) q.push(c - 'a');   // edge case 4: isolated letters seed normally

    string result;
    int processed = 0;
    while (!q.empty()) {
        int u = q.front(); q.pop();
        result += char('a' + u);
        processed++;
        for (int v : childIdx[u]) { indeg[v]--; if (indeg[v] == 0) q.push(v); }
    }

    if (processed != n) return "";   // edge case 3: cycle in the derived graph -> invalid ordering
    return result;
}
````

````c++
// Test cell: input -> actual (Expected: X)

// Standard case: wrt, wrf, er, ett, rftt -> implies w<e, r<t, e<r, t<f
{
    vector<string> words = {"wrt","wrf","er","ett","rftt"};
    string order = alienOrder(words);
    cout << "alienOrder(standard) -> " << order << " (Expected: some valid order like wertf, length 5)\n";
    cout << "  length -> " << order.size() << " (Expected: 5)\n";
}

// Invalid prefix ordering: "abc" before "ab" is impossible
{
    vector<string> words = {"abc","ab"};
    string order = alienOrder(words);
    cout << "alienOrder(invalid prefix) -> \"" << order << "\" (Expected: empty string)\n";
}

// Valid prefix ordering: "ab" before "abc" is fine, no edge extracted
{
    vector<string> words = {"ab","abc"};
    string order = alienOrder(words);
    cout << "alienOrder(valid prefix) -> " << order << " (Expected: non-empty, contains a and b)\n";
}

// Cycle in derived graph: a<b, b<c, c<a
{
    vector<string> cyc = {"ab","bc","ca"};   // pairwise constraints: a<b, b<c, c<a -- forms a cycle
    string order = alienOrder(cyc);
    cout << "alienOrder(cyclic constraints) -> \"" << order << "\" (Expected: empty string, cycle a->b->c->a)\n";
}

// Single word: no adjacent pairs at all, every letter isolated
{
    vector<string> words = {"abc"};
    string order = alienOrder(words);
    cout << "alienOrder(single word) length -> " << order.size() << " (Expected: 3, all letters present, any order valid)\n";
}
````

````c++
// ================= UNIFIED MENTAL MODEL — Subtopic 7 =================
#include <iostream>
#include <vector>
#include <stack>
#include <queue>
using namespace std;
#define vi vector<int>
#define vvi vector<vector<int>>

enum Color { WHITE, GRAY, BLACK };

// --- Template: DFS-based topo sort (7.1 pattern) ---
void dfsTopoTemplate(int u, vvi& adj, vector<Color>& color, stack<int>& out) {
    color[u] = GRAY;
    for (int v : adj[u]) if (color[v] == WHITE) dfsTopoTemplate(v, adj, color, out);
    color[u] = BLACK;
    out.push(u);
}

// --- Template: Kahn's BFS topo sort + cycle detection (7.2/7.3 pattern) ---
// Returns {order, isValidTopoOrder}. Reused directly for Course Schedule I/II (7.4/7.5).
pair<vi,bool> kahnTemplate(int V, vvi& adj) {
    vi indeg(V, 0);
    for (int u = 0; u < V; u++) for (int v : adj[u]) indeg[v]++;
    queue<int> q;
    for (int v = 0; v < V; v++) if (indeg[v] == 0) q.push(v);
    vi order;
    while (!q.empty()) {
        int u = q.front(); q.pop();
        order.push_back(u);
        for (int v : adj[u]) if (--indeg[v] == 0) q.push(v);
    }
    return {order, (int)order.size() == V};
}

// --- Template: reverse-graph safety peeling (7.6 pattern) ---
vi reverseGraphPeelTemplate(int V, vvi& adj) {
    vvi radj(V); vi outdeg(V, 0);
    for (int u = 0; u < V; u++) for (int v : adj[u]) { radj[v].push_back(u); outdeg[u]++; }
    queue<int> q;
    for (int v = 0; v < V; v++) if (outdeg[v] == 0) q.push(v);
    vi safe(V, 0);
    while (!q.empty()) {
        int u = q.front(); q.pop(); safe[u] = 1;
        for (int w : radj[u]) if (--outdeg[w] == 0) q.push(w);
    }
    return safe;
}

int main() {
    vvi adj(4); adj[0]={1,2}; adj[1]={3}; adj[2]={3};
    auto [order, valid] = kahnTemplate(4, adj);
    cout << "Unified template sanity check: valid topo order -> " << valid << " size -> " << order.size() << " (Expected: 1, 4)\n";
    return 0;
}
````

### Decision Tree — Subtopic 7

```
Does the problem require a linear ordering respecting directed precedence constraints?
│
├── NO — just need to know IF a valid ordering exists (feasibility only)
│    → Kahn's algorithm, check processed count == V (7.3/7.4). DFS 3-color (Subtopic 5.3) also works.
│
└── YES — need the actual ordering
     │
     Do you prefer building it via finish-time recursion, or via indegree-draining iteration?
     ├── DFS finish-time (7.1) — simpler if you already have DFS machinery in place, reuses
     │                            Subtopic 2.2/5.3's disc/fin and color[] infrastructure
     └── Kahn's BFS (7.2) — more natural when the problem is ALREADY phrased in terms of
                             "dependencies must resolve before X becomes eligible" (scheduling-flavored
                             problems: Course Schedule, Alien Dictionary) -- it's a multi-source BFS
                             (Subtopic 4) where the source set grows dynamically


Does the problem ask "which nodes are safe / valid / non-cyclic-reaching" rather than a full order?
→ Reverse the graph, peel from terminal (originally out-degree-0) nodes using Kahn's exhaustion
  logic (7.6) — converts an expensive forward "can I reach a cycle" check into one linear pass.


Does the problem give you constraints IMPLICITLY (e.g. sorted word list) rather than an explicit
edge list?
→ First build the precedence graph correctly (watch for: only-first-difference edges, prefix
  invalidity, isolated vertices -- 7.7's deep dive), THEN apply topo sort/Kahn's exactly as above.
  Bugs in these problems are almost always in graph CONSTRUCTION, not in the topo sort itself.
```


### Complexity Summary — Subtopic 7

| Algorithm | Time | Space | When to use | Key invariant | Failure mode if misused |
|---|---|---|---|---|---|
| Topological Sort (DFS) | $O(V+E)$ | $O(V)$ recursion + stack | Have DFS infrastructure already; order via reverse-finish-time | every edge satisfies $\text{fin}[v] < \text{fin}[u]$ (DAG only) | running on a graph with an undetected cycle produces a meaningless but plausible-looking order |
| Topological Sort (Kahn's, BFS) | $O(V+E)$ | $O(V)$ | Scheduling-flavored problems; need cycle detection as a byproduct | every popped node has all dependencies popped first | forgetting to check `order.size() == V` silently accepts cyclic input as if it were valid |
| Cycle detection via Kahn's | $O(V+E)$ | $O(V)$ | Directed graphs, need cycle existence alongside/instead of full order | processed count $< V$ $\iff$ cycle exists | none specific — equivalent alternative to Subtopic 5.3's DFS three-color method |
| Course Schedule I/II | $O(V+E)$ | $O(V+E)$ | Direct application of Kahn's to a "prerequisite" word problem | feasibility = acyclicity, exactly as in general topo sort | getting the edge direction backwards (prerequisite vs. dependent) silently inverts the whole schedule |
| Find Eventual Safe States | $O(V+E)$ | $O(V+E)$ (reversed graph) | "Which nodes provably terminate cleanly" queries | reverse-graph Kahn's peeling propagates "confirmed safe" backward from terminal nodes | forgetting to reverse the graph turns this into a different (incorrect) computation entirely |
| Alien Dictionary | $O(\sum \|word\|)$ graph construction $+\ O(V+E)$ topo sort | $O(26^2)$ or $O(\text{alphabet size}^2)$ for edge-dedup + $O(V+E)$ | Reconstructing implicit ordering constraints from sorted data | topo sort correctness is unchanged; all subtlety is in edge EXTRACTION | invalid-prefix case (longer word is a prefix-extension of a later, shorter word) must be caught explicitly — topo sort alone won't catch it, since no edge encodes it |
