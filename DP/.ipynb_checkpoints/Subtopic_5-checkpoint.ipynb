# Dynamic Programming — Subtopic 5

## DP on Strings

**Problems covered:**
1. Longest Common Subsequence (LCS)
2. Print LCS
3. Longest Common Substring
4. Longest Palindromic Subsequence (LPS)
5. Minimum Insertions to Make Palindrome
6. Minimum Insertions/Deletions to Convert A to B
7. Shortest Common Supersequence (SCS)
8. Edit Distance
9. Wildcard Matching
10. Regex Matching

> All ten problems share a 2D state $(i, j)$ over **prefix lengths** of two strings (or one string + pattern). The recurrence universally case-splits on **the last character of each prefix**: match or mismatch. The whole subtopic is the disciplined application of this case split, with one variation per problem in how the mismatch case decomposes.

---

# The state — `(i, j) = prefix lengths`, and the universal case split

All ten problems share this state convention:

$$\text{dp}[i][j] = \text{(answer for the subproblem on)}\ \ s[0..i-1]\ \text{and}\ t[0..j-1]$$

i.e., $\text{dp}[i][j]$ is the answer when we consider the first $i$ characters of $s$ and the first $j$ characters of $t$. **Critically**: $i$ and $j$ are *prefix lengths*, not indices. So $\text{dp}[0][\cdot]$ and $\text{dp}[\cdot][0]$ correspond to **empty prefixes** — and that's where the base cases live.

**Why prefix-length indexing rather than 0-indexed positions.** With prefix lengths:
- The DP table has dimensions $(n+1) \times (m+1)$, not $n \times m$.
- The base row/column at $i = 0$ or $j = 0$ corresponds to one prefix being empty — a natural "nothing matched yet" base.
- The recurrence reads $s[i-1]$ and $t[j-1]$ — the **last** characters of the two prefixes.
- Negative indices never appear in the recursion.

This convention is so standard that you should adopt it as muscle memory.

---

## The match / mismatch case split

For every problem in this subtopic, the recurrence asks: **does the last character of each prefix participate in the optimal solution?**

**Match case** ($s[i-1] = t[j-1]$): both last characters align. The subproblem reduces to the same question on the shorter prefixes $(i-1, j-1)$, often with a "+ 1" or "+ char" contribution.

**Mismatch case** ($s[i-1] \neq t[j-1]$): at least one of the two last characters is *not* used (or is consumed by an operation like insert/delete/replace). This branches into several sub-options, combined by $\min$, $\max$, or $\vee$ depending on flavor.

The whole subtopic is variations on this skeleton:

$$\text{dp}[i][j] = \begin{cases} \text{transform}_{\text{match}}(\text{dp}[i-1][j-1]) & s[i-1] = t[j-1] \\ \text{combine}\big(\text{branches involving smaller subproblems}\big) & s[i-1] \neq t[j-1] \end{cases}$$

What changes per problem:
1. **The transform on match** — `+1`, append-char, or pure inheritance.
2. **Which branches appear on mismatch** — usually two (drop from $s$, drop from $t$), occasionally three (Edit Distance: insert/delete/replace), occasionally none (Common Substring: mismatch hard-resets to 0).
3. **The combine operator** — $\max$ for LCS, $\min$ for Edit Distance, OR for Pattern Matching, etc.

---

## Direction of traversal & space optimization

Every problem here reads $\text{dp}[i-1][\cdot]$ and/or $\text{dp}[i][j-1]$, $\text{dp}[i-1][j-1]$. All reads are "above and to the left" (in standard tabulation orientation), so the iteration order is **row-major: $i$ from $1$ to $n$, $j$ from $1$ to $m$**.

For space optimization to $O(\min(n, m))$ we typically use **two rows** (`prev`, `cur`) and swap. A one-row in-place optimization is sometimes possible but needs care because three cells in the previous row interact (above-left, above, left).

The two-row pattern is robust and what we'll use throughout — clarity over micro-optimization.

---

# Family A — LCS and its derivatives

LCS is the kernel DP of this subtopic. Eight of the ten problems either *are* LCS, or *reduce* to LCS via a closed-form formula, or share its recurrence shape with one tweak.

## 5.1 Longest Common Subsequence (LCS)

**Problem.** Given two strings $s$ (length $n$) and $t$ (length $m$), find the length of the longest sequence appearing as a subsequence in both. (A *subsequence* of a string is obtained by deleting zero or more characters without rearranging the rest.)

---

### Theory

**State definition.**
$\text{dp}[i][j]$ = length of the longest common subsequence of $s[0..i-1]$ and $t[0..j-1]$ (the prefixes of length $i$ and $j$).

**Invariant.** For every $(i, j)$, $\text{dp}[i][j]$ correctly reports the LCS length over the chosen prefixes. The recursion case-splits on the **last character of each prefix**:
- If $s[i-1] = t[j-1]$, this matching character can extend the LCS of the previous prefixes.
- Otherwise, at least one of $s[i-1]$ or $t[j-1]$ is **not** in the LCS — drop one and recurse.

**Recurrence.**
$$\text{dp}[i][j] = \begin{cases} 1 + \text{dp}[i-1][j-1] & \text{if}\ s[i-1] = t[j-1] \\ \max\!\big(\text{dp}[i-1][j],\ \text{dp}[i][j-1]\big) & \text{if}\ s[i-1] \neq t[j-1] \end{cases}$$

**Base cases.**
- $\text{dp}[0][j] = 0$ for all $j$ — empty $s$ prefix means no common subsequence.
- $\text{dp}[i][0] = 0$ for all $i$ — symmetric.

**Boundary transitions table.**

| Case | Recurrence value |
|---|---|
| $i = 0$ or $j = 0$ | $0$ |
| $s[i-1] = t[j-1]$ | $1 + \text{dp}[i-1][j-1]$ |
| $s[i-1] \neq t[j-1]$ | $\max(\text{dp}[i-1][j],\ \text{dp}[i][j-1])$ |

**Why it works.**
- *Optimal substructure:* Let $L = \text{LCS}(s[0..i-1], t[0..j-1])$. If $s[i-1] = t[j-1]$, the last char of $L$ either *is* this matching char (and the rest is $\text{LCS}(s[0..i-2], t[0..j-2])$) or *isn't*; in the latter case $L \subseteq s[0..i-2]$, so $L = \text{LCS}(s[0..i-2], t[0..j-1])$, and similarly $L = \text{LCS}(s[0..i-1], t[0..j-2])$. **A simple inequality argument shows the matching-char option dominates** when $s[i-1] = t[j-1]$, justifying the unconditional `1 + dp[i-1][j-1]`. If $s[i-1] \neq t[j-1]$, the last char of $L$ cannot be both characters, so at least one is unused; take the better of the two options.
- *Overlapping subproblems:* Naive recursion is $\Theta(2^{n+m})$; state space is $\Theta(nm)$.

**Complexity.** Time $O(nm)$. Space $O(nm)$ → $O(\min(n, m))$ with two rows.

**Space optimization — validity proof.**
$\text{dp}[i][j]$ reads $\text{dp}[i-1][j-1]$, $\text{dp}[i-1][j]$, $\text{dp}[i][j-1]$. The first two are in the previous row; the third is in the current row at $j-1$ (already computed when we reach $j$). Keep `prev[]` (row $i-1$) and `cur[]` (row $i$), iterate $j$ ascending, and swap after each row.

> Note: the *length* of the LCS can be computed with $O(m)$ space, but **reconstructing** the LCS string (next problem) requires the full $O(nm)$ table.

**Delta.** This is the kernel. Everything else in Family A is "LCS plus a twist."

---

````c++
// LCS — three implementations
#include <vector>
#include <string>
#include <algorithm>
#include <iostream>
using namespace std;

// (A) Top-down memoization
//     dp[i][j] = LCS length of s[0..i-1] and t[0..j-1].
int lcs_memo_helper(int i, int j, const string& s, const string& t, vector<vector<int>>& dp) {
    if (i == 0 || j == 0) return 0;                        // base: empty prefix
    if (dp[i][j] != -1) return dp[i][j];
    if (s[i-1] == t[j-1])                                  // match: both last chars align
        dp[i][j] = 1 + lcs_memo_helper(i-1, j-1, s, t, dp);
    else                                                   // mismatch: drop one last char
        dp[i][j] = max(lcs_memo_helper(i-1, j, s, t, dp),
                       lcs_memo_helper(i, j-1, s, t, dp));
    return dp[i][j];
}
int lcs_memo(const string& s, const string& t) {
    int n = (int)s.size(), m = (int)t.size();
    vector<vector<int>> dp(n + 1, vector<int>(m + 1, -1));
    return lcs_memo_helper(n, m, s, t, dp);
}

// (B) Tabulation — O(nm) time, O(nm) space
int lcs_tab(const string& s, const string& t) {
    int n = (int)s.size(), m = (int)t.size();
    vector<vector<int>> dp(n + 1, vector<int>(m + 1, 0));  // row/col 0 are zero (base)
    for (int i = 1; i <= n; ++i) {
        for (int j = 1; j <= m; ++j) {
            if (s[i-1] == t[j-1])                          // match
                dp[i][j] = 1 + dp[i-1][j-1];
            else                                           // mismatch
                dp[i][j] = max(dp[i-1][j], dp[i][j-1]);
        }
    }
    return dp[n][m];
}

// (C) Space-optimized — O(min(n,m)) using two rows
//     Valid: dp[i][j] reads {dp[i-1][j-1], dp[i-1][j], dp[i][j-1]} — all in prev row or cur row prefix.
int lcs_opt(const string& s, const string& t) {
    int n = (int)s.size(), m = (int)t.size();
    // Ensure t is the shorter string so the row length is min(n,m)+1
    if (m > n) return lcs_opt(t, s);
    vector<int> prev(m + 1, 0), cur(m + 1, 0);             // rows i-1 and i
    for (int i = 1; i <= n; ++i) {
        for (int j = 1; j <= m; ++j) {
            if (s[i-1] == t[j-1])
                cur[j] = 1 + prev[j-1];                    // match reads NW
            else
                cur[j] = max(prev[j], cur[j-1]);           // mismatch reads N and W
        }
        swap(prev, cur);                                   // promote cur -> prev for next row
    }
    return prev[m];                                        // after final swap, prev holds row n
}
````

````c++
// Tests — LCS
auto run_lcs = [](string s, string t, int expected) {
    int a = lcs_memo(s, t);
    int b = lcs_tab(s, t);
    int c = lcs_opt(s, t);
    bool ok = (a == expected && b == expected && c == expected);
    cout << "s=\"" << s << "\" t=\"" << t << "\""
         << " -> memo=" << a << " tab=" << b << " opt=" << c
         << " (Expected: " << expected << ")"
         << (ok ? " OK" : " FAIL") << '\n';
};

run_lcs("",          "",          0);   // both empty
run_lcs("abc",       "",          0);   // one empty
run_lcs("",          "abc",       0);
run_lcs("abc",       "abc",       3);   // identical
run_lcs("abc",       "def",       0);   // disjoint
run_lcs("abcde",     "ace",       3);   // classic: "ace"
run_lcs("AGGTAB",    "GXTXAYB",   4);   // "GTAB"
run_lcs("aaaa",      "aa",        2);   // repeats: pick any two
run_lcs("abcdef",    "fedcba",    1);   // reverse: only one common char
run_lcs("xyzabczyx", "abc",       3);
````

## 5.2 Print LCS

**Problem.** Same input as LCS, but return *one* longest common subsequence string (any valid one if multiple exist).

---

### Theory — reconstruction by backtracking through the DP table

The LCS *length* tells us "what" — we now want "which characters." Two approaches:

1. **Parent-pointer table** — alongside $\text{dp}[i][j]$, store which branch was taken (match / drop-from-$s$ / drop-from-$t$). Then trace back from $(n, m)$.
2. **Re-derive the branch from $\text{dp}[i][j]$ alone** — no extra storage; just inspect the DP table during traceback.

We adopt approach 2: the **full DP table already contains enough information** to reconstruct any valid LCS, by reversing the recurrence.

**Backtrack invariant.** Starting at $(i, j) = (n, m)$, at each step:
- If $s[i-1] = t[j-1]$, then the LCS we are building includes this character, and the suffix problem is at $(i-1, j-1)$.
- Otherwise, $\text{dp}[i][j] = \max(\text{dp}[i-1][j], \text{dp}[i][j-1])$ — move to whichever is larger (ties broken arbitrarily; this is why "any valid LCS" is acceptable).

Append the matched character to a stack (or prepend to a string). Stop when $i = 0$ or $j = 0$.

**Correctness.** At each step, we move to a state whose $\text{dp}$ value is exactly $\text{dp}[i][j] - 1$ (on match) or $\text{dp}[i][j]$ (on mismatch). The walk terminates in at most $n + m$ steps. The collected characters form a sequence of length $\text{dp}[n][m]$ that, by the case analysis, is a subsequence of both strings.

**Complexity.** DP fill: $O(nm)$ time and space. Backtrack: $O(n + m)$ extra time, $O(\text{LCS length})$ extra space for the result string.

**Why we need the full table.** Space-optimized LCS (two rows) only knows the *length*; it discards the table needed for traceback. Reconstruction *requires* $O(nm)$ space.

**Boundary transitions during traceback.**

| Condition at $(i, j)$ | Action |
|---|---|
| $i = 0$ or $j = 0$ | Stop |
| $s[i-1] = t[j-1]$ | Emit $s[i-1]$, move to $(i-1, j-1)$ |
| $s[i-1] \neq t[j-1],\ \text{dp}[i-1][j] \ge \text{dp}[i][j-1]$ | Move to $(i-1, j)$ |
| $s[i-1] \neq t[j-1],\ \text{dp}[i-1][j] < \text{dp}[i][j-1]$ | Move to $(i, j-1)$ |

**Delta vs LCS.** No new DP; just a post-processing backtrack. The lesson: a 2D DP table is a directed acyclic graph of decisions, and tracing back along the *winning* edges reconstructs the actual solution.

---

````c++
// Print LCS — fill the table, then backtrack
#include <vector>
#include <string>
#include <iostream>
using namespace std;

string printLCS(const string& s, const string& t) {
    int n = (int)s.size(), m = (int)t.size();
    // Build the full DP table (cannot space-optimize: traceback needs all cells)
    vector<vector<int>> dp(n + 1, vector<int>(m + 1, 0));
    for (int i = 1; i <= n; ++i)
        for (int j = 1; j <= m; ++j) {
            if (s[i-1] == t[j-1])
                dp[i][j] = 1 + dp[i-1][j-1];
            else
                dp[i][j] = max(dp[i-1][j], dp[i][j-1]);
        }
    // Backtrack from (n, m) to a boundary
    string result;
    int i = n, j = m;
    while (i > 0 && j > 0) {
        if (s[i-1] == t[j-1]) {
            result.push_back(s[i-1]);                      // this char is part of the LCS
            --i; --j;                                      // diagonal step
        } else if (dp[i-1][j] >= dp[i][j-1]) {
            --i;                                           // drop from s
        } else {
            --j;                                           // drop from t
        }
    }
    reverse(result.begin(), result.end());                 // we built it back-to-front
    return result;
}
````

````c++
// Tests — Print LCS
auto run_pLCS = [](string s, string t, int expectedLen, string oneValidLCS) {
    string r = printLCS(s, t);
    bool lenOk = ((int)r.size() == expectedLen);
    // Verify r is a subsequence of both s and t (correctness check that doesn't pin one specific output)
    auto isSubseq = [](const string& sub, const string& str) {
        size_t k = 0;
        for (char c : str) if (k < sub.size() && c == sub[k]) ++k;
        return k == sub.size();
    };
    bool subOk = isSubseq(r, s) && isSubseq(r, t);
    bool ok = lenOk && subOk;
    cout << "s=\"" << s << "\" t=\"" << t << "\""
         << " -> \"" << r << "\" (len=" << r.size()
         << ", one valid is \"" << oneValidLCS << "\" len=" << expectedLen << ")"
         << (ok ? " OK" : " FAIL") << '\n';
};

run_pLCS("",          "",         0,  "");
run_pLCS("abc",       "",         0,  "");
run_pLCS("abc",       "abc",      3,  "abc");
run_pLCS("abc",       "def",      0,  "");
run_pLCS("abcde",     "ace",      3,  "ace");
run_pLCS("AGGTAB",    "GXTXAYB",  4,  "GTAB");
run_pLCS("aaaa",      "aa",       2,  "aa");
run_pLCS("abcdef",    "fedcba",   1,  "a");  // multiple valid 1-char LCSs
run_pLCS("xyzabczyx", "abc",      3,  "abc");
````

## 5.3 Longest Common Substring

**Problem.** Find the longest **contiguous** substring that appears in both $s$ and $t$. Return its length.

> **Critical distinction**: substring (contiguous) ≠ subsequence (any order-preserving selection). "abc" is a substring of "xabcy"; "ac" is a subsequence but not a substring.

---

### Theory

**State definition — the crucial twist.**
$\text{dp}[i][j]$ = length of the longest common substring that **ends exactly at** $s[i-1]$ **and** $t[j-1]$.

This is *not* the same as "length of LCS over prefixes." The "ends exactly here" anchoring is what enforces contiguity.

**Invariant.** $\text{dp}[i][j]$ requires both prefixes to end at a matching character at $(s[i-1], t[j-1])$ that extends a common substring backwards through equal-length prefixes. If they don't match here, the substring **cannot end here**, so $\text{dp}[i][j] = 0$.

**Recurrence.**
$$\text{dp}[i][j] = \begin{cases} 1 + \text{dp}[i-1][j-1] & \text{if}\ s[i-1] = t[j-1] \\ 0 & \text{if}\ s[i-1] \neq t[j-1] \end{cases}$$

**Answer.** $\text{answer} = \max_{i, j}\ \text{dp}[i][j]$ — the longest substring may end anywhere.

**Base cases.** $\text{dp}[0][j] = \text{dp}[i][0] = 0$.

**Boundary transitions table.**

| Case | Recurrence value |
|---|---|
| $i = 0$ or $j = 0$ | $0$ |
| $s[i-1] = t[j-1]$ | $1 + \text{dp}[i-1][j-1]$ |
| $s[i-1] \neq t[j-1]$ | $0$ |

**Why it works.**
- *Optimal substructure for substrings:* if a common substring of length $L$ ends at $(s[i-1], t[j-1])$, then dropping the last char gives a common substring of length $L - 1$ ending at $(s[i-2], t[j-2])$. Strict contiguity ⟹ we *must* read from $(i-1, j-1)$ specifically — no other source works.
- *Mismatch ⟹ hard reset:* a common substring ending exactly at non-matching characters cannot exist. Any longer common substring crossing this position would have had to end at a *match*, recorded elsewhere in the table.

**Complexity.** Time $O(nm)$. Space $O(nm)$ → $O(\min(n, m))$ with two rows. The "track running max" doesn't need the full table, so unlike Print LCS, space optimization is fine here.

**Delta vs LCS.**
- The two recurrences look almost identical, but the **mismatch case differs profoundly**: LCS inherits via $\max(\text{dp}[i-1][j], \text{dp}[i][j-1])$; Common Substring resets to $0$.
- LCS answer is at $\text{dp}[n][m]$; Common Substring answer is the *running max* over the entire table.
- The state semantics are completely different despite looking similar: LCS = "longest over prefixes," Common Substring = "longest ending exactly here." **Confusing these is the most common bug in this subtopic.**

---

````c++
// Longest Common Substring — three implementations
#include <vector>
#include <string>
#include <algorithm>
#include <iostream>
using namespace std;

// (A) Top-down memoization
//     dp[i][j] = length of common substring ending exactly at s[i-1], t[j-1].
//     Need a separate global tracker for the running max.
int lcsubstr_memo_helper(int i, int j, const string& s, const string& t,
                           vector<vector<int>>& dp, int& best) {
    if (i == 0 || j == 0) return 0;
    if (dp[i][j] != -1) return dp[i][j];
    int val;
    if (s[i-1] == t[j-1])
        val = 1 + lcsubstr_memo_helper(i-1, j-1, s, t, dp, best);
    else
        val = 0;                                           // hard reset on mismatch
    // Still recurse into i-1,j and i,j-1 to populate dp table so the running max sees all cells.
    // Memo will short-circuit repeated work.
    lcsubstr_memo_helper(i-1, j, s, t, dp, best);
    lcsubstr_memo_helper(i, j-1, s, t, dp, best);
    dp[i][j] = val;
    if (val > best) best = val;
    return val;
}
int lcsubstr_memo(const string& s, const string& t) {
    int n = (int)s.size(), m = (int)t.size();
    vector<vector<int>> dp(n + 1, vector<int>(m + 1, -1));
    int best = 0;
    lcsubstr_memo_helper(n, m, s, t, dp, best);
    return best;
}

// (B) Tabulation — O(nm)
int lcsubstr_tab(const string& s, const string& t) {
    int n = (int)s.size(), m = (int)t.size();
    vector<vector<int>> dp(n + 1, vector<int>(m + 1, 0));
    int best = 0;
    for (int i = 1; i <= n; ++i)
        for (int j = 1; j <= m; ++j) {
            if (s[i-1] == t[j-1]) {
                dp[i][j] = 1 + dp[i-1][j-1];               // extend the diagonal
                if (dp[i][j] > best) best = dp[i][j];      // track running max
            }
            // mismatch: dp[i][j] stays at 0 (initialized)
        }
    return best;
}

// (C) Space-optimized — O(min(n,m)) two rows
//     Valid: only reads prev[j-1] (NW diagonal). No row-i lookups needed.
int lcsubstr_opt(const string& s, const string& t) {
    int n = (int)s.size(), m = (int)t.size();
    if (m > n) return lcsubstr_opt(t, s);                  // ensure t shorter
    vector<int> prev(m + 1, 0), cur(m + 1, 0);
    int best = 0;
    for (int i = 1; i <= n; ++i) {
        cur[0] = 0;                                        // base: empty t prefix
        for (int j = 1; j <= m; ++j) {
            if (s[i-1] == t[j-1]) {
                cur[j] = 1 + prev[j-1];
                if (cur[j] > best) best = cur[j];
            } else {
                cur[j] = 0;                                // explicit reset (cur from prev iteration may not be 0)
            }
        }
        swap(prev, cur);
    }
    return best;
}
````

````c++
// Tests — Longest Common Substring
auto run_lcss = [](string s, string t, int expected) {
    int a = lcsubstr_memo(s, t);
    int b = lcsubstr_tab(s, t);
    int c = lcsubstr_opt(s, t);
    bool ok = (a == expected && b == expected && c == expected);
    cout << "s=\"" << s << "\" t=\"" << t << "\""
         << " -> memo=" << a << " tab=" << b << " opt=" << c
         << " (Expected: " << expected << ")"
         << (ok ? " OK" : " FAIL") << '\n';
};

run_lcss("",          "",         0);
run_lcss("abc",       "",         0);
run_lcss("abc",       "abc",      3);
run_lcss("abc",       "def",      0);
run_lcss("abcde",     "ace",      1);   // LCS is "ace" (len 3) but longest *substring* is "a","c","e" — all len 1
run_lcss("abcdxyz",   "xyzabcd",  4);   // "abcd"
run_lcss("ABABC",     "BABCA",    4);   // "BABC"
run_lcss("aaaa",      "aa",       2);   // "aa"
run_lcss("hello",     "world",    1);   // "l" or "o"
run_lcss("OldSite:GeeksforGeeks.org", "NewSite:GeeksQuiz.com", 10);  // "Site:Geeks"
````

## 5.4 Longest Palindromic Subsequence (LPS)

**Problem.** Given a string $s$, find the length of the longest subsequence that is a palindrome.

---

### Theory — reduction to LCS

A subsequence of $s$ is a palindrome iff it is equal to its own reverse. So we ask: **what is the longest sequence that is a subsequence of both $s$ and $\text{rev}(s)$?** That's $\text{LCS}(s,\ \text{rev}(s))$.

**Theorem.** $\text{LPS}(s) = \text{LCS}(s,\ \text{rev}(s))$.

**Proof.**
- ($\le$) Any palindromic subsequence $P$ of $s$ is also a subsequence of $\text{rev}(s)$ (since $P = \text{rev}(P)$ and $P$ appears in $s$ ⟹ $\text{rev}(P) = P$ appears in $\text{rev}(s)$). So $\text{LPS}(s) \le \text{LCS}(s, \text{rev}(s))$.
- ($\ge$) Any common subsequence $C$ of $s$ and $\text{rev}(s)$ corresponds to one set of positions in $s$ and one set in $\text{rev}(s)$. The character sequence $C$ appears identically in both. The positions in $\text{rev}(s)$ correspond to *reversed* positions in $s$. A characters-then-positions argument shows $C$ must equal $\text{rev}(C)$ (it appears both forward and backward in $s$ at consistent positions), so $C$ is a palindrome.  Hence $\text{LCS}(s, \text{rev}(s)) \le \text{LPS}(s)$.

So we just call LCS. No new DP machinery.

**Alternative direct formulation** (worth knowing, not implemented here).
$\text{dp}[i][j]$ = LPS of substring $s[i..j]$. Recurrence:
$$\text{dp}[i][j] = \begin{cases} \text{dp}[i+1][j-1] + 2 & \text{if}\ s[i] = s[j],\ j - i \ge 2 \\ 2 & \text{if}\ s[i] = s[j],\ j - i = 1 \\ \max(\text{dp}[i+1][j], \text{dp}[i][j-1]) & \text{otherwise} \end{cases}$$
This is an **interval DP** (preview of Subtopic 8). The reduction-to-LCS version is cleaner and reuses our kernel; we'll use it.

**Complexity.** $O(n^2)$ time, $O(n)$ space (inherited from LCS-opt with $m = n$).

**Delta.** Reduction-by-reversal. The combinatorial trick: a palindrome is fixed by reversal, so it's exactly a common element of "$s$" and "reversed $s$."

---

````c++
// Longest Palindromic Subsequence — LCS with the reversed string
#include <string>
#include <algorithm>
#include <iostream>
using namespace std;

int LPS(const string& s) {
    string r = s;
    reverse(r.begin(), r.end());
    return lcs_opt(s, r);                                  // reuse the O(n)-space LCS
}
````

````c++
// Tests — Longest Palindromic Subsequence
auto run_lps = [](string s, int expected) {
    int a = LPS(s);
    cout << "s=\"" << s << "\""
         << " -> " << a << " (Expected: " << expected << ")"
         << (a == expected ? " OK" : " FAIL") << '\n';
};

run_lps("",            0);
run_lps("a",           1);
run_lps("aa",          2);
run_lps("ab",          1);
run_lps("bbbab",       4);   // "bbbb"
run_lps("cbbd",        2);   // "bb"
run_lps("abcdba",      5);   // "abcba" or "abdba"
run_lps("racecar",     7);   // already a palindrome
run_lps("character",   5);   // "carac"
run_lps("aaaaaa",      6);   // already palindrome
````

## 5.5 Minimum Insertions to Make a String a Palindrome

**Problem.** Given $s$, find the minimum number of characters to *insert* anywhere in $s$ so that the result is a palindrome.

---

### Theory — reduction via LPS

**Key claim.** $\text{answer} = n - \text{LPS}(s)$, where $n = |s|$.

**Proof / intuition.** The LPS of $s$ is a palindromic skeleton already present in $s$. The remaining $n - \text{LPS}$ characters are "unmatched" — each requires exactly one insertion to provide a mirror partner across the palindromic center.

More formally: consider any optimal sequence of insertions producing a palindrome $P$ of length $n + k$. Looking at which characters of $P$ came from $s$ versus from insertions: the original characters of $s$ form a subsequence of $P$. Since $P$ is a palindrome, the original characters of $s$ also form a *palindromic* subsequence of $s$ (read forward in $s$ and check they're symmetric in $P$). So the number of original chars retained is $\le \text{LPS}(s)$, hence insertions $\ge n - \text{LPS}(s)$.

Conversely, $n - \text{LPS}(s)$ insertions suffice: for each char of $s$ not in the LPS, insert a matching mirror. (Explicit construction omitted.)

**Complexity.** $O(n^2)$ time, $O(n)$ space — dominated by LPS.

**Delta.** Pure formula on LPS. The lesson: many "minimum edits to make a property hold" problems unfold as $n - (\text{largest subsequence with that property})$.

---

````c++
// Minimum Insertions to make Palindrome
#include <string>
#include <iostream>
using namespace std;

int minInsertionsToPalindrome(const string& s) {
    return (int)s.size() - LPS(s);                          // n - LPS(s)
}
````

````c++
// Tests — Min Insertions to Palindrome
auto run_mip = [](string s, int expected) {
    int a = minInsertionsToPalindrome(s);
    cout << "s=\"" << s << "\""
         << " -> " << a << " (Expected: " << expected << ")"
         << (a == expected ? " OK" : " FAIL") << '\n';
};

run_mip("",          0);   // empty is already palindrome
run_mip("a",         0);   // single char
run_mip("aa",        0);   // already palindrome
run_mip("ab",        1);   // insert 'a' or 'b': "aba" or "bab"
run_mip("abcd",      3);   // LPS=1, need 3 insertions
run_mip("aebcbda",   2);   // LPS = "abcba" (len 5), need 2: "adebcbeda"
run_mip("leetcode",  5);   // LPS = "eee" (len 3), need 5
run_mip("racecar",   0);   // already palindrome
````

## 5.6 Minimum Insertions/Deletions to Convert A to B

**Problem.** Given two strings $A$ (length $n$) and $B$ (length $m$), find the minimum number of insertions and deletions (in $A$) to make it equal to $B$. Replacements are not allowed in this variant.

---

### Theory — reduction via LCS

**Key claim.** $\text{answer} = (n - L) + (m - L)$ where $L = \text{LCS}(A, B)$.

**Proof.** Any sequence of inserts-and-deletes maps $A$ to $B$. The characters of $A$ that *survive* (are not deleted) form a subsequence of $A$. Those same survivor characters appear in $B$ in the same relative order (since insertions don't shuffle them), so they form a subsequence of $B$. Hence the survivor count $\le \text{LCS}(A, B)$.

- Deletions = (chars of $A$ not in survivors) $\ge n - L$.
- Insertions = (chars of $B$ not in survivors) $\ge m - L$.
- Total $\ge (n - L) + (m - L) = n + m - 2L$.

Conversely, this is achievable: delete from $A$ every char not in the LCS ($n - L$ ops), then insert into the resulting string every char of $B$ not in the LCS, at the right positions ($m - L$ ops).

**Complexity.** $O(nm)$ time, $O(\min(n, m))$ space.

**Delta vs LCS.** Pure post-processing formula. The "Editing" trichotomy is showing up: this problem allows only inserts/deletes; Edit Distance (§5.8) adds replace as a third operation, which generally gives a strictly smaller minimum.

---

````c++
// Min Insertions + Deletions to convert A to B
#include <string>
#include <iostream>
using namespace std;

int minInsDel(const string& A, const string& B) {
    int L = lcs_opt(A, B);
    return ((int)A.size() - L) + ((int)B.size() - L);       // (deletions from A) + (insertions to reach B)
}
````

````c++
// Tests — Min Ins/Del A → B
auto run_id = [](string A, string B, int expected) {
    int a = minInsDel(A, B);
    cout << "A=\"" << A << "\" B=\"" << B << "\""
         << " -> " << a << " (Expected: " << expected << ")"
         << (a == expected ? " OK" : " FAIL") << '\n';
};

run_id("",        "",        0);    // both empty
run_id("abc",     "",        3);    // delete 3
run_id("",        "abc",     3);    // insert 3
run_id("abc",     "abc",     0);    // identical
run_id("heap",    "pea",     3);    // LCS="ea" len 2: (4-2)+(3-2) = 2+1 = 3
run_id("sea",     "eat",     2);    // LCS="ea" len 2
run_id("abcdef",  "azced",   5);     // LCS="acd" or "ace" length 3: (6-3)+(5-3) = 5
````

## 5.7 Shortest Common Supersequence (SCS)

**Problem.** Given $A$ and $B$, find the shortest string $S$ such that *both* $A$ and $B$ are subsequences of $S$. Return $S$ (or just its length, depending on the variant).

---

### Theory

**Length formula.** $|\text{SCS}(A, B)| = n + m - \text{LCS}(A, B)$.

**Proof.** $S$ contains $A$ and $B$ as subsequences. The shared characters can overlap maximally in $S$ — overlap is exactly the LCS. So total length is $n + m - (\text{overlap}) \ge n + m - \text{LCS}$. Conversely, we can construct an $S$ achieving this length (see reconstruction below).

**Reconstruction by backtracking through the LCS DP table.**

Starting at $(i, j) = (n, m)$, walk backwards while emitting characters:

1. If $A[i-1] = B[j-1]$: emit this single character (it represents the shared LCS char); move to $(i-1, j-1)$.
2. Else if $\text{dp}[i-1][j] \ge \text{dp}[i][j-1]$: emit $A[i-1]$ (it's unique to $A$); move to $(i-1, j)$.
3. Else: emit $B[j-1]$ (unique to $B$); move to $(i, j-1)$.
4. After reaching a boundary, flush the remaining prefix of the non-exhausted string.

Reverse the emitted string at the end (we built it back-to-front).

**Why this works.** At each step we emit exactly one character of the supersequence. The case analysis ensures that:
- Shared characters (LCS) are emitted once and serve as part of both $A$ and $B$.
- Unique characters of $A$ and $B$ are emitted exactly when their source-string traversal is "ahead" in producing them.

The total emit count is $\text{dp}[n][m]$ shared + remaining chars from $A$ and $B$ = $n + m - \text{LCS}$.

**Complexity.** $O(nm)$ time and space (cannot skip the table — needed for backtrack).

**Delta vs Print LCS.**
- Same DP table.
- Print LCS emits *only* matched characters during backtrack.
- SCS emits *every* character along the path — matched ones once, unmatched ones each.
- Structurally identical traceback, different emission rule.

---

````c++
// Shortest Common Supersequence — full reconstruction
#include <string>
#include <vector>
#include <algorithm>
#include <iostream>
using namespace std;

string SCS(const string& A, const string& B) {
    int n = (int)A.size(), m = (int)B.size();
    // Build LCS DP table
    vector<vector<int>> dp(n + 1, vector<int>(m + 1, 0));
    for (int i = 1; i <= n; ++i)
        for (int j = 1; j <= m; ++j) {
            if (A[i-1] == B[j-1]) dp[i][j] = 1 + dp[i-1][j-1];
            else dp[i][j] = max(dp[i-1][j], dp[i][j-1]);
        }
    // Backtrack, emitting every character on the path
    string result;
    int i = n, j = m;
    while (i > 0 && j > 0) {
        if (A[i-1] == B[j-1]) {
            result.push_back(A[i-1]);                      // shared char, emit once
            --i; --j;
        } else if (dp[i-1][j] >= dp[i][j-1]) {
            result.push_back(A[i-1]);                      // A's unique char
            --i;
        } else {
            result.push_back(B[j-1]);                      // B's unique char
            --j;
        }
    }
    // Flush remaining prefix of whichever string isn't exhausted
    while (i > 0) { result.push_back(A[i-1]); --i; }
    while (j > 0) { result.push_back(B[j-1]); --j; }
    reverse(result.begin(), result.end());                 // built back-to-front
    return result;
}

int SCS_length(const string& A, const string& B) {
    return (int)A.size() + (int)B.size() - lcs_opt(A, B);  // closed-form length
}
````

````c++
// Tests — Shortest Common Supersequence
auto run_scs = [](string A, string B, int expectedLen) {
    string s = SCS(A, B);
    int lenFormula = SCS_length(A, B);
    // Correctness: A and B must each be subsequences of s, and s has expected length
    auto isSubseq = [](const string& sub, const string& str) {
        size_t k = 0;
        for (char c : str) if (k < sub.size() && c == sub[k]) ++k;
        return k == sub.size();
    };
    bool lenOk = ((int)s.size() == expectedLen) && (lenFormula == expectedLen);
    bool subOk = isSubseq(A, s) && isSubseq(B, s);
    bool ok = lenOk && subOk;
    cout << "A=\"" << A << "\" B=\"" << B << "\""
         << " -> \"" << s << "\" (len=" << s.size() << ", formula=" << lenFormula
         << ", expected " << expectedLen << ")"
         << (ok ? " OK" : " FAIL") << '\n';
};

run_scs("",       "",        0);
run_scs("abc",    "",        3);
run_scs("",       "xyz",     3);
run_scs("abc",    "abc",     3);
run_scs("abc",    "def",     6);    // disjoint: n+m
run_scs("AGGTAB", "GXTXAYB", 9);    // n=6, m=7, LCS=4 -> 9
run_scs("abac",   "cab",     5);    // LCS="ab" len 2: 4+3-2 = 5
run_scs("brute",  "groot",   8);    // LCS="rot" len 3: 5+5-3 = 7? Let me check: b-r-u-t-e vs g-r-o-o-t. Common subsequences "r","t","rt". LCS=2 ("rt"). So 5+5-2 = 8.
````

# Family B — Edit Distance

We now allow a third operation — replace — which generally lowers the minimum-ops count below what Family A's $n + m - 2L$ formula achieves.

## 5.8 Edit Distance (Levenshtein)

**Problem.** Given $s$ and $t$, find the minimum number of edits (insert / delete / replace, each costing 1) to convert $s$ into $t$.

---

### Theory

**State definition.**
$\text{dp}[i][j]$ = minimum edits to convert $s[0..i-1]$ into $t[0..j-1]$.

**Invariant.** $\text{dp}[i][j]$ is the optimum over all edit sequences transforming the $s$-prefix of length $i$ into the $t$-prefix of length $j$. The case-split is on the **last edit operation** (or non-operation):

- If $s[i-1] = t[j-1]$: no edit needed at this position; the cost is the cost of editing the shorter prefixes $\text{dp}[i-1][j-1]$.
- Otherwise, exactly one of three operations was the last one performed (any optimal sequence's *final* op):
  - **Replace** $s[i-1]$ with $t[j-1]$ (cost 1): reduces to $\text{dp}[i-1][j-1] + 1$.
  - **Delete** $s[i-1]$ (cost 1): reduces to $\text{dp}[i-1][j] + 1$.
  - **Insert** $t[j-1]$ at the end of the $s$-prefix (cost 1): reduces to $\text{dp}[i][j-1] + 1$.

**Recurrence — the three transitions on mismatch.**
$$\text{dp}[i][j] = \begin{cases} \text{dp}[i-1][j-1] & \text{if}\ s[i-1] = t[j-1] \\ 1 + \min\!\big(\underbrace{\text{dp}[i-1][j-1]}_{\text{replace}},\ \underbrace{\text{dp}[i-1][j]}_{\text{delete}},\ \underbrace{\text{dp}[i][j-1]}_{\text{insert}}\big) & \text{otherwise} \end{cases}$$

**Why the $\min$ over the three is correct.** Any sequence of edits ends with *exactly one* last operation (the trivial case "no edits needed" is captured by the match branch). The three branches correspond to the three possible identities of the last edit. The minimum over them is the optimum.

**Base cases.**
- $\text{dp}[0][j] = j$ — converting empty to $t[0..j-1]$ requires $j$ insertions.
- $\text{dp}[i][0] = i$ — converting $s[0..i-1]$ to empty requires $i$ deletions.

**Boundary transitions table.**

| Case | Recurrence value |
|---|---|
| $i = 0$ | $j$ (insert all of $t$) |
| $j = 0$ | $i$ (delete all of $s$) |
| $s[i-1] = t[j-1]$ | $\text{dp}[i-1][j-1]$ |
| $s[i-1] \neq t[j-1]$ | $1 + \min(\text{dp}[i-1][j-1], \text{dp}[i-1][j], \text{dp}[i][j-1])$ |

**Why it works.**
- *Optimal substructure:* Each of the three operations cleanly reduces to a strictly smaller subproblem (one prefix shrinks).
- *Overlapping subproblems:* $\Theta(3^{n+m})$ naïve calls vs $\Theta(nm)$ states.

**Complexity.** Time $O(nm)$. Space $O(nm)$ → $O(\min(n, m))$ with two rows.

**Delta vs Min Ins/Del.** A third branch (replace) joins the mismatch case. Replace costs 1 op; the Min Ins/Del approach would use 2 ops (delete + insert) to achieve the same character change. So Edit Distance $\le$ Min Ins/Del always.

**Why $\min$, not $\max$.** Trichotomy axis: optimization with $\min$ since we're counting operations and want fewest. Same skeleton as Coin Change (min coins).

---

````c++
// Edit Distance — three implementations
#include <vector>
#include <string>
#include <algorithm>
#include <iostream>
using namespace std;

// (A) Top-down memoization
//     dp[i][j] = min edits to convert s[0..i-1] to t[0..j-1].
int editDist_memo_helper(int i, int j, const string& s, const string& t, vector<vector<int>>& dp) {
    if (i == 0) return j;                                  // base: insert all of t's prefix
    if (j == 0) return i;                                  // base: delete all of s's prefix
    if (dp[i][j] != -1) return dp[i][j];
    if (s[i-1] == t[j-1]) {
        dp[i][j] = editDist_memo_helper(i-1, j-1, s, t, dp);    // match: no edit, inherit
    } else {
        int replace = editDist_memo_helper(i-1, j-1, s, t, dp); // replace last char of s
        int del     = editDist_memo_helper(i-1, j,   s, t, dp); // delete last char of s
        int ins     = editDist_memo_helper(i,   j-1, s, t, dp); // insert t's last char at end of s
        dp[i][j] = 1 + min({replace, del, ins});           // recurrence: 1 + min over three ops
    }
    return dp[i][j];
}
int editDist_memo(const string& s, const string& t) {
    int n = (int)s.size(), m = (int)t.size();
    vector<vector<int>> dp(n + 1, vector<int>(m + 1, -1));
    return editDist_memo_helper(n, m, s, t, dp);
}

// (B) Tabulation — O(nm)
int editDist_tab(const string& s, const string& t) {
    int n = (int)s.size(), m = (int)t.size();
    vector<vector<int>> dp(n + 1, vector<int>(m + 1, 0));
    for (int j = 0; j <= m; ++j) dp[0][j] = j;             // base row
    for (int i = 0; i <= n; ++i) dp[i][0] = i;             // base column
    for (int i = 1; i <= n; ++i) {
        for (int j = 1; j <= m; ++j) {
            if (s[i-1] == t[j-1]) {
                dp[i][j] = dp[i-1][j-1];                   // match
            } else {
                int replace = dp[i-1][j-1];                // 3 predecessors:
                int del     = dp[i-1][j];                  //   replace at NW
                int ins     = dp[i][j-1];                  //   delete  at N
                dp[i][j] = 1 + min({replace, del, ins});   //   insert  at W
            }
        }
    }
    return dp[n][m];
}

// (C) Space-optimized — O(min(n,m)) using two rows
//     Valid: reads dp[i-1][j-1] (NW), dp[i-1][j] (N), dp[i][j-1] (W) — all in prev row or cur row prefix.
int editDist_opt(const string& s, const string& t) {
    int n = (int)s.size(), m = (int)t.size();
    if (m > n) return editDist_opt(t, s);                  // minor optimization: shorter row
    vector<int> prev(m + 1, 0), cur(m + 1, 0);
    for (int j = 0; j <= m; ++j) prev[j] = j;              // base row
    for (int i = 1; i <= n; ++i) {
        cur[0] = i;                                        // base column entry for this row
        for (int j = 1; j <= m; ++j) {
            if (s[i-1] == t[j-1]) {
                cur[j] = prev[j-1];                        // match: NW
            } else {
                int replace = prev[j-1];                   // NW
                int del     = prev[j];                     // N
                int ins     = cur[j-1];                    // W (already computed in this row)
                cur[j] = 1 + min({replace, del, ins});
            }
        }
        swap(prev, cur);
    }
    return prev[m];
}
````

````c++
// Tests — Edit Distance
auto run_ed = [](string s, string t, int expected) {
    int a = editDist_memo(s, t);
    int b = editDist_tab(s, t);
    int c = editDist_opt(s, t);
    bool ok = (a == expected && b == expected && c == expected);
    cout << "s=\"" << s << "\" t=\"" << t << "\""
         << " -> memo=" << a << " tab=" << b << " opt=" << c
         << " (Expected: " << expected << ")"
         << (ok ? " OK" : " FAIL") << '\n';
};

run_ed("",       "",        0);   // both empty
run_ed("abc",    "",        3);   // delete all
run_ed("",       "xyz",     3);   // insert all
run_ed("abc",    "abc",     0);   // identical
run_ed("horse",  "ros",     3);   // classic LeetCode: replace h->r, delete r, delete e
run_ed("intention", "execution", 5);   // classic
run_ed("kitten", "sitting",  3);   // classic Levenshtein example
run_ed("a",      "b",        1);   // single replace
run_ed("abc",    "yabd",     2);   // insert 'y' + replace 'c'->'d'
run_ed("Sunday", "Saturday", 3);   // insert 'a', insert 't', replace 'n'->'r'
````

# Family C — Pattern Matching

The state shape stays $(i, j)$, but now one axis is a **pattern with metacharacters**, not a literal string. The recurrence has special cases for each metacharacter. The combine becomes OR (existence flavor — does the pattern match?).

## 5.9 Wildcard Matching

**Problem.** Given a text $s$ and a pattern $p$ where:
- `?` matches **any single character**.
- `*` matches **any sequence of characters, including the empty sequence**.

Determine whether $p$ matches the *entire* string $s$.

---

### Theory

**State definition.**
$\text{dp}[i][j]$ = true iff the pattern prefix $p[0..i-1]$ matches the text prefix $s[0..j-1]$.

> **Convention here:** $i$ ranges over the pattern, $j$ over the text. This swap from previous problems is just a labeling — it makes the special cases for $p[i-1]$ read naturally.

**Invariant.** $\text{dp}[i][j]$ captures the existence of any alignment between pattern and text prefixes. The recursion case-splits on $p[i-1]$ (the last character of the pattern prefix):

- **Literal match** ($p[i-1] = s[j-1]$) or **`?`** ($p[i-1] = $ `?`): one pattern char consumes one text char. Reduces to $\text{dp}[i-1][j-1]$.
- **`*`**: two sub-options.
  - `*` matches **empty** (consumes nothing in $s$): reduces to $\text{dp}[i-1][j]$.
  - `*` matches **one more character** (consumes $s[j-1]$, but `*` remains in play): reduces to $\text{dp}[i][j-1]$.
  
  Combine: OR.

- **Literal mismatch** ($p[i-1]$ is a literal $\neq s[j-1]$): no match possible.

**Recurrence.**
$$\text{dp}[i][j] = \begin{cases} \text{dp}[i-1][j-1] & \text{if}\ p[i-1] = s[j-1]\ \text{or}\ p[i-1] = \text{`?'} \\ \text{dp}[i-1][j]\ \vee\ \text{dp}[i][j-1] & \text{if}\ p[i-1] = \text{`*'} \\ \text{false} & \text{if}\ p[i-1]\ \text{literal}\ \text{and}\ p[i-1] \neq s[j-1] \end{cases}$$

**Base cases — subtle.**
- $\text{dp}[0][0] = \text{true}$ — empty pattern matches empty text.
- $\text{dp}[0][j] = \text{false}$ for $j \ge 1$ — empty pattern, non-empty text: cannot match.
- $\text{dp}[i][0]$ for $i \ge 1$: empty text. The pattern prefix matches iff **every** character of $p[0..i-1]$ is `*`. Compute: $\text{dp}[i][0] = \text{dp}[i-1][0] \wedge (p[i-1] = \text{`*'})$.

**Boundary transitions table.**

| Case | Recurrence value |
|---|---|
| $i = 0, j = 0$ | true |
| $i = 0, j \ge 1$ | false |
| $i \ge 1, j = 0$ | $\text{dp}[i-1][0] \wedge (p[i-1] = $ `*` $)$ |
| $p[i-1]$ literal $= s[j-1]$ or $p[i-1] = $ `?` | $\text{dp}[i-1][j-1]$ |
| $p[i-1] = $ `*` | $\text{dp}[i-1][j] \vee \text{dp}[i][j-1]$ |
| $p[i-1]$ literal $\neq s[j-1]$ | false |

**Why it works.**
- *Optimal substructure (existence):* a successful match must end with the pattern's last char either matching one text char (literal or `?`) or being a `*` that contributes zero-or-more chars. Each case reduces to a strictly smaller subproblem in both dimensions or just one.
- *Overlapping subproblems:* the `*`-branch creates two children per state, and they overlap heavily; naive recursion is exponential.

**Complexity.** Time $O(nm)$, Space $O(nm)$ → $O(m)$ with two rows ($n$ = pattern length, $m$ = text length).

**Delta vs LCS.**
- Existence (OR combine) replaces optimization (max).
- The match case is the same shape (NW transition).
- The mismatch case **shrinks** for literals (no recurrence — just false) and **expands** for `*` (two branches: empty-match goes "up," one-more-match goes "left").
- Asymmetry: the state is "pattern matches text," not "two strings are similar." So which dimension means what matters.

---

````c++
// Wildcard Matching — three implementations
#include <vector>
#include <string>
#include <iostream>
using namespace std;

// (A) Top-down memoization
//     dp[i][j] = pattern p[0..i-1] matches text s[0..j-1]?
int wildcard_memo_helper(int i, int j, const string& p, const string& s, vector<vector<int>>& dp) {
    if (i == 0 && j == 0) return 1;                        // empty matches empty
    if (i == 0) return 0;                                  // empty pattern, non-empty text
    if (j == 0) {
        // non-empty pattern, empty text: only matches if pattern is all '*'
        for (int k = 0; k < i; ++k) if (p[k] != '*') return 0;
        return 1;
    }
    if (dp[i][j] != -1) return dp[i][j];
    int result;
    if (p[i-1] == s[j-1] || p[i-1] == '?') {
        result = wildcard_memo_helper(i-1, j-1, p, s, dp); // single-char match
    } else if (p[i-1] == '*') {
        // * matches empty (consume pattern) OR * matches one more (consume text)
        result = wildcard_memo_helper(i-1, j, p, s, dp) || wildcard_memo_helper(i, j-1, p, s, dp);
    } else {
        result = 0;                                        // literal mismatch
    }
    dp[i][j] = result ? 1 : 0;
    return dp[i][j];
}
bool wildcard_memo(const string& s, const string& p) {
    int n = (int)p.size(), m = (int)s.size();
    vector<vector<int>> dp(n + 1, vector<int>(m + 1, -1));
    return wildcard_memo_helper(n, m, p, s, dp) == 1;
}

// (B) Tabulation — O(nm)
bool wildcard_tab(const string& s, const string& p) {
    int n = (int)p.size(), m = (int)s.size();
    vector<vector<bool>> dp(n + 1, vector<bool>(m + 1, false));
    dp[0][0] = true;                                       // base
    for (int i = 1; i <= n; ++i)                            // empty text: only all-* prefix matches
        dp[i][0] = dp[i-1][0] && (p[i-1] == '*');
    for (int i = 1; i <= n; ++i) {
        for (int j = 1; j <= m; ++j) {
            if (p[i-1] == s[j-1] || p[i-1] == '?') {
                dp[i][j] = dp[i-1][j-1];                   // match
            } else if (p[i-1] == '*') {
                dp[i][j] = dp[i-1][j] || dp[i][j-1];       // * = empty OR * = one more
            }
            // else: literal mismatch → dp[i][j] stays false
        }
    }
    return dp[n][m];
}

// (C) Space-optimized — O(m) two rows
bool wildcard_opt(const string& s, const string& p) {
    int n = (int)p.size(), m = (int)s.size();
    vector<bool> prev(m + 1, false), cur(m + 1, false);
    prev[0] = true;                                        // dp[0][0]
    // prev now represents row 0; we'll iterate rows 1..n, swapping into prev each time.
    for (int i = 1; i <= n; ++i) {
        cur[0] = prev[0] && (p[i-1] == '*');               // empty text: all-* prefix
        for (int j = 1; j <= m; ++j) {
            if (p[i-1] == s[j-1] || p[i-1] == '?') {
                cur[j] = prev[j-1];                        // NW
            } else if (p[i-1] == '*') {
                cur[j] = prev[j] || cur[j-1];              // N (empty match) or W (one more match)
            } else {
                cur[j] = false;
            }
        }
        swap(prev, cur);
    }
    return prev[m];
}
````

````c++
// Tests — Wildcard Matching
auto run_wm = [](string s, string p, bool expected) {
    bool a = wildcard_memo(s, p);
    bool b = wildcard_tab(s, p);
    bool c = wildcard_opt(s, p);
    bool ok = (a == expected && b == expected && c == expected);
    cout << "s=\"" << s << "\" p=\"" << p << "\""
         << " -> memo=" << a << " tab=" << b << " opt=" << c
         << " (Expected: " << expected << ")"
         << (ok ? " OK" : " FAIL") << '\n';
};

run_wm("",         "",        true);   // both empty
run_wm("",         "*",       true);   // * matches empty
run_wm("",         "***",     true);
run_wm("",         "?",       false);  // ? needs one char
run_wm("a",        "?",       true);
run_wm("a",        "*",       true);
run_wm("abc",      "*",       true);
run_wm("abc",      "a*c",     true);
run_wm("abc",      "a?c",     true);
run_wm("abc",      "a*b*c",   true);
run_wm("abc",      "abcd",    false);
run_wm("abcde",    "*e",      true);
run_wm("aa",       "a",       false);  // pattern shorter than text without *
run_wm("adceb",    "*a*b",    true);   // LeetCode example
run_wm("acdcb",    "a*c?b",   false);  // LeetCode example
run_wm("mississippi", "m??*ss*?i*pi", false);  // LeetCode example
````

## 5.10 Regex Matching

**Problem.** Given text $s$ and pattern $p$ where:
- `.` matches **any single character**.
- `*` matches **zero or more of the preceding element**.

(Unlike wildcard, `*` is a *modifier* of the immediately preceding pattern character, not a standalone wildcard.) Determine whether $p$ matches the entire $s$.

---

### Theory

**State definition.**
$\text{dp}[i][j]$ = true iff $p[0..i-1]$ matches $s[0..j-1]$.

**Invariant.** Same as wildcard, but the special-case handling of `*` is fundamentally different because **`*` binds to the preceding character**. So when we see $p[i-1] = $ `*`, we look at $p[i-2]$ to determine what is being repeated.

**Case-split on $p[i-1]$.**

**Case A — literal or `.` match.**
If $p[i-1] = s[j-1]$ or $p[i-1] = $ `.`: $\text{dp}[i][j] = \text{dp}[i-1][j-1]$.

**Case B — `*`.**
The pair `c*` (where `c = p[i-2]`) can match:
- **Zero occurrences** of `c`: skip the pair entirely → $\text{dp}[i-2][j]$.
- **One or more occurrences** of `c`: but only if `c` matches $s[j-1]$ (i.e., $p[i-2] = s[j-1]$ or $p[i-2] = $ `.`). In that case, consume one $s$ character; `c*` remains in play → $\text{dp}[i][j-1]$.

Combine: OR.

**Case C — literal mismatch.**
$p[i-1]$ literal and $\neq s[j-1]$, not `.`, not `*`: $\text{dp}[i][j] = \text{false}$.

**Recurrence.**
$$\text{dp}[i][j] = \begin{cases} \text{dp}[i-1][j-1] & \text{if}\ p[i-1] = s[j-1]\ \text{or}\ p[i-1] = \text{`.'} \\[6pt] \text{dp}[i-2][j]\ \vee\ \big(\text{matches}(p[i-2], s[j-1])\ \wedge\ \text{dp}[i][j-1]\big) & \text{if}\ p[i-1] = \text{`*'} \\[6pt] \text{false} & \text{otherwise} \end{cases}$$

where $\text{matches}(a, b) := (a = b\ \vee\ a = \text{`.'})$.

**Base cases — even more subtle than wildcard.**
- $\text{dp}[0][0] = \text{true}$.
- $\text{dp}[0][j] = \text{false}$ for $j \ge 1$ — empty pattern, non-empty text.
- $\text{dp}[i][0]$ — empty text:
  - A pattern matches empty text iff it consists entirely of `c*` pairs (each contributing zero occurrences).
  - Compute as: $\text{dp}[i][0] = (i \ge 2)\ \wedge\ (p[i-1] = $ `*` $)\ \wedge\ \text{dp}[i-2][0]$.

**Boundary transitions table.**

| Case | Recurrence value |
|---|---|
| $i = 0, j = 0$ | true |
| $i = 0, j \ge 1$ | false |
| $i \ge 1, j = 0,\ p[i-1] = $ `*` | $\text{dp}[i-2][0]$ |
| $i \ge 1, j = 0,\ p[i-1] \neq $ `*` | false |
| $p[i-1] = s[j-1]$ or $p[i-1] = $ `.` | $\text{dp}[i-1][j-1]$ |
| $p[i-1] = $ `*`, $j \ge 1$ | $\text{dp}[i-2][j]\ \vee\ (\text{matches}(p[i-2], s[j-1]) \wedge \text{dp}[i][j-1])$ |
| literal mismatch | false |

**Why it works.**
- *Optimal substructure (existence):* every successful regex match decomposes by what the last pattern token does. For `c*`, that token can absorb zero or more chars, each absorption being a one-step reduction.
- The `dp[i-2]` reach (skipping two pattern chars) is what makes regex distinctive — the pattern has *units* of size 1 or 2 (`c` alone, or `c*`).

**Complexity.** Time $O(nm)$. Space $O(m)$ achievable with **three rows** (need rows $i-2$, $i-1$, $i$) — see code.

**Delta vs Wildcard.**
- `*` binds to the preceding character → look at $p[i-2]$ in the recurrence.
- "Zero occurrences" branch jumps two pattern positions ($\text{dp}[i-2][j]$), not one ($\text{dp}[i-1][j]$).
- "One or more occurrences" requires the matched character to actually match $s[j-1]$ — extra guard.
- Standalone `*` is illegal as a pattern in regex (must follow a char); our code assumes well-formed input.

> Regex is the most intricate pattern-matching DP in standard interview catalogs. Internalize the two sub-cases for `*` (zero vs one-more) and the dp[i-2] jump; the rest follows mechanically.

---

````c++
// Regex Matching — three implementations
#include <vector>
#include <string>
#include <iostream>
using namespace std;

// Helper: does pattern char p_char match text char s_char (treating '.' as wildcard)?
inline bool matchOne(char p_char, char s_char) {
    return p_char == s_char || p_char == '.';
}

// (A) Top-down memoization
int regex_memo_helper(int i, int j, const string& p, const string& s, vector<vector<int>>& dp) {
    if (i == 0) return (j == 0) ? 1 : 0;                   // empty pattern matches only empty text
    if (j == 0) {
        // Empty text: pattern matches iff it's entirely c*c*c*...
        if (p[i-1] == '*') return regex_memo_helper(i-2, 0, p, s, dp);
        return 0;
    }
    if (dp[i][j] != -1) return dp[i][j];
    int result;
    if (p[i-1] == '*') {
        // Case B: 'c*' as a unit. c is p[i-2].
        int zeroOccurrences = regex_memo_helper(i-2, j, p, s, dp);  // skip the c* pair
        int oneOrMore = 0;
        if (i >= 2 && matchOne(p[i-2], s[j-1])) {
            // c matches s[j-1]; consume that one char of s, c* remains
            oneOrMore = regex_memo_helper(i, j-1, p, s, dp);
        }
        result = (zeroOccurrences || oneOrMore) ? 1 : 0;
    } else if (matchOne(p[i-1], s[j-1])) {
        // Case A: literal or '.' match
        result = regex_memo_helper(i-1, j-1, p, s, dp);
    } else {
        // Case C: literal mismatch
        result = 0;
    }
    dp[i][j] = result;
    return dp[i][j];
}
bool regex_memo(const string& s, const string& p) {
    int n = (int)p.size(), m = (int)s.size();
    vector<vector<int>> dp(n + 1, vector<int>(m + 1, -1));
    return regex_memo_helper(n, m, p, s, dp) == 1;
}

// (B) Tabulation — O(nm)
bool regex_tab(const string& s, const string& p) {
    int n = (int)p.size(), m = (int)s.size();
    vector<vector<bool>> dp(n + 1, vector<bool>(m + 1, false));
    dp[0][0] = true;                                       // base
    // Empty text column: dp[i][0] = (p[i-1] == '*' && dp[i-2][0])
    for (int i = 2; i <= n; ++i)                            // start at i=2; dp[1][0] is necessarily false (no preceding char for '*')
        if (p[i-1] == '*') dp[i][0] = dp[i-2][0];
    for (int i = 1; i <= n; ++i) {
        for (int j = 1; j <= m; ++j) {
            if (p[i-1] == '*') {
                bool zero = (i >= 2) ? dp[i-2][j] : false;
                bool oneOrMore = (i >= 2 && matchOne(p[i-2], s[j-1])) ? dp[i][j-1] : false;
                dp[i][j] = zero || oneOrMore;
            } else if (matchOne(p[i-1], s[j-1])) {
                dp[i][j] = dp[i-1][j-1];
            }
            // else: dp[i][j] stays false
        }
    }
    return dp[n][m];
}

// (C) Space-optimized — O(m) using THREE rows (need rows i-2, i-1, i)
//     The '*' branch reads dp[i-2][j], so we cannot use just two rows.
bool regex_opt(const string& s, const string& p) {
    int n = (int)p.size(), m = (int)s.size();
    vector<vector<bool>> dp(3, vector<bool>(m + 1, false));
    // We'll cycle row indices: dp[(i) % 3], dp[(i-1) % 3], dp[(i-2) % 3].
    dp[0][0] = true;                                       // represents dp[0][.]; ring at row index 0
    for (int i = 1; i <= n; ++i) {
        int curIdx  = i % 3;
        int prev1Idx = (i - 1) % 3;
        int prev2Idx = (i - 2) % 3;                        // valid for i >= 2; for i=1, never read since '*' needs i>=2
        // Reset cur row before populating
        for (int j = 0; j <= m; ++j) dp[curIdx][j] = false;
        // Empty-text column entry
        if (i >= 2 && p[i-1] == '*') dp[curIdx][0] = dp[prev2Idx][0];
        for (int j = 1; j <= m; ++j) {
            if (p[i-1] == '*') {
                bool zero = (i >= 2) ? dp[prev2Idx][j] : false;
                bool oneOrMore = (i >= 2 && matchOne(p[i-2], s[j-1])) ? dp[curIdx][j-1] : false;
                dp[curIdx][j] = zero || oneOrMore;
            } else if (matchOne(p[i-1], s[j-1])) {
                dp[curIdx][j] = dp[prev1Idx][j-1];
            }
            // else stays false
        }
    }
    return dp[n % 3][m];
}
````

````c++
// Tests — Regex Matching
auto run_re = [](string s, string p, bool expected) {
    bool a = regex_memo(s, p);
    bool b = regex_tab(s, p);
    bool c = regex_opt(s, p);
    bool ok = (a == expected && b == expected && c == expected);
    cout << "s=\"" << s << "\" p=\"" << p << "\""
         << " -> memo=" << a << " tab=" << b << " opt=" << c
         << " (Expected: " << expected << ")"
         << (ok ? " OK" : " FAIL") << '\n';
};

run_re("",        "",        true);
run_re("",        "a*",      true);    // a* = zero a's
run_re("",        "a*b*c*",  true);    // all zero occurrences
run_re("",        "a",       false);
run_re("a",       "a",       true);
run_re("a",       ".",       true);    // '.' matches any single
run_re("aa",      "a",       false);   // pattern too short
run_re("aa",      "a*",      true);    // a* = two a's
run_re("ab",      ".*",      true);    // .* = any sequence
run_re("aab",     "c*a*b",   true);    // classic LeetCode: c* = empty, a* = aa, b = b
run_re("mississippi", "mis*is*p*.",  false);  // LeetCode: doesn't match
run_re("mississippi", "mis*is*ip*.", true);
run_re("aaa",     "a*a",     true);
run_re("aaa",     "ab*a*c*a", true);   // a, b*=empty, a*=a, c*=empty, a
run_re("aaa",     "ab*ac*a", true);
run_re("ab",      ".*c",     false);
````

# Unified Mental Model — DP on Strings

All ten problems are realizations of one skeleton — a 2D DP on prefixes with a match/mismatch case-split. The variation lies in:

1. **Match-case transform** — pure inheritance, `+ 1`, append-char.
2. **Mismatch-case branches** — how many predecessors are combined, and what combine operator.
3. **Special characters** — wildcards or regex `*` add extra branches that read non-standard cells (`dp[i-2][j]` for regex).
4. **Post-processing** — pure $\text{dp}[n][m]$ read, or scan whole table (Common Substring), or apply closed-form formula (LPS-derivatives).

---

````c++
// ============================================================
// THE UNIFIED 2D-STRING-DP SKELETON
// ============================================================
//
//   dp[i][j] = answer for s[0..i-1] and t[0..j-1]    (prefix-length indexing)
//
//   case-split on last char of each prefix:
//     match  (s[i-1] == t[j-1])  → transform(dp[i-1][j-1])
//     mismatch                    → combine over predecessor cells
//
// ============================================================
// INSTANTIATIONS  (combine in shorthand: max/min/||;  pred cells in {NW, N, W, NN, ...})
// ============================================================
//
// LCS                       combine=max,  match=+1·NW,  mismatch picks {N, W}
// Print LCS                 same DP as LCS, then backtrack the table
// Longest Common Substring  combine=max(running),  match=+1·NW,  mismatch=0 (HARD RESET)
//                                                                ^ answer = max over ENTIRE table
//
// LPS                       = LCS(s, reverse(s))                                          (reduction)
// Min Insertions Palindrome = n - LPS(s)                                                  (formula)
// Min Ins/Del A→B           = (n - LCS(A,B)) + (m - LCS(A,B)) = n + m - 2·LCS(A,B)        (formula)
// SCS length                = n + m - LCS(A,B)                                            (formula)
// SCS reconstruction        = backtrack LCS table emitting EVERY cell visited
//
// Edit Distance             combine=min,  match=NW (cost 0),  mismatch=1+min{NW, N, W}    (three ops)
//
// Wildcard                  combine=||,
//                           match (literal=lit, or '?' wildcard): NW
//                           '*' = N (consume pattern char) OR W (consume text char)
//                           literal mismatch: false
//
// Regex                     combine=||,
//                           match (literal=lit, or '.'): NW
//                           '*' (binds to preceding char c = p[i-2]):
//                              NN  (zero c's: skip c*)
//                              W   (one more c, only if c matches s[j-1])
//                           literal mismatch: false
//
// ============================================================
// SPACE-OPTIMIZATION TABLE
// ============================================================
//   LCS, LCSubstr, LPS, Min Ins/Del, Edit Distance, Wildcard:  O(min(n,m)) — TWO rows
//   Regex:                                                     O(m)        — THREE rows (dp[i-2] is read)
//   Print LCS, SCS reconstruction:                             O(nm)        — full table required for backtrack
// ============================================================
````

# Decision Tree — recognizing string DP

```
                ┌──────────────────────────────────────────────────────────┐
                │ Input is one or two strings (or string + pattern)?       │
                └────────────────────┬─────────────────────────────────────┘
                                     │ yes
                                     ▼
                ┌──────────────────────────────────────────────────────────┐
                │ State = (prefix-length of A, prefix-length of B)?         │
                │  → 2D DP on prefixes, recurrence at the last char of each │
                └────────────────────┬─────────────────────────────────────┘
                                     ▼
                ┌──────────────────────────────────────────────────────────┐
                │ Is the answer about CONTIGUOUS substrings of A and B?     │
                │  • Yes → state is "ending exactly here"; mismatch=0;     │
                │          answer = max over full table  (Common Substring)│
                │  • No  → state is "any subsequence"; standard recurrence │
                └────────────────────┬─────────────────────────────────────┘
                                     ▼
                ┌──────────────────────────────────────────────────────────┐
                │ Question flavor (trichotomy):                            │
                │  • Max len / count / value             → max combine     │
                │  • Min edits / insertions / deletions  → min combine     │
                │  • Match / not match                   → OR combine      │
                └────────────────────┬─────────────────────────────────────┘
                                     ▼
                ┌──────────────────────────────────────────────────────────┐
                │ Mismatch-branch arity:                                   │
                │  • 2 branches (LCS-shaped)        → drop from one or other│
                │  • 3 branches (Edit Distance)     → insert/delete/replace│
                │  • Specific predecessors          → pattern matching:    │
                │     wildcard '*' or regex 'c*'                           │
                └────────────────────┬─────────────────────────────────────┘
                                     ▼
                ┌──────────────────────────────────────────────────────────┐
                │ Need to RECONSTRUCT the solution?                        │
                │  • Yes (Print LCS, SCS string): keep full O(nm) table;    │
                │        backtrack from (n, m) emitting along winning edges │
                │  • No (length only): O(min(n,m)) space with 2 rows       │
                │        (or 3 rows if recurrence reads dp[i-2] like regex)│
                └──────────────────────────────────────────────────────────┘
```

**Three traps to avoid:**

1. **LCS vs Longest Common Substring.** Both have $(i, j)$ state, both read NW on match. But Common Substring's mismatch is a **hard 0**, not "max of neighbors," and the answer is the **max over the whole table**, not the last cell. Confusing them is the most common error.

2. **Wildcard `*` vs Regex `*`.** Wildcard `*` is *standalone* and matches any sequence. Regex `*` is a *modifier* of the preceding character. The recurrences differ on the "zero occurrence" branch: wildcard jumps one pattern position ($\text{dp}[i-1][j]$), regex jumps two ($\text{dp}[i-2][j]$).

3. **Asymmetry of Edit Distance branches.** "Delete" reduces $s$, "Insert" reduces $t$. Many students conflate them or get the recurrence backwards. Anchoring on **"what does the last edit do?"** keeps them straight: delete consumes the last char of $s$; insert produces the last char of $t$; replace does both.

---

# Complexity Summary

| Problem | Family | Time | Naive Space | Optimized Space | Key Insight |
|---|---|---|---|---|---|
| **LCS** | A | $O(nm)$ | $O(nm)$ | $O(\min(n,m))$ | Kernel DP: match=NW+1, mismatch=max(N, W) |
| **Print LCS** | A | $O(nm)$ | $O(nm)$ | $O(nm)$ (forced) | Backtrack the LCS table from $(n,m)$ |
| **Longest Common Substring** | A | $O(nm)$ | $O(nm)$ | $O(\min(n,m))$ | "Ends exactly here"; mismatch=0; answer=max over table |
| **LPS** | A | $O(n^2)$ | $O(n^2)$ | $O(n)$ | $\text{LPS}(s) = \text{LCS}(s, \text{rev}(s))$ |
| **Min Ins. to Palindrome** | A | $O(n^2)$ | $O(n^2)$ | $O(n)$ | $n - \text{LPS}(s)$ |
| **Min Ins/Del A→B** | A | $O(nm)$ | $O(nm)$ | $O(\min(n,m))$ | $n + m - 2 \cdot \text{LCS}$ |
| **SCS** | A | $O(nm)$ | $O(nm)$ | $O(nm)$ (for string) | Length = $n + m - \text{LCS}$; reconstruct by full-path backtrack |
| **Edit Distance** | B | $O(nm)$ | $O(nm)$ | $O(\min(n,m))$ | Three mismatch branches: replace/delete/insert |
| **Wildcard Matching** | C | $O(nm)$ | $O(nm)$ | $O(m)$ | `*` = N (empty) ∨ W (one more), 2 rows enough |
| **Regex Matching** | C | $O(nm)$ | $O(nm)$ | $O(m)$ (3 rows) | `c*` reads $\text{dp}[i-2][j]$ for "zero occurrences" |

---

# Closing Notes

**What you should now be able to do without reaching for any reference:**

1. State the recurrence for LCS in under five seconds; everything in Family A follows by mechanical reduction or a one-character twist.
2. Recognize the **contiguity invariant** — "ends exactly here" — as the distinguishing feature of substring problems versus subsequence problems.
3. Write Edit Distance's three-branch recurrence and explain *which physical edit* each branch represents (delete $\leftrightarrow$ $\text{dp}[i-1][j]$, insert $\leftrightarrow$ $\text{dp}[i][j-1]$, replace $\leftrightarrow$ $\text{dp}[i-1][j-1] + 1$).
4. Handle the special-character cases for wildcard and regex without re-deriving — `*` in wildcard is a free wildcard; `*` in regex modifies the preceding character.
5. Decide whether you need full $O(nm)$ table (for reconstruction) or can drop to $O(\min(n,m))$ (length only).

**A pattern observed across this subtopic:** five of the ten problems (Print LCS, LPS, Min Ins. Palindrome, Min Ins/Del, SCS) are not new DPs at all — they are **reductions** to or backtracks over the LCS table. This pattern is central to interview velocity: when you see a string DP, your first move should be to ask "**can this reduce to LCS, or be expressed as $n - \text{(something)}$?**"

**A useful unifying perspective:** every problem in this subtopic is computing a **dissimilarity measure** or **alignment** of two sequences. LCS measures similarity (more shared = more aligned); Edit Distance measures dissimilarity (fewer edits = more aligned). The same DP table can answer both — they are dual views of the same alignment lattice. Bioinformatics calls this "sequence alignment" and uses the same DP machinery with a richer cost model (Needleman–Wunsch and Smith–Waterman are direct extensions of LCS and Common Substring respectively).

**Looking ahead — Subtopic 6 (DP on Stocks).** A new state dimension joins: **a flag for "currently holding stock or not."** The recurrence becomes a state machine in the holding-flag dimension. This is the cleanest example yet of when adding a state dimension is necessary because the problem cannot be Markovian in just $(i, j)$ — the past matters in a discrete way captured by the flag.
