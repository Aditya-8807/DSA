## 9.1 MST Theory — Cut Property and Cycle Property

### State definition
Given a connected, undirected, weighted graph $G=(V,E,w)$, a **Minimum Spanning Tree (MST)** is a
subset $T \subseteq E$ such that $(V,T)$ is a tree (connected, acyclic, $|T|=|V|-1$) and
$\sum_{e \in T} w(e)$ is minimized among all such spanning trees.

### The cut property (formal statement and proof)

**Definition — cut:** a partition of $V$ into two nonempty disjoint sets $S$ and $V \setminus S$.
An edge **crosses** the cut if it has one endpoint in each side.

**Cut Property:** for any cut $(S, V\setminus S)$, if $e$ is a **strictly minimum-weight** edge
crossing that cut, then $e$ belongs to **some** MST of $G$.

*Proof (exchange argument):* let $T$ be any MST not containing $e = (u,v)$ with $u \in S$,
$v \in V \setminus S$. Since $T$ is a spanning tree, adding $e$ to $T$ creates exactly one cycle
(trees have exactly one path between any two vertices, so adding an edge between two tree-connected
vertices closes exactly one cycle). This cycle, since it contains $e$ which crosses the cut, must
contain at least one **other** edge $f$ that also crosses the cut (a cycle must cross any cut an
even number of times — it has to "come back"). Since $e$ is the strict minimum-weight edge crossing
the cut, $w(f) \ge w(e)$ (in fact $w(f) > w(e)$ if $e$ is strictly minimum and $f \neq e$, or
$w(f) \ge w(e)$ if ties are allowed). Remove $f$ and add $e$: the result $T' = T - f + e$ is still a
spanning tree (removing $f$ breaks the one cycle we just created, restoring acyclicity; the graph
remains connected since $f$ was part of a cycle, not a bridge), with total weight
$w(T') = w(T) - w(f) + w(e) \le w(T)$. Since $T$ was already minimum, $w(T') = w(T)$ (cannot be
strictly less), so $T'$ is also an MST, and it contains $e$. $\blacksquare$

### The cycle property (formal statement and proof)

**Cycle Property:** for any cycle $C$ in $G$, if $e$ is a **strictly maximum-weight** edge in $C$,
then $e$ belongs to **no** MST of $G$.

*Proof (exchange argument, dual to the cut property):* suppose for contradiction some MST $T$
contains $e$. Removing $e$ from $T$ splits $T$ into exactly two components $S$ and $V \setminus S$
(removing any tree edge disconnects the tree into exactly two pieces). Since $e$ is part of cycle
$C$, and $C$ is a cycle in $G$ (not necessarily in $T$), $C$ must contain at least one other edge
$f$ that also crosses the $(S, V\setminus S)$ cut (same even-crossing argument as above). Since $e$
is the strict maximum-weight edge of $C$, $w(f) < w(e)$. Swap: $T' = T - e + f$ is a spanning tree
(symmetric argument to the cut property proof) with $w(T') = w(T) - w(e) + w(f) < w(T)$ — strictly
less than $T$'s weight, contradicting that $T$ was already minimum. $\blacksquare$

### Why these two properties are the foundation of both MST algorithms
The cut property directly justifies **Prim's algorithm** (9.2): repeatedly picking the cheapest
edge crossing the cut between "tree so far" and "everything else" is always safe, by the cut
property applied to that exact cut at every step. The cycle property directly justifies
**Kruskal's algorithm** (9.4): processing edges in increasing weight order and skipping any edge
that would close a cycle is safe, because any skipped edge is (at the moment it's skipped) the
maximum-weight edge of the cycle it would have closed among edges considered so far, so the cycle
property guarantees it's excludable.

### Boundary transitions table

| Decision point | Condition | Consequence |
|---|---|---|
| Multiple edges tie for minimum across a cut | ties allowed by "some MST," not necessarily unique | any of the tied edges can be safely included — MST is not always unique |
| Graph is disconnected | no spanning tree exists at all | MST is undefined; compute a **minimum spanning forest** per component instead |
| Self-loops or multi-edges present | self-loops never useful (never part of any spanning tree, they can't help connectivity); multi-edges — only the minimum-weight parallel edge can ever be useful | filter out self-loops; keep only the cheapest of any parallel edges as a preprocessing step |

### The delta
The non-obvious insight: **the cut property and cycle property are two views of the same
exchange argument**, one "from the outside in" (justifying inclusion of cheap edges) and one "from
the inside out" (justifying exclusion of expensive edges within a cycle) — every correctness proof
for Prim's and Kruskal's in this subtopic reduces to one of these two properties, so understanding
both once eliminates the need to re-derive correctness separately for each algorithm.


````c++
// MST theory: a verification helper -- checks total weight and confirms tree structure (V-1 edges, connected, acyclic)
#include <iostream>
#include <vector>
using namespace std;
#define vi vector<int>
#define pii pair<int,int>

struct DSU {
    vi parent, rank_;
    DSU(int n) : parent(n), rank_(n, 0) { for (int i = 0; i < n; i++) parent[i] = i; }
    int find(int x) { if (parent[x] != x) parent[x] = find(parent[x]); return parent[x]; }
    bool unite(int a, int b) {
        int ra = find(a), rb = find(b);
        if (ra == rb) return false;              // already connected -- this edge would close a cycle
        if (rank_[ra] < rank_[rb]) swap(ra, rb);
        parent[rb] = ra;
        if (rank_[ra] == rank_[rb]) rank_[ra]++;
        return true;
    }
};

// isValidSpanningTree: confirms exactly V-1 edges, all vertices connected (single DSU component), no cycles
// (implicitly: if union() ever returns false for an edge in the supposed tree, it wasn't acyclic).
bool isValidSpanningTree(int V, vector<array<int,2>>& treeEdges) {
    if ((int)treeEdges.size() != V - 1) return false;   // structural requirement: exactly V-1 edges
    DSU dsu(V);
    for (auto& e : treeEdges) {
        if (!dsu.unite(e[0], e[1])) return false;          // union failed -- a cycle exists, not a valid tree
    }
    // confirm full connectivity: all V vertices in one component
    int root = dsu.find(0);
    for (int v = 1; v < V; v++) if (dsu.find(v) != root) return false;
    return true;
}
````

````c++
// Test cell: input -> actual (Expected: X)

// Valid spanning tree: path 0-1-2-3
{
    vector<array<int,2>> edges = {{0,1},{1,2},{2,3}};
    cout << "isValidSpanningTree(path, V=4) -> " << isValidSpanningTree(4, edges) << " (Expected: 1)\n";
}

// Invalid: has a cycle (too many edges implicitly caught by union failure, and count check)
{
    vector<array<int,2>> edges = {{0,1},{1,2},{2,0}};   // triangle, only 3 vertices but forms a cycle, V should be 3 not tree
    cout << "isValidSpanningTree(cycle, V=3) -> " << isValidSpanningTree(3, edges) << " (Expected: 0, only 3 edges != V-1=2)\n";
}

// Invalid: disconnected (two separate components, not a single spanning tree)
{
    vector<array<int,2>> edges = {{0,1},{2,3}};
    cout << "isValidSpanningTree(disconnected, V=4) -> " << isValidSpanningTree(4, edges) << " (Expected: 0)\n";
}

// Single node, no edges -- trivially a valid "tree"
{
    vector<array<int,2>> edges = {};
    cout << "isValidSpanningTree(single node, V=1) -> " << isValidSpanningTree(1, edges) << " (Expected: 1)\n";
}

// Star spanning tree
{
    vector<array<int,2>> edges = {{0,1},{0,2},{0,3},{0,4}};
    cout << "isValidSpanningTree(star, V=5) -> " << isValidSpanningTree(5, edges) << " (Expected: 1)\n";
}
````

## 9.2 Prim's Algorithm — Grow One Tree

### State definition
$\text{inMST}[v] \in \{\text{false}, \text{true}\}$ — whether $v$ has been added to the growing
tree. $\text{key}[v]$ = the minimum edge weight connecting $v$ to the current tree (analogous to
Dijkstra's `dist[v]`, but representing "cheapest connecting edge," not "cheapest path from source").

### Invariant — "the current tree is contained in some MST"
At every point during execution, the set of edges chosen so far forms a tree $T_{\text{partial}}$
that is a **subset of some MST** of $G$. This invariant is maintained by construction: at each
step, the algorithm adds the cheapest edge crossing the cut between $\{v : \text{inMST}[v]\}$ and
the rest — which the **cut property (9.1)** guarantees is always safe to add without ever leaving
the "subset of some MST" invariant.

### Why it works — direct application of the cut property

**Claim:** at every step, adding the minimum-weight edge crossing the current cut
$(S, V\setminus S)$ where $S = \{v : \text{inMST}[v]\}$ preserves the invariant.

*Proof:* by the cut property (9.1), the minimum-weight edge crossing any cut belongs to *some* MST.
Since $T_{\text{partial}}$ (by the inductive invariant) is already a subset of some MST $M$, and the
cut $(S, V\setminus S)$ is exactly the cut induced by that partial tree's vertex set, the
minimum-weight crossing edge $e$ either is already in $M$ (trivially safe to add, stays a subset of
$M$) or — if $M$ doesn't contain $e$ — the same exchange argument from 9.1's cut-property proof
constructs a *different* MST $M'$ that does contain $e$ **and** still contains all of
$T_{\text{partial}}$ (since the swap only removes an edge crossing the same cut, which
$T_{\text{partial}}$'s edges, being entirely within $S$ or entirely within $V \setminus S$ except
for tree edges already accounted for, does not disturb). So $T_{\text{partial}} \cup \{e\}$ is a
subset of $M'$, maintaining the invariant. By induction from the empty tree (trivially a subset of
any MST) to the full spanning tree, the final result is a complete MST. $\blacksquare$

### Boundary transitions table

| Decision point | Condition | Action |
|---|---|---|
| Vertex not yet in tree, edge to it improves `key` | $w(u,v) < \text{key}[v]$, $u \in$ tree | update `key[v] = w(u,v)`, push to priority queue |
| Vertex already in tree | `inMST[v] == true` | skip — no longer part of any future cut |
| Graph disconnected | some vertex never gets a finite `key` | Prim's from a single start vertex only finds the MST of that vertex's component — must restart per component for a full minimum spanning forest |

### The delta
The non-obvious insight: Prim's algorithm is **structurally identical to Dijkstra** (Subtopic
8.3) — same priority-queue-driven "extract minimum, relax neighbors" loop — but the *meaning* of
the value being minimized is different: Dijkstra's `dist[v]` accumulates total path cost from a
fixed source, while Prim's `key[v]` is memoryless, always just "the cheapest single edge connecting
$v$ to the tree *right now*," reset independently at each vertex rather than summed along a path.
Confusing these two (e.g. accidentally accumulating cost like Dijkstra) silently computes the wrong
thing while looking like syntactically valid Prim's code.


````c++
// Prim's Algorithm: grow one tree, cut-property-justified greedy edge selection
#include <iostream>
#include <vector>
#include <queue>
#include <climits>
using namespace std;
#define vi vector<int>
#define vb vector<bool>
#define pii pair<int,int>
#define vpii vector<pair<int,int>>
#define vvpii vector<vector<pii>>

// primMST: returns total MST weight. Priority queue keyed by (edge weight to tree, vertex).
long long primMST(int V, vvpii& adj) {
    vb inMST(V, false);
    priority_queue<pii, vpii, greater<pii>> pq;   // (weight, vertex)
    pq.push({0, 0});           // start from vertex 0, key[0] = 0 (free to "connect" the first vertex)
    long long totalWeight = 0;

    while (!pq.empty()) {
        auto [w, u] = pq.top(); pq.pop();
        if (inMST[u]) continue;         // stale entry -- u already added via a cheaper edge, skip
        inMST[u] = true;                // u joins the tree -- key[u] (this edge's weight) is FINAL, by the cut property
        totalWeight += w;
        for (auto& [v, ew] : adj[u]) {
            if (!inMST[v]) pq.push({ew, v});   // candidate edge crossing the NEW cut -- no need to compare to old key,
                                                 // stale/worse entries are simply skipped later via the inMST check
        }
    }
    return totalWeight;
}
````

````c++
// Test cell: input -> actual (Expected: X)

// Standard case: square with a diagonal
{
    vvpii adj(4);
    auto addEdge = [&](int u, int v, int w){ adj[u].push_back({v,w}); adj[v].push_back({u,w}); };
    addEdge(0,1,1); addEdge(1,2,2); addEdge(2,3,3); addEdge(3,0,4); addEdge(0,2,5);
    cout << "primMST(square+diagonal) -> " << primMST(4, adj) << " (Expected: 6, edges 0-1,1-2,2-3)\n";
}

// Single node, no edges
{
    vvpii adj(1);
    cout << "primMST(single node) -> " << primMST(1, adj) << " (Expected: 0)\n";
}

// Already a tree (no choices to make)
{
    vvpii adj(3);
    auto addEdge = [&](int u, int v, int w){ adj[u].push_back({v,w}); adj[v].push_back({u,w}); };
    addEdge(0,1,5); addEdge(1,2,3);
    cout << "primMST(already a tree) -> " << primMST(3, adj) << " (Expected: 8)\n";
}

// Complete graph K4 with distinct weights
{
    vvpii adj(4);
    auto addEdge = [&](int u, int v, int w){ adj[u].push_back({v,w}); adj[v].push_back({u,w}); };
    addEdge(0,1,1); addEdge(0,2,2); addEdge(0,3,3); addEdge(1,2,4); addEdge(1,3,5); addEdge(2,3,6);
    cout << "primMST(K4) -> " << primMST(4, adj) << " (Expected: 6, star from 0: edges 0-1,0-2,0-3 costing 1+2+3)\n";
}
````

## 9.3 Disjoint Set Union (Union-Find)

### State definition
$\text{parent}[v]$ = the immediate parent of $v$ in a forest of trees representing disjoint sets;
$\text{parent}[v] = v$ means $v$ is a root (representative of its set). $\text{rank}[v]$ (or
`size[v]`) = an upper bound on the height of the subtree rooted at $v$ (used only when $v$ is a
root), used to decide merge direction.

### Invariant
Two elements $u, v$ are in the same set **iff** $\text{find}(u) = \text{find}(v)$, where
$\text{find}$ follows parent pointers to the root. This must hold after every `unite` call.

### Union by rank/size — why merge direction matters

**Without** union by rank/size, naive unite could always attach $b$'s root under $a$'s root
regardless of tree size — repeatedly doing this in the worst order (e.g. always merging a
size-1 tree under a growing chain) produces a **degenerate chain** of height $O(n)$, making `find`
cost $O(n)$ per call in the worst case.

**With union by rank** (attach the shorter tree under the taller one's root): the resulting tree's
height increases **only** when the two merged trees have **equal** rank (in which case the new
height is old height $+1$). This directly bounds tree height: a tree of rank $r$ must have been
built from merging two rank-$(r-1)$ trees, each of which required merging two rank-$(r-2)$ trees,
and so on — by induction, a tree of rank $r$ has **at least $2^r$ nodes**. Since the total number of
nodes is $n$, $2^r \le n \implies r \le \log_2 n$ — **tree height (and hence naive `find` cost) is
bounded by $O(\log n)$** even without path compression.

### Path compression — the second half of the optimization

During `find(x)`, after locating the root, **re-point every node visited along the way directly to
the root** (not just $x$ itself). This flattens the tree structure for all future queries touching
any of those nodes.

### Amortized $O(\alpha(n))$ derivation (sketch)

**Claim:** with **both** union by rank and path compression combined, a sequence of $m$
`find`/`unite` operations on $n$ elements takes $O(m \cdot \alpha(n))$ total time, where $\alpha$ is
the inverse Ackermann function — for all practically conceivable $n$ (even $n$ far larger than the
number of atoms in the observable universe), $\alpha(n) \le 4$, making each operation *effectively*
$O(1)$ amortized.

*Proof sketch (full proof is a substantial potential-function argument, standard result, sketched
here):* union by rank alone bounds tree height at $O(\log n)$, giving $O(\log n)$ per `find`. Path
compression alone, without rank, still gives an amortized $O(\log n)$ bound via a different
argument (each compression roughly halves the "distance to root" for nodes involved, similar in
spirit to the halving argument in binary counters). **Combined**, the two optimizations interact
superadditively: path compression performed *within* rank-bounded trees means each node can only be
moved to a shallower position a limited number of times before it's essentially attached directly
under a root of much higher rank — this is formalized via a potential function that charges each
`find` operation against a slowly-growing rank-based "level" structure, and the standard result
(Tarjan) shows the total charged cost across $m$ operations is $O((m+n)\alpha(n))$. The full proof
is beyond what's reproduced here, but the practical takeaway is unconditional: **always implement
both optimizations together** — the combination is what achieves the near-constant amortized bound,
neither alone is sufficient for the tightest guarantee.

### Boundary transitions table

| Decision point | Condition | Action |
|---|---|---|
| `find(x)` where $x$ is a root | $\text{parent}[x] = x$ | return $x$ directly, base case |
| `find(x)` where $x$ is not a root | $\text{parent}[x] \neq x$ | recurse to find the true root, then re-point $\text{parent}[x]$ directly to it (path compression) |
| `unite(a,b)` where $a, b$ already in the same set | $\text{find}(a) = \text{find}(b)$ | no-op — union would be redundant (this is exactly the "edge closes a cycle" signal used in Kruskal's, 9.4) |
| `unite(a,b)`, different sets, equal rank | $\text{rank}[\text{find}(a)] = \text{rank}[\text{find}(b)]$ | attach either under the other, increment the new root's rank by 1 |
| `unite(a,b)`, different sets, unequal rank | — | attach the lower-rank root under the higher-rank root, rank unchanged |

### The delta
The non-obvious insight, sharpening Subtopic 1.4's earlier DSU introduction: the near-$O(1)$
amortized claim is **not** free from either optimization alone — union by rank alone gives a solid
but not optimal $O(\log n)$, and path compression alone (without balancing merge direction) can
still degrade in adversarial sequences. The combination is a genuinely emergent property, which is
why virtually every production DSU implementation includes both unconditionally, rather than
treating them as independent, separately-optional optimizations.


````c++
// Disjoint Set Union: union by rank + path compression, the combination needed for near-O(1) amortized ops
#include <iostream>
#include <vector>
using namespace std;
#define vi vector<int>

struct DSU {
    vi parent, rank_;
    int components;          // bonus tracked field: number of distinct sets remaining

    DSU(int n) : parent(n), rank_(n, 0), components(n) {
        for (int i = 0; i < n; i++) parent[i] = i;   // base case: every element is its own root initially
    }

    // find: path compression -- every node visited on the way up gets re-pointed directly to the root.
    int find(int x) {
        if (parent[x] != x) parent[x] = find(parent[x]);   // recursive compression
        return parent[x];
    }

    // unite: union by rank -- attach the shorter tree under the taller one's root.
    bool unite(int a, int b) {
        int ra = find(a), rb = find(b);
        if (ra == rb) return false;              // already in the same set -- this "edge" would close a cycle
        if (rank_[ra] < rank_[rb]) swap(ra, rb);
        parent[rb] = ra;
        if (rank_[ra] == rank_[rb]) rank_[ra]++;   // height only grows on equal-rank merges (bounds height to O(log n))
        components--;
        return true;
    }

    bool connected(int a, int b) { return find(a) == find(b); }
};
````

````c++
// Test cell: input -> actual (Expected: X)

// Basic union and find
{
    DSU dsu(5);
    dsu.unite(0, 1);
    dsu.unite(1, 2);
    cout << "DSU connected(0,2) -> " << dsu.connected(0, 2) << " (Expected: 1)\n";
    cout << "DSU connected(0,3) -> " << dsu.connected(0, 3) << " (Expected: 0)\n";
}

// Component counting
{
    DSU dsu(6);
    dsu.unite(0,1); dsu.unite(2,3); dsu.unite(4,5);
    cout << "DSU components after 3 unions on 6 elements -> " << dsu.components << " (Expected: 3)\n";
}

// Redundant union (already connected) should not change component count
{
    DSU dsu(3);
    dsu.unite(0,1);
    bool result = dsu.unite(0,1);   // redundant
    cout << "DSU redundant unite returns -> " << result << " (Expected: 0, false)\n";
    cout << "DSU components after redundant union -> " << dsu.components << " (Expected: 2)\n";
}

// Single element, no unions
{
    DSU dsu(1);
    cout << "DSU single element components -> " << dsu.components << " (Expected: 1)\n";
}

// Chain of unions collapsing to one component
{
    DSU dsu(5);
    for (int i = 0; i < 4; i++) dsu.unite(i, i+1);
    cout << "DSU chain, components -> " << dsu.components << " (Expected: 1)\n";
    cout << "DSU chain, connected(0,4) -> " << dsu.connected(0, 4) << " (Expected: 1)\n";
}
````

## 9.4 Kruskal's Algorithm

### State definition
Sort all edges by weight ascending. Process each edge $(u,v,w)$ in order; include it in the MST
**iff** $u$ and $v$ are not already connected (checked via DSU).

### Invariant
At every point during processing, the set of included edges forms a **forest** (never a cycle),
and this forest is a subset of some MST — directly justified by the **cycle property (9.1)**: any
edge skipped because it would close a cycle is, at the moment it's examined, the maximum-weight
edge considered so far among the cycle it would form (since edges are processed in increasing
weight order), so the cycle property guarantees it is safely excludable from every MST.

### Why it works — direct application of the cycle property

**Claim:** processing edges in increasing weight order, including an edge iff it doesn't close a
cycle with already-included edges, produces an MST.

*Proof:* every included edge connects two previously-disconnected components — by the cut
property (9.1) applied to the cut between those two components at the moment this edge is the
cheapest ever considered connecting them (guaranteed by sorted order — no cheaper edge between these
two specific components could have been skipped, since all cheaper edges were already processed),
this edge is safe to include, exactly as in Prim's argument. Every **excluded** edge, at the moment
it's skipped (endpoints already connected via cheaper edges), would close a cycle in which it is —
by sorted processing order — the maximum-weight edge seen among that cycle's edges so far; the
cycle property directly excludes it from being necessary in any MST. Since the process terminates
with exactly $V-1$ included edges connecting all vertices (guaranteed by DSU's component-count
tracking, 9.3) forming a tree, and every inclusion/exclusion decision is individually justified by
the cut/cycle properties, the final result is a valid MST. $\blacksquare$

### Boundary transitions table

| Decision point | Condition | Action |
|---|---|---|
| Edge endpoints already connected (`unite` returns false) | `find(u) == find(v)` | skip — including it would close a cycle, cycle property says it's excludable |
| Edge endpoints in different components | `unite(u,v)` returns true | include the edge, add its weight to the MST total |
| MST edge count reaches $V-1$ | all vertices connected | can early-exit — no further edges needed (early termination optimization) |
| Graph disconnected | fewer than $V-1$ edges included after processing all edges | no full MST exists — result is a minimum spanning **forest** |

### The delta
The non-obvious insight, closing this subtopic's central theme: **Prim's and Kruskal's are the
same greedy correctness guarantee (safe-edge exchange arguments) applied via two different
traversal orders** — Prim's grows outward from one vertex, always taking the cheapest edge crossing
the current frontier (cut property, vertex-driven); Kruskal's processes globally by weight,
always taking the cheapest edge that doesn't create a cycle (cycle property, edge-driven). Neither
is "more correct" than the other; the choice between them is purely about data-structure fit
(Kruskal's needs a fast DSU and edge sorting — natural when working from an edge list, Subtopic
1.2; Prim's needs a fast priority queue over vertices — natural when working from an adjacency
list, and is preferable on dense graphs for the same reasons Dijkstra's naive array version can
beat its heap version, Subtopic 8.4).


````c++
// Kruskal's Algorithm: sort edges, union non-cycle-forming, using the DSU from 9.3
#include <iostream>
#include <vector>
#include <algorithm>
using namespace std;
#define vi vector<int>

struct DSUK {
    vi parent, rank_;
    DSUK(int n) : parent(n), rank_(n, 0) { for (int i = 0; i < n; i++) parent[i] = i; }
    int find(int x) { if (parent[x] != x) parent[x] = find(parent[x]); return parent[x]; }
    bool unite(int a, int b) {
        int ra = find(a), rb = find(b);
        if (ra == rb) return false;
        if (rank_[ra] < rank_[rb]) swap(ra, rb);
        parent[rb] = ra;
        if (rank_[ra] == rank_[rb]) rank_[ra]++;
        return true;
    }
};

struct WEdge { int u, v, w; };

long long kruskalMST(int V, vector<WEdge> edges) {
    sort(edges.begin(), edges.end(), [](const WEdge& a, const WEdge& b) { return a.w < b.w; });   // ascending weight
    DSUK dsu(V);
    long long totalWeight = 0;
    int edgesUsed = 0;

    for (auto& e : edges) {
        if (dsu.unite(e.u, e.v)) {          // true iff this edge does NOT close a cycle
            totalWeight += e.w;
            edgesUsed++;
            if (edgesUsed == V - 1) break;    // early exit: full spanning tree complete
        }
    }
    return totalWeight;   // caller can separately check edgesUsed == V-1 to confirm full connectivity
}
````

````c++
// Test cell: input -> actual (Expected: X)

// Standard case: matches the same graph used in Prim's 9.2 test -- must agree
{
    vector<WEdge> edges = {{0,1,1},{1,2,2},{2,3,3},{3,0,4},{0,2,5}};
    cout << "kruskalMST(square+diagonal) -> " << kruskalMST(4, edges) << " (Expected: 6, matches primMST from 9.2)\n";
}

// Single node, no edges
{
    vector<WEdge> edges = {};
    cout << "kruskalMST(single node) -> " << kruskalMST(1, edges) << " (Expected: 0)\n";
}

// Already a tree, no cycles to skip
{
    vector<WEdge> edges = {{0,1,5},{1,2,3}};
    cout << "kruskalMST(already a tree) -> " << kruskalMST(3, edges) << " (Expected: 8)\n";
}

// Complete graph K4 -- must agree with Prim's K4 result from 9.2
{
    vector<WEdge> edges = {{0,1,1},{0,2,2},{0,3,3},{1,2,4},{1,3,5},{2,3,6}};
    cout << "kruskalMST(K4) -> " << kruskalMST(4, edges) << " (Expected: 6, matches primMST from 9.2)\n";
}

// Graph with duplicate/parallel edges -- cheaper parallel edge should be used, more expensive one skipped as redundant
{
    vector<WEdge> edges = {{0,1,10},{0,1,2},{1,2,1}};
    cout << "kruskalMST(parallel edges) -> " << kruskalMST(3, edges) << " (Expected: 3, uses the cheap 0-1 edge (2) + 1-2 (1))\n";
}
````

### 9.5.1 Number of Operations to Make Network Connected (DSU Counting Components)

**State definition.** Given $n$ computers and a list of existing cables (edges), each "operation"
moves one cable to connect two currently-disconnected computers. Find the minimum number of
operations to make the whole network connected, or $-1$ if impossible.

**Invariant.** Direct application of Subtopic 1.4's DSU-based component counting: build a DSU over
all $n$ computers, union every existing cable's endpoints; the final `components` count tells you
exactly how many separate "islands" need to be stitched together.

**Why it works.** To connect $k$ disjoint components into one, you need **exactly $k-1$** extra
edges (this is the tree-edge-count fact restated: a spanning structure over $k$ components needs
$k-1$ connecting edges, regardless of which specific edges — any redundant/already-useless cable
can be repurposed as one of these $k-1$ "extra" edges). This is only possible if there are **at
least $k-1$ redundant cables available** — a cable is redundant iff its `unite()` call returned
false (both endpoints already connected before this cable was processed). Feasibility check:
$(\text{number of edges}) \ge n - 1$ (you need at least $n-1$ cables total to have any hope of
connecting $n$ computers, by the same tree-edge-count fact applied to the whole network).

**The delta.** This is Subtopic 1.4's plain "count components" DSU application, with one added
insight: **counting how many union() calls returned false directly counts available "spare" edges**
— no separate bookkeeping is needed, since DSU's own success/failure signal from each union call is
exactly the redundancy indicator required.


````c++
// Number of Operations to Make Network Connected: DSU component counting + redundant-edge feasibility check
#include <iostream>
#include <vector>
using namespace std;
#define vi vector<int>

struct DSUN {
    vi parent, rank_;
    int components;
    DSUN(int n) : parent(n), rank_(n,0), components(n) { for (int i=0;i<n;i++) parent[i]=i; }
    int find(int x) { if (parent[x]!=x) parent[x]=find(parent[x]); return parent[x]; }
    bool unite(int a, int b) {
        int ra=find(a), rb=find(b);
        if (ra==rb) return false;
        if (rank_[ra]<rank_[rb]) swap(ra,rb);
        parent[rb]=ra;
        if (rank_[ra]==rank_[rb]) rank_[ra]++;
        components--;
        return true;
    }
};

int makeConnected(int n, vector<vector<int>>& connections) {
    if ((int)connections.size() < n - 1) return -1;   // not enough cables to possibly connect n computers
    DSUN dsu(n);
    for (auto& c : connections) dsu.unite(c[0], c[1]);
    return dsu.components - 1;   // k components need exactly k-1 extra edges to merge into one
}
````

### 9.5.2 Most Stones Removed with Same Row or Column (Row-Column Union Trick)

**State definition.** Given stones at grid coordinates, a stone can be removed if it shares a row
or column with another **remaining** stone. Find the maximum number of stones removable.

**Invariant — the row-column union trick.** Model rows and columns as a **single shared
namespace**: union each stone's row-id and column-id together (e.g. row $r$ as node $r$, column
$c$ as node $\text{numRows} + c$ to keep them distinct). Two stones end up in the same DSU
component **iff** they are connected through a chain of shared rows/columns (stone $A$ shares a row
with stone $B$, which shares a column with stone $C$, etc.).

**Why it works.** **Claim:** within any connected component of stones (in the row/column-sharing
sense), all but **one** stone can always be removed. *Proof:* a connected component with $k$ stones
spans some set of rows and columns; since it is connected (by definition of the DSU component), you
can always remove stones **one at a time in an order that keeps the remaining set connected** —
concretely, repeatedly remove any stone that is not a "cut vertex" of the row-column bipartite
connectivity graph of remaining stones (such a stone always exists in any connected graph with
$\ge 2$ nodes — any spanning-tree-like structure has a leaf). This continues until exactly one stone
remains, which trivially cannot be removed further (needs *another* stone sharing its row/column).
Hence the maximum removable count is $(\text{total stones}) - (\text{number of connected
components})$ — exactly the same formula shape as 9.5.1's "$k$ components need $k-1$ connections,"
here inverted to "$k$ stones in a component leave exactly 1 behind."

**The delta.** This is the least "obviously DSU" problem so far: the graph being unioned is **not**
the stones directly, but an auxiliary bipartite structure (rows and columns as the actual DSU
elements, with stones only implicitly connecting them) — recognizing that rows/columns, not stones,
should be the union-find elements is the entire trick.


````c++
// Most Stones Removed: union rows and columns directly (not the stones), count components
#include <iostream>
#include <vector>
#include <unordered_map>
using namespace std;
#define vi vector<int>

struct DSUS {
    unordered_map<int,int> parent;
    int find(int x) {
        if (!parent.count(x)) parent[x] = x;         // lazily initialize -- row/col ids are sparse
        if (parent[x] != x) parent[x] = find(parent[x]);
        return parent[x];
    }
    void unite(int a, int b) {
        int ra = find(a), rb = find(b);
        if (ra != rb) parent[ra] = rb;
    }
};

int removeStones(vector<vector<int>>& stones) {
    DSUS dsu;
    for (auto& s : stones) {
        int row = s[0];
        int col = ~s[1];       // bitwise complement: guarantees columns never collide with row ids (rows are >= 0, ~col < 0)
        dsu.unite(row, col);    // union the ROW and COLUMN directly -- stones are never DSU elements themselves
    }
    unordered_map<int,bool> seenRoot;
    for (auto& s : stones) seenRoot[dsu.find(s[0])] = true;
    int components = seenRoot.size();
    return (int)stones.size() - components;   // total stones - components = max removable
}
````

````c++
// Test cell: input -> actual (Expected: X)

{
    vector<vector<int>> stones = {{0,0},{0,1},{1,0},{1,2},{2,1},{2,2}};
    cout << "removeStones(standard) -> " << removeStones(stones) << " (Expected: 5)\n";
}
{
    vector<vector<int>> stones = {{0,0},{0,2},{1,1},{2,0},{2,2}};
    cout << "removeStones(two components) -> " << removeStones(stones) << " (Expected: 3)\n";
}
{
    vector<vector<int>> stones = {{0,0}};
    cout << "removeStones(single stone) -> " << removeStones(stones) << " (Expected: 0)\n";
}
{
    vector<vector<int>> stones = {{0,0},{1,1},{2,2}};   // no shared rows/cols at all -- 3 isolated components
    cout << "removeStones(all isolated) -> " << removeStones(stones) << " (Expected: 0)\n";
}
````

### 9.5.3 Accounts Merge (Email Union Across Accounts)

**State definition.** Given accounts, each a name plus a list of emails, merge accounts that share
**any** email into one, and return each merged account's name plus its sorted, deduplicated emails.

**Invariant.** Union every pair of emails that appear **together in the same account's email
list**. Two accounts should be merged **iff** their emails end up in the same DSU component — this
correctly handles transitive merging (account A shares an email with B, B shares a different email
with C $\Rightarrow$ A, B, C all merge), which a naive pairwise-only comparison would miss.

**Why it works.** Direct instantiation of Subtopic 1.4's component-counting/grouping logic, with
**emails as the DSU elements** (not accounts) — this mirrors 9.5.2's insight that the DSU element
set is not always the "obvious" object in the problem statement. Once components are found, group
all emails by their DSU root, and any email in a group determines which account's name to attach
(since all emails from the same original account are, by construction, unioned together as part of
processing that account).

**The delta.** This is a second instance (after 9.5.2) of "the DSU elements are not the top-level
objects named in the problem" — here it's individual email strings, requiring a string-to-index
mapping as a preprocessing step, exactly analogous to how Word Ladder (Subtopic 6) needed a
dictionary/bucket structure before any traversal could begin.


````c++
// Accounts Merge: union emails (not accounts) that co-occur, then group by DSU root
#include <iostream>
#include <vector>
#include <string>
#include <unordered_map>
#include <algorithm>
using namespace std;
#define vi vector<int>

struct DSUE {
    vi parent;
    DSUE(int n) : parent(n) { for (int i=0;i<n;i++) parent[i]=i; }
    int find(int x) { if (parent[x]!=x) parent[x]=find(parent[x]); return parent[x]; }
    void unite(int a, int b) { int ra=find(a), rb=find(b); if (ra!=rb) parent[ra]=rb; }
};

vector<vector<string>> accountsMerge(vector<vector<string>>& accounts) {
    unordered_map<string,int> emailToId;
    unordered_map<string,string> emailToName;
    int idCounter = 0;

    // Preprocessing: map every distinct email to an integer id (emails ARE the DSU elements).
    for (auto& acc : accounts) {
        string name = acc[0];
        for (int i = 1; i < (int)acc.size(); i++) {
            if (!emailToId.count(acc[i])) emailToId[acc[i]] = idCounter++;
            emailToName[acc[i]] = name;
        }
    }

    DSUE dsu(idCounter);
    for (auto& acc : accounts) {
        for (int i = 2; i < (int)acc.size(); i++) {
            dsu.unite(emailToId[acc[1]], emailToId[acc[i]]);   // union every email in this account together
        }
    }

    unordered_map<int, vector<string>> groups;
    for (auto& [email, id] : emailToId) groups[dsu.find(id)].push_back(email);

    vector<vector<string>> result;
    for (auto& [root, emails] : groups) {
        vector<string> sortedEmails = emails;
        sort(sortedEmails.begin(), sortedEmails.end());
        vector<string> entry;
        entry.push_back(emailToName[sortedEmails[0]]);
        for (auto& e : sortedEmails) entry.push_back(e);
        result.push_back(entry);
    }
    return result;
}
````

````c++
// Test cell: input -> actual (Expected: X)

// Standard case: two John accounts share an email, transitively merging
{
    vector<vector<string>> accounts = {
        {"John","johnsmith@mail.com","john_newyork@mail.com"},
        {"John","johnsmith@mail.com","john00@mail.com"},
        {"Mary","mary@mail.com"},
        {"John","johnnybravo@mail.com"}
    };
    auto result = accountsMerge(accounts);
    cout << "accountsMerge(standard) #merged accounts -> " << result.size() << " (Expected: 3)\n";
}

// No shared emails at all -- everything stays separate
{
    vector<vector<string>> accounts = {{"A","a@mail.com"},{"B","b@mail.com"}};
    auto result = accountsMerge(accounts);
    cout << "accountsMerge(no overlap) #accounts -> " << result.size() << " (Expected: 2)\n";
}

// Single account
{
    vector<vector<string>> accounts = {{"A","a1@mail.com","a2@mail.com"}};
    auto result = accountsMerge(accounts);
    cout << "accountsMerge(single account) #accounts -> " << result.size() << " (Expected: 1)\n";
    cout << "  #emails in it -> " << (result.empty() ? 0 : (int)result[0].size()-1) << " (Expected: 2)\n";
}
````

### 9.5.4 Number of Islands II (Dynamic Connectivity With DSU)

**State definition.** Cells are added to a grid one at a time (land conversions), in a given
sequence. After **each** addition, report the current number of islands (4-connected land
components).

**Invariant — why this needs DSU, not repeated BFS/DFS.** Subtopic 3.2's flood-fill component
counting is a **static, one-shot** algorithm — re-running it after every single addition would cost
$O(k \cdot R \cdot C)$ for $k$ additions, exactly the same asymptotic trap proven suboptimal in
Subtopic 4.2's deep dive (repeated full passes instead of incrementally maintained state). DSU is
built exactly for this: Subtopic 1.4's boundary table already flagged "edges arrive incrementally,
need repeated connectivity queries" as the DSU-favoring case — this problem is precisely that case,
applied to a grid.

**Why it works.** Maintain a running island count. When a new land cell is added: increment the
count by 1 (it starts as its own new island), then for each of its 4-directional land neighbors
already present, `unite()` — each **successful** union (merging two previously-separate components)
decrements the count by exactly 1 (two islands becoming one). This is the same "union call's
success/failure signals a real merge" pattern used in 9.5.1.

**The delta.** This is the clearest demonstration in this subtopic of *why* DSU exists as a
structure distinct from BFS/DFS component counting (Subtopic 1.4/3's static tools) — it is
specifically built for workloads where connectivity structure changes incrementally and must be
queried after every change, which static flood-fill fundamentally cannot support efficiently.


````c++
// Number of Islands II: dynamic connectivity, DSU maintained incrementally as cells are added
#include <iostream>
#include <vector>
using namespace std;
#define vi vector<int>

struct DSUI {
    vi parent, rank_;
    DSUI(int n) : parent(n), rank_(n,0) { for (int i=0;i<n;i++) parent[i]=i; }
    int find(int x) { if (parent[x]!=x) parent[x]=find(parent[x]); return parent[x]; }
    bool unite(int a, int b) {
        int ra=find(a), rb=find(b);
        if (ra==rb) return false;
        if (rank_[ra]<rank_[rb]) swap(ra,rb);
        parent[rb]=ra;
        if (rank_[ra]==rank_[rb]) rank_[ra]++;
        return true;
    }
};

vi numIslands2(int R, int C, vector<vector<int>>& positions) {
    DSUI dsu(R * C);
    vector<vector<bool>> isLand(R, vector<bool>(C, false));
    vi result;
    int islandCount = 0;
    int dr[4]={-1,1,0,0}, dc[4]={0,0,-1,1};

    for (auto& p : positions) {
        int r = p[0], c = p[1];
        if (isLand[r][c]) { result.push_back(islandCount); continue; }   // duplicate addition, no change
        isLand[r][c] = true;
        islandCount++;                        // new cell starts as its own island
        int id = r * C + c;
        for (int d = 0; d < 4; d++) {
            int nr = r+dr[d], nc = c+dc[d];
            if (nr<0||nr>=R||nc<0||nc>=C || !isLand[nr][nc]) continue;
            if (dsu.unite(id, nr*C+nc)) islandCount--;   // successful merge -> two islands become one
        }
        result.push_back(islandCount);
    }
    return result;
}
````

````c++
// Test cell: input -> actual (Expected: X)

{
    vector<vector<int>> positions = {{0,0},{0,1},{1,2},{1,0},{2,1}};
    vi result = numIslands2(3, 3, positions);
    cout << "numIslands2(standard) -> ";
    for (int x : result) cout << x << " ";
    cout << "(Expected: 1 1 2 3 3)\n";
}

// Duplicate addition
{
    vector<vector<int>> positions = {{0,0},{0,0}};
    vi result = numIslands2(2, 2, positions);
    cout << "numIslands2(duplicate) -> ";
    for (int x : result) cout << x << " ";
    cout << "(Expected: 1 1)\n";
}

// Single cell grid
{
    vector<vector<int>> positions = {{0,0}};
    vi result = numIslands2(1, 1, positions);
    cout << "numIslands2(single cell) -> " << result[0] << " (Expected: 1)\n";
}

// Sequential additions that never touch -- island count keeps growing
{
    vector<vector<int>> positions = {{0,0},{2,2},{4,4}};
    vi result = numIslands2(5, 5, positions);
    cout << "numIslands2(spread out) -> ";
    for (int x : result) cout << x << " ";
    cout << "(Expected: 1 2 3)\n";
}
````

### 9.5.5 Making A Large Island (Union Then Query Max Component Size on Flip)

**State definition.** Given a binary grid, you may change **at most one** 0 to a 1. Find the
maximum possible size of a connected island afterward.

**Invariant.** First, build DSU over all existing land, unioning 4-connected land cells, tracking
each component's **size** (not just its root). Then, for **every** water cell (0), compute what its
resulting island size *would* be if flipped: the water cell itself ($+1$) plus the sizes of every
**distinct** land component adjacent to it (summed, each counted only once even if the water cell
touches the same component from multiple directions — this is the subtlety).

**Why the "distinct component" check is essential.** A single water cell can be 4-directionally
adjacent to the **same** land component more than once (e.g. a water cell in a small pocket
surrounded by land that's all one component via a path elsewhere) — naively summing every
neighboring land cell's component size without deduplicating roots would **double-count** that
component's contribution, producing an inflated, incorrect answer. Deduplicate via a small
set/hash of roots examined per water cell before summing.

**Why it works.** This is a direct query on top of Subtopic 1.4/9.3's static DSU component
structure — no new correctness proof beyond "component sizes are correctly maintained by DSU" (a
direct consequence of the union-by-size/rank invariant) plus the deduplication argument above.

**The delta.** This problem tests whether the "query, don't mutate" mode of DSU is understood —
unlike 9.5.4's *dynamic* (state-changing) DSU usage, this problem builds the DSU **once**
(static, over existing land) and then performs **read-only** what-if queries against it for every
candidate flip — the DSU structure itself never changes during the query phase, which is a
meaningfully different usage pattern from Subtopic 9's other applications.


````c++
// Making A Large Island: static DSU over existing land, then read-only what-if queries per water cell
#include <iostream>
#include <vector>
#include <unordered_set>
using namespace std;
#define vi vector<int>
#define vvi vector<vector<int>>

struct DSUL {
    vi parent, sz;
    DSUL(int n) : parent(n), sz(n, 1) { for (int i=0;i<n;i++) parent[i]=i; }
    int find(int x) { if (parent[x]!=x) parent[x]=find(parent[x]); return parent[x]; }
    void unite(int a, int b) {
        int ra=find(a), rb=find(b);
        if (ra==rb) return;
        if (sz[ra]<sz[rb]) swap(ra,rb);
        parent[rb]=ra;
        sz[ra]+=sz[rb];             // maintain true component size on every merge
    }
};

int largestIsland(vvi& grid) {
    int R = grid.size(), C = grid[0].size();
    DSUL dsu(R*C);
    int dr[4]={-1,1,0,0}, dc[4]={0,0,-1,1};

    // Phase 1: build static DSU over existing land, unioning 4-connected land cells.
    for (int r = 0; r < R; r++) {
        for (int c = 0; c < C; c++) {
            if (grid[r][c] != 1) continue;
            for (int d = 0; d < 4; d++) {
                int nr=r+dr[d], nc=c+dc[d];
                if (nr<0||nr>=R||nc<0||nc>=C || grid[nr][nc]!=1) continue;
                dsu.unite(r*C+c, nr*C+nc);
            }
        }
    }

    int best = 0;
    bool anyWater = false;
    // Phase 2: read-only what-if query for every water cell -- DSU is never mutated here.
    for (int r = 0; r < R; r++) {
        for (int c = 0; c < C; c++) {
            if (grid[r][c] != 0) continue;
            anyWater = true;
            unordered_set<int> distinctRoots;    // dedup -- avoid double-counting a component seen twice
            for (int d = 0; d < 4; d++) {
                int nr=r+dr[d], nc=c+dc[d];
                if (nr<0||nr>=R||nc<0||nc>=C || grid[nr][nc]!=1) continue;
                distinctRoots.insert(dsu.find(nr*C+nc));
            }
            int total = 1;   // the flipped cell itself
            for (int root : distinctRoots) total += dsu.sz[root];
            best = max(best, total);
        }
    }

    if (!anyWater) return R * C;   // no water cells at all -- entire grid is already one island
    return best;
}
````

````c++
// Test cell: input -> actual (Expected: X)

{
    vvi g = {{1,0},{0,1}};
    cout << "largestIsland(diagonal islands) -> " << largestIsland(g) << " (Expected: 3)\n";
}
{
    vvi g = {{1,1},{1,1}};
    cout << "largestIsland(all land already) -> " << largestIsland(g) << " (Expected: 4)\n";
}
{
    vvi g = {{0,0},{0,0}};
    cout << "largestIsland(all water) -> " << largestIsland(g) << " (Expected: 1)\n";
}
{
    // Water cell touching the SAME component twice (via a pocket) -- must not double count
    vvi g = {{1,1,1},{1,0,1},{1,1,1}};
    cout << "largestIsland(pocket, same component twice) -> " << largestIsland(g) << " (Expected: 9, whole ring is one component + the flipped center)\n";
}
````

### 9.5.6 Swim in Rising Water (Union-Find on Edges Sorted by Weight)

**State definition.** Given an $n \times n$ grid of elevations, at time $t$ you may move
4-directionally between cells whose elevation is $\le t$. Find the minimum $t$ such that a path
exists from top-left to bottom-right.

**Invariant — Union-Find on edges sorted by weight (Kruskal-flavored framing).** Treat each grid
cell as a vertex; each 4-directional adjacency as an edge with weight $= \max(\text{elev}(u),
\text{elev}(v))$ (the water level needed to traverse that specific edge). Process edges in
**increasing weight order** (exactly Kruskal's processing order, 9.4), unioning endpoints as you
go — the answer is the weight of the edge whose processing **first** causes source and destination
to become connected.

**Why it works.** This is the **cycle/cut property machinery repurposed for a bottleneck-path
query** rather than MST total weight: by processing edges in increasing order and unioning greedily
(exactly Kruskal's loop), the moment top-left and bottom-right first share a DSU root, the
maximum-weight edge used along *some* path connecting them (which, by Kruskal's own MST-correctness
proof, is a **minimum bottleneck path** — the path minimizing its own maximum edge weight) has just
been processed. This is a direct corollary of the **cut property (9.1)**: connectivity via
cheapest-first edge processing always uses the minimum possible "worst edge" to first bridge any
two components, by the same cut-property exchange argument that justifies Prim's/Kruskal's MST
correctness generally.

**Alternative framing — Dijkstra with max-of-path relaxation.** Exactly Subtopic 8.7.1's Path with
Minimum Effort: this is the identical "minimize the maximum edge weight along a path" problem,
solvable equally validly via Dijkstra's extraction proof with $\max$ substituted for $+$
(Subtopic 8.7.1's monotonicity argument applies unchanged here).

**The delta.** This problem is the deliberate bridge between Subtopic 8's Dijkstra-generalization
theme and Subtopic 9's Kruskal/DSU theme: **the exact same "minimize the bottleneck edge on a
path" objective is solvable by either a modified Dijkstra (8.7.1's proof) or a Kruskal-style
sorted-edge Union-Find sweep (9.1's cut property, applied here)** — recognizing that these are two
correct routes to the identical answer, not competing algorithms, is the capstone insight tying
Subtopics 8 and 9 together.


````c++
// Swim in Rising Water: Kruskal-style sorted-edge Union-Find, minimum bottleneck path
#include <iostream>
#include <vector>
#include <algorithm>
using namespace std;
#define vi vector<int>
#define vvi vector<vector<int>>

struct DSUW {
    vi parent, rank_;
    DSUW(int n) : parent(n), rank_(n,0) { for (int i=0;i<n;i++) parent[i]=i; }
    int find(int x) { if (parent[x]!=x) parent[x]=find(parent[x]); return parent[x]; }
    void unite(int a, int b) {
        int ra=find(a), rb=find(b);
        if (ra==rb) return;
        if (rank_[ra]<rank_[rb]) swap(ra,rb);
        parent[rb]=ra;
        if (rank_[ra]==rank_[rb]) rank_[ra]++;
    }
    bool connected(int a, int b) { return find(a) == find(b); }
};

int swimInWater(vvi& grid) {
    int n = grid.size();
    struct WE { int u, v, w; };
    vector<WE> edges;
    int dr[2] = {0, 1}, dc[2] = {1, 0};    // only need right + down to enumerate every grid edge once

    for (int r = 0; r < n; r++) {
        for (int c = 0; c < n; c++) {
            for (int d = 0; d < 2; d++) {
                int nr = r+dr[d], nc = c+dc[d];
                if (nr>=n || nc>=n) continue;
                int u = r*n+c, v = nr*n+nc;
                int w = max(grid[r][c], grid[nr][nc]);   // water level needed to cross THIS edge
                edges.push_back({u, v, w});
            }
        }
    }
    sort(edges.begin(), edges.end(), [](const WE& a, const WE& b){ return a.w < b.w; });   // Kruskal ordering

    DSUW dsu(n*n);
    int src = 0, dst = n*n - 1;
    // Source and destination cells' own elevations set a lower bound -- must be able to stand on both.
    int answer = max(grid[0][0], grid[n-1][n-1]);

    for (auto& e : edges) {
        dsu.unite(e.u, e.v);
        if (dsu.connected(src, dst)) { answer = max(answer, e.w); return answer; }   // first connection = bottleneck answer
    }
    return answer;
}
````

````c++
// Test cell: input -> actual (Expected: X)

{
    vvi g = {{0,2},{1,3}};
    cout << "swimInWater(2x2) -> " << swimInWater(g) << " (Expected: 3)\n";
}
{
    vvi g = {{0,1,2,3,4},{24,23,22,21,5},{12,13,14,15,16},{11,17,18,19,20},{10,9,8,7,6}};
    cout << "swimInWater(5x5 spiral) -> " << swimInWater(g) << " (Expected: 16)\n";
}
{
    vvi g = {{0}};
    cout << "swimInWater(single cell) -> " << swimInWater(g) << " (Expected: 0)\n";
}
{
    vvi g = {{3,2},{0,1}};
    cout << "swimInWater(high start elevation) -> " << swimInWater(g) << " (Expected: 3, must start standing at elevation 3)\n";
}
````

````c++
// ================= UNIFIED MENTAL MODEL — Subtopic 9 =================
#include <iostream>
#include <vector>
#include <queue>
#include <algorithm>
#include <climits>
using namespace std;
#define vi vector<int>
#define pii pair<int,int>
#define vpii vector<pair<int,int>>
#define vvpii vector<vector<pii>>

// --- Template: DSU, union by rank + path compression (9.3, the foundation for everything below) ---
struct DSUTemplate {
    vi parent, rank_;
    int components;
    DSUTemplate(int n) : parent(n), rank_(n,0), components(n) { for (int i=0;i<n;i++) parent[i]=i; }
    int find(int x) { if (parent[x]!=x) parent[x]=find(parent[x]); return parent[x]; }
    bool unite(int a, int b) {
        int ra=find(a), rb=find(b);
        if (ra==rb) return false;
        if (rank_[ra]<rank_[rb]) swap(ra,rb);
        parent[rb]=ra;
        if (rank_[ra]==rank_[rb]) rank_[ra]++;
        components--;
        return true;
    }
};

// --- Template: Prim's -- priority queue over VERTICES, key = cheapest edge to the current tree ---
long long primTemplate(int V, vvpii& adj) {
    vector<bool> inMST(V, false);
    priority_queue<pii, vpii, greater<pii>> pq;
    pq.push({0, 0});
    long long total = 0;
    while (!pq.empty()) {
        auto [w, u] = pq.top(); pq.pop();
        if (inMST[u]) continue;
        inMST[u] = true; total += w;
        for (auto& [v, ew] : adj[u]) if (!inMST[v]) pq.push({ew, v});
    }
    return total;
}

// --- Template: Kruskal's -- sort EDGES globally, DSU to reject cycle-forming edges ---
struct WE { int u, v, w; };
long long kruskalTemplate(int V, vector<WE> edges) {
    sort(edges.begin(), edges.end(), [](const WE&a, const WE&b){ return a.w < b.w; });
    DSUTemplate dsu(V);
    long long total = 0;
    for (auto& e : edges) if (dsu.unite(e.u, e.v)) total += e.w;
    return total;
}

int main() {
    vvpii adj(3); adj[0]={{1,1},{2,3}}; adj[1]={{0,1},{2,1}}; adj[2]={{0,3},{1,1}};
    cout << "Unified template sanity check: primTemplate -> " << primTemplate(3, adj) << " (Expected: 2)\n";
    return 0;
}
````

### Decision Tree — Subtopic 9

```
Does the problem ask for a Minimum Spanning Tree (or forest) directly?
│
├── YES
│    │
│    Is the graph given as an ADJACENCY LIST (vertex-centric), and is it dense?
│    ├── YES → Prim's Algorithm (9.2) -- priority queue over vertices, cut-property justified
│    │
│    └── Is the graph given as an EDGE LIST, or is it sparse?
│         └── YES → Kruskal's Algorithm (9.4) -- sort edges globally, DSU rejects cycles,
│                   cycle-property justified
│    (Both are always correct on any graph -- this is purely a data-structure/density fit choice,
│     exactly like Dijkstra's heap-vs-array choice in Subtopic 8.4)
│
└── NO -- not an MST problem directly, but connectivity/grouping is still central
     │
     Does the "graph" need repeated/incremental connectivity queries as edges arrive over time?
     ├── YES → plain DSU (9.3), NOT repeated BFS/DFS (Subtopic 4.2's asymptotic penalty applies
     │         identically here) -- e.g. Number of Islands II (9.5.4)
     │
     Are the "elements" to union NOT the objects named directly in the problem (e.g. rows/columns
     instead of stones, emails instead of accounts)?
     ├── YES → identify the TRUE underlying element set first (9.5.2, 9.5.3), then apply plain DSU
     │
     Does the problem ask "minimize the maximum edge weight along a path" (bottleneck path)?
     ├── YES → EITHER Kruskal-style sorted-edge Union-Find sweep (9.5.6, cut-property justified)
     │         OR Dijkstra with max() relaxation (Subtopic 8.7.1) -- both are correct, pick based
     │         on whether you already have a DSU or a heap set up
     │
     Does the problem ask for max component size after a hypothetical single change?
     └── YES → build DSU ONCE (static) over existing structure, then run READ-ONLY what-if
               queries per candidate change, deduplicating component roots per query (9.5.5)
```


### Complexity Summary — Subtopic 9

| Algorithm / Application | Time | Space | When to use | Key invariant | Failure mode if misused |
|---|---|---|---|---|---|
| Cut property / Cycle property (9.1) | — (proof tools) | — | Justifying MST algorithm correctness | min-weight cut edge is safe to include; max-weight cycle edge is safe to exclude | — |
| Prim's Algorithm (9.2) | $O((V+E)\log V)$ heap, $O(V^2)$ naive dense-array variant | $O(V+E)$ | Adjacency list, dense graphs favor the array variant | current tree ⊆ some MST | confusing `key[v]` (single edge) with Dijkstra's accumulated `dist[v]` gives wrong totals |
| Disjoint Set Union (9.3) | $O(\alpha(n))$ amortized per op (union by rank + path compression together) | $O(n)$ | Any connectivity/grouping query, especially incremental | same-root $\iff$ same set | using only ONE of rank/compression loses the tightest amortized bound |
| Kruskal's Algorithm (9.4) | $O(E \log E)$ (sort dominates) | $O(V+E)$ | Edge list given directly, sparse graphs | sorted-order + DSU-rejects-cycles ⊆ some MST | forgetting to sort first, or checking connectivity via BFS instead of DSU, reintroduces the $O(V \cdot E)$ penalty DSU exists to avoid |
| Network Connected Ops (9.5.1) | $O(E \cdot \alpha(V))$ | $O(V)$ | Count components, feasibility via edge-count | components-1 = min edges needed | forgetting the `edges >= n-1` feasibility pre-check |
| Most Stones Removed (9.5.2) | $O((R+C)\cdot \alpha(R+C))$ | $O(R+C)$ | Union an auxiliary structure (rows/cols), not the named objects directly | stones - components = max removable | unioning stones directly (instead of rows/columns) misses transitive connections |
| Accounts Merge (9.5.3) | $O(N \alpha(N) + N\log N)$ for $N$ total emails | $O(N)$ | Union emails, not accounts | same-root emails belong to one merged account | pairwise-only comparison (no DSU) misses transitive merges |
| Number of Islands II (9.5.4) | $O(k \cdot \alpha(RC))$ for $k$ additions | $O(RC)$ | Incremental connectivity on a grid | successful union = one fewer island | using flood-fill (Subtopic 3) per addition costs $O(k \cdot RC)$ instead |
| Making a Large Island (9.5.5) | $O(RC \cdot \alpha(RC))$ | $O(RC)$ | Static DSU + read-only what-if queries | dedup component roots per query to avoid double-counting | forgetting to dedup roots inflates the answer when a water cell touches one component twice |
| Swim in Rising Water (9.5.6) | $O(E \log E)$, $E=O(n^2)$ | $O(n^2)$ | Bottleneck-path (min-max-edge) queries | first Kruskal-order union connecting src/dst = the answer | forgetting the `max(grid[0][0], grid[n-1][n-1])` lower bound undercounts when start/end elevation alone exceeds the path's max edge |
