# Graphs — Subtopic 6: Word Transformation (Implicit Graphs)

**Kernel:** C++17 (xeus-cling, `xcpp17`)

Pedagogical order — cleanest baseline first, then the variant that stresses the hardest new idea:
1. Word Ladder I — building the implicit graph, BFS for shortest transformation, optimizations (bidirectional BFS, pattern buckets)
2. Word Ladder II — BFS for layer distances, then DFS/backtracking to enumerate ALL shortest paths; deep dive on why the naive "collect parents during BFS" is subtle
3. General lesson — how to model any state-space problem as an implicit graph, and how to identify the transition function


## 6.1 Word Ladder I

### State definition
Given a `beginWord`, a `endWord`, and a `wordList`, find the length of the shortest transformation
sequence from `beginWord` to `endWord`, changing exactly one letter per step, where every
intermediate word must be in `wordList`.

$\text{dist}[w]$ = minimum number of one-letter transformations from `beginWord` to reach word $w$.

### Building the implicit graph (direct instance of Subtopic 1.1's framing)
$$
V = \text{wordList} \cup \{\text{beginWord}\}, \qquad
E = \{(u,v) : \text{hamming}(u,v) = 1\}
$$
where $\text{hamming}(u,v)$ is the number of positions at which $u$ and $v$ differ (both words
have equal length by problem constraints). This is never materialized as an adjacency list —
neighbors of $u$ are generated lazily by trying every possible single-letter substitution and
checking dictionary membership. This is *exactly* Subtopic 1.1's $\delta(u)$ transition function,
here concretely: $\delta(u) = \{v : v$ differs from $u$ in exactly one position, $v \in$ dictionary$\}$.

### Invariant
Subtopic 2.1's BFS layer invariant, unchanged — the only new content is that $\text{adj}(u)$ is
computed on demand rather than read from a stored structure.

### Why it works
Direct corollary of Subtopic 2.1's proof: since every transformation (edge) has equal cost (one
letter change = one edge), BFS's non-decreasing-distance dequeue order guarantees
$\text{dist}[\text{endWord}]$, once dequeued, is the true minimum transformation count.

### Optimization 1 — bidirectional BFS

**Idea:** run BFS simultaneously from `beginWord` forward and `endWord` backward, alternating
which frontier expands, and stop the moment the two frontiers meet.

**Why it's faster — complexity argument.** A single-direction BFS explores a "ball" of radius $d$
(the true shortest distance) around the source, visiting $O(b^d)$ vertices in the worst case where
$b$ is the branching factor (here, roughly $26 \times \text{word length}$ candidate substitutions
per word, filtered by dictionary membership). Bidirectional BFS instead grows two balls of radius
$d/2$ each, visiting $O(2 \cdot b^{d/2})$ vertices — since $b^{d/2} \ll b^d$ for large $d$, this is
an exponential-in-the-exponent improvement whenever the branching factor $b > 1$. **Correctness is
preserved** because meeting-in-the-middle at layer $d/2$ from each side is equivalent, by the same
non-decreasing-distance argument applied to both frontiers simultaneously, to finding the true
midpoint of a shortest path — the total path length is the sum of the two layers at the moment of
first intersection.

### Optimization 2 — pattern buckets (avoiding the $O(26 \cdot L)$-per-word generation cost)

**Idea:** precompute, for every word in the dictionary, its $L$ "wildcard patterns" (e.g. `"hot"`
$\to$ `"*ot"`, `"h*t"`, `"ho*"`), and group dictionary words by shared pattern. Two words are
neighbors (Hamming distance 1) **iff** they share at least one wildcard pattern. This turns
neighbor lookup from $O(26 \cdot L)$ brute-force substitution-and-lookup into $O(L)$ pattern
lookups into a precomputed hash map, at the cost of $O(N \cdot L)$ preprocessing space for $N$
dictionary words.

### Boundary transitions table

| Decision point | Condition | Action |
|---|---|---|
| `endWord` not in wordList | dictionary membership check fails | return 0 immediately — no valid sequence exists (endWord must be reachable through the dictionary) |
| Candidate substitution not in dictionary | generated word absent from wordList | skip — not a valid vertex in $V$ |
| Candidate substitution already visited | `dist[v] != -1` | skip — visited-set discipline, standard BFS rule |
| `beginWord == endWord` | trivial (not typical per problem constraints, but a valid boundary check) | distance is 0 |

### The delta
The non-obvious insight: Word Ladder I is a **pure application** of Subtopic 1.1's implicit-graph
framing plus Subtopic 2.1's BFS proof — nothing new is required for correctness. The genuinely new
content is entirely in the *engineering* of neighbor generation (bidirectional search, pattern
buckets), because unlike adjacency-list/grid BFS (Subtopic 2.4), computing $\delta(u)$ here is
itself a nontrivial cost ($O(26L)$ per vertex naively) that dominates the runtime if not optimized.


````c++
// Word Ladder I: implicit graph BFS with pattern-bucket neighbor generation
#include <iostream>
#include <vector>
#include <string>
#include <unordered_map>
#include <unordered_set>
#include <queue>
using namespace std;

// buildPatternBuckets: for every word, generate its L wildcard patterns and bucket words by pattern.
// This is the delta() optimization -- neighbor lookup becomes O(L) pattern hits instead of O(26*L).
unordered_map<string, vector<string>> buildPatternBuckets(vector<string>& wordList) {
    unordered_map<string, vector<string>> buckets;
    for (string& w : wordList) {
        for (int i = 0; i < (int)w.size(); i++) {
            string pattern = w;
            pattern[i] = '*';                       // wildcard at position i
            buckets[pattern].push_back(w);
        }
    }
    return buckets;
}

// ladderLength: multi-... actually single-source BFS on the implicit word graph, using pattern buckets.
int ladderLength(string beginWord, string endWord, vector<string>& wordList) {
    unordered_set<string> dict(wordList.begin(), wordList.end());
    if (!dict.count(endWord)) return 0;              // endWord unreachable via the dictionary -- boundary case

    auto buckets = buildPatternBuckets(wordList);
    unordered_map<string, int> dist;                 // dist[w] = transformations from beginWord (implicit graph dist)
    queue<string> q;
    dist[beginWord] = 1;                              // problem convention: path LENGTH counts words, so begin=1
    q.push(beginWord);

    while (!q.empty()) {
        string u = q.front(); q.pop();
        if (u == endWord) return dist[u];             // early exit: BFS layer invariant guarantees optimality
        for (int i = 0; i < (int)u.size(); i++) {
            string pattern = u;
            pattern[i] = '*';
            if (!buckets.count(pattern)) continue;
            for (string& v : buckets[pattern]) {       // delta(u): all dictionary words sharing this pattern
                if (dist.count(v)) continue;            // visited-set discipline
                dist[v] = dist[u] + 1;
                q.push(v);
            }
        }
    }
    return 0;    // endWord never reached
}
````

````c++
// Test cell: input -> actual (Expected: X)

// Standard case: hit -> hot -> dot -> dog -> cog
{
    vector<string> wl = {"hot","dot","dog","lot","log","cog"};
    cout << "ladderLength(hit->cog) -> " << ladderLength("hit", "cog", wl) << " (Expected: 5)\n";
}

// endWord not in dictionary -> impossible
{
    vector<string> wl = {"hot","dot","dog","lot","log"};
    cout << "ladderLength(endWord missing) -> " << ladderLength("hit", "cog", wl) << " (Expected: 0)\n";
}

// Empty word list
{
    vector<string> wl = {};
    cout << "ladderLength(empty dict) -> " << ladderLength("hit", "cog", wl) << " (Expected: 0)\n";
}

// beginWord directly one letter away from endWord
{
    vector<string> wl = {"dot"};
    cout << "ladderLength(one hop) -> " << ladderLength("hot", "dot", wl) << " (Expected: 2)\n";
}

// Single-word dictionary that IS the endWord, no path needed beyond direct hop
{
    vector<string> wl = {"a","b","c"};
    cout << "ladderLength(single-letter words) -> " << ladderLength("a", "c", wl) << " (Expected: 2, a->c differs in 1 position)\n";
}
````

## 6.2 Word Ladder II

### State definition
Same setup as 6.1, but now return **every** shortest transformation sequence from `beginWord` to
`endWord`, not just the length.

### Invariant — layer distances first, then reconstruct

**Phase 1 (BFS):** exactly 6.1's BFS, computing $\text{dist}[w]$ for every reachable word — the
minimum layer at which $w$ is first reached.

**Phase 2 (DFS/backtracking):** starting from `endWord`, walk **backward** through words $w$ such
that $\text{dist}[w] = \text{dist}[\text{current}] - 1$ **and** $w$ is a valid one-letter neighbor
of `current` — this reconstructs every shortest path by construction, since any path using only
strictly-decreasing-by-exactly-1 layer steps is automatically a shortest path (it has length
exactly $\text{dist}[\text{endWord}]$, which is optimal by Phase 1's BFS proof).

### Deep dive — why the naive "collect parents during BFS" is subtle

**The naive idea:** during the single BFS pass, whenever word $v$ is first discovered from $u$,
record $u$ as $v$'s "parent," exactly as in Subtopic 2's traversal templates. Then reconstruct
paths by following parent pointers backward from `endWord`.

**Why this naive idea is *insufficient* for Word Ladder II specifically:** in vanilla BFS (single
shortest path, e.g. Subtopic 2.1 or 4.3), each vertex needs only **one** parent, because only one
shortest path is required. But Word Ladder II wants **all** shortest paths — and a word $v$ can
have **multiple** valid predecessors $u_1, u_2, \ldots$, all at the same layer $\text{dist}[v]-1$,
each contributing a distinct valid shortest path. A single-parent-pointer BFS silently keeps only
the *first* discoverer of $v$ and discards the rest — this is precisely the standard visited-set
discipline (Subtopic 2.5) doing exactly what it's designed to do for *reachability* purposes
(preventing redundant re-exploration), but that same discipline actively **destroys** information
needed for *path-enumeration* purposes.

**The fix:** decouple "has $v$'s distance been finalized" (still checked once, to keep BFS
$O(V+E)$ and prevent the exponential blowup proven in Subtopic 2.5) from "which words can precede
$v$ on a shortest path" (which must record **every** $u$ with an edge to $v$ satisfying
$\text{dist}[u] = \text{dist}[v] - 1$, not just the first one found). Concretely: allow a word to be
enqueued/expanded-from only once (distance finalization, preserving $O(V+E)$ BFS cost), but during
Phase 2 (backtracking), **recompute** which neighbors of the current word satisfy the layer
condition, rather than trusting a single stored parent pointer.

### Memory / time tradeoffs

| Approach | Time | Space | Correctness |
|---|---|---|---|
| Single parent pointer per word | $O(V+E)$ BFS | $O(V)$ | **Wrong** — only reconstructs one shortest path, silently drops others |
| Store ALL valid predecessors per word (multi-parent map) during BFS | $O(V+E)$ BFS (each edge still examined once) | $O(V+E)$ to store the predecessor lists | Correct — Phase 2 backtracks through all recorded predecessors |
| Recompute predecessors during backtracking (no storage during BFS) | $O(V+E)$ BFS, $O(P \cdot 26L)$ backtracking for $P$ total path-cells visited during reconstruction | $O(V)$ (just distances) | Correct — trades backtracking-time recomputation for BFS-phase memory savings |
| Enumerate ALL paths themselves | Can be exponential in the number of shortest paths (unavoidable — the *output size itself* can be exponential) | proportional to total output size | Correct, but caution: this is why layer-restricted backtracking (only ever stepping to $\text{dist}-1$) is essential to avoid exploring non-shortest paths, which would be even more wasteful |

### Boundary transitions table

| Decision point | Condition | Action |
|---|---|---|
| BFS discovery of $v$, first time ever | $\text{dist}[v]$ unset | set $\text{dist}[v] = \text{dist}[u]+1$, enqueue (finalizes distance) |
| BFS discovery of $v$, already finalized at a strictly greater layer | impossible in correct BFS (layer invariant) | — |
| BFS discovery of $v$, already finalized at the SAME layer as this edge would produce | $\text{dist}[v] = \text{dist}[u]+1$ already | do **not** re-enqueue (keeps $O(V+E)$), but this edge is still a valid predecessor relationship for Phase 2 |
| Backtracking step from `current` to candidate `prev` | $\text{dist}[\text{prev}] = \text{dist}[\text{current}] - 1$ AND hamming distance 1 | valid predecessor — recurse |

### The delta
The non-obvious insight, stated precisely: **the visited-set discipline (Subtopic 2.5) that makes
BFS $O(V+E)$ for distance/reachability purposes is exactly what must be relaxed — carefully, only
for path *enumeration* — when the problem needs all shortest paths, not just the shortest
distance.** Getting this wrong (blindly reusing a single-parent-pointer BFS template from Subtopic
2) is the single most common bug in Word Ladder II: it doesn't crash and doesn't obviously misbehave
— it just silently returns fewer paths than exist, since the missing branches are structurally
indistinguishable from "there were no other equally-short paths" without careful auditing.


````c++
// Word Ladder II: BFS for layer distances, then backtracking to enumerate ALL shortest paths
#include <iostream>
#include <vector>
#include <string>
#include <unordered_map>
#include <unordered_set>
#include <queue>
#include <algorithm>
using namespace std;

unordered_map<string, vector<string>> buildBuckets(vector<string>& wordList) {
    unordered_map<string, vector<string>> buckets;
    for (string& w : wordList) {
        for (int i = 0; i < (int)w.size(); i++) {
            string p = w; p[i] = '*';
            buckets[p].push_back(w);
        }
    }
    return buckets;
}

// phase1BFS: computes dist[w] for every reachable word. Distance is FINALIZED once (O(V+E) total),
// matching the visited-set discipline -- but note we do NOT store single parent pointers here.
unordered_map<string,int> phase1BFS(string& beginWord, string& endWord,
                                     unordered_map<string, vector<string>>& buckets) {
    unordered_map<string,int> dist;
    queue<string> q;
    dist[beginWord] = 0;
    q.push(beginWord);
    while (!q.empty()) {
        string u = q.front(); q.pop();
        if (u == endWord) break;                 // distances beyond endWord's layer are never needed
        for (int i = 0; i < (int)u.size(); i++) {
            string p = u; p[i] = '*';
            if (!buckets.count(p)) continue;
            for (string& v : buckets[p]) {
                if (dist.count(v)) continue;      // finalize distance ONCE -- keeps BFS O(V+E)
                dist[v] = dist[u] + 1;
                q.push(v);
            }
        }
    }
    return dist;
}

// isOneHamming: true iff a and b differ in exactly one character position.
bool isOneHamming(const string& a, const string& b) {
    int diff = 0;
    for (int i = 0; i < (int)a.size(); i++) if (a[i] != b[i]) { diff++; if (diff > 1) return false; }
    return diff == 1;
}

// phase2Backtrack: from `current`, recompute ALL valid predecessors at layer dist[current]-1,
// rather than trusting a single stored parent -- this is the fix for the naive single-parent bug.
void phase2Backtrack(string current, string& beginWord, unordered_map<string,int>& dist,
                      vector<string>& path, vector<vector<string>>& results) {
    path.push_back(current);
    if (current == beginWord) {
        vector<string> rev(path.rbegin(), path.rend());
        results.push_back(rev);
    } else {
        for (auto& [w, d] : dist) {
            if (d == dist[current] - 1 && isOneHamming(w, current)) {
                phase2Backtrack(w, beginWord, dist, path, results);
            }
        }
    }
    path.pop_back();
}

vector<vector<string>> findLadders(string beginWord, string endWord, vector<string>& wordList) {
    unordered_set<string> dictSet(wordList.begin(), wordList.end());
    if (!dictSet.count(endWord)) return {};

    auto buckets = buildBuckets(wordList);
    unordered_map<string,int> dist = phase1BFS(beginWord, endWord, buckets);
    if (!dist.count(endWord)) return {};          // endWord unreachable

    vector<vector<string>> results;
    vector<string> path;
    phase2Backtrack(endWord, beginWord, dist, path, results);
    return results;
}
````

````c++
// Test cell: input -> actual (Expected: X)

// Classic case with TWO distinct shortest paths of equal length
{
    vector<string> wl = {"hot","dot","dog","lot","log","cog"};
    auto res = findLadders("hit", "cog", wl);
    cout << "findLadders(hit->cog) #paths -> " << res.size() << " (Expected: 2)\n";
    if (!res.empty()) cout << "  each path length -> " << res[0].size() << " (Expected: 5)\n";
}

// endWord not in dictionary -> no paths
{
    vector<string> wl = {"hot","dot","dog","lot","log"};
    auto res = findLadders("hit", "cog", wl);
    cout << "findLadders(endWord missing) #paths -> " << res.size() << " (Expected: 0)\n";
}

// Empty dictionary
{
    vector<string> wl = {};
    auto res = findLadders("hit", "cog", wl);
    cout << "findLadders(empty dict) #paths -> " << res.size() << " (Expected: 0)\n";
}

// Single unique shortest path (no branching)
{
    vector<string> wl = {"dot"};
    auto res = findLadders("hot", "dot", wl);
    cout << "findLadders(single hop) #paths -> " << res.size() << " (Expected: 1)\n";
}

// beginWord itself not required to be in wordList (per typical problem constraints)
{
    vector<string> wl = {"a","b","c"};
    auto res = findLadders("a", "c", wl);
    cout << "findLadders(single-letter, a->c) #paths -> " << res.size() << " (Expected: 1, path a->c directly)\n";
}
````

## 6.3 General Lesson — Modeling Any State-Space Problem as an Implicit Graph

### The recipe (restated with full precision, building on Subtopic 1.1)

To model *any* state-space search as a graph problem, answer these three questions precisely:

1. **What is a state (a node)?** It must satisfy the Markov property from Subtopic 1.1: the only
   information needed to decide legal next moves is the state itself. If your "obvious" state
   definition doesn't satisfy this, augment it (e.g. `(word)` for distance, but `(word, layer)` or
   richer for path enumeration in Word Ladder II — see 6.2's deep dive for exactly this kind of
   augmentation, where "predecessor set" had to become part of what's tracked, not just "distance").

2. **What is the transition function $\delta$ (the edges)?** Given a state $u$, what are *all* the
   states reachable in one legal move? This must be computable (even if expensive) for any given
   $u$ — you do not need to enumerate all of $V$ upfront, only to generate $\delta(u)$ on demand.
   - Word Ladder: $\delta(u) = \{v : \text{hamming}(u,v)=1, v \in \text{dictionary}\}$
   - Subtopic 1.1's min-ops example: $\delta(u) = \{u+1, u-1, 2u\}$
   - Any puzzle (sliding tile, Rubik's cube, etc.): $\delta(u) = \{$ all board states reachable by one legal move from $u \}$

3. **What is the query?** This determines which graph algorithm to reach for:
   - "Minimum number of moves to reach a target" → BFS (unweighted shortest path, Subtopic 2.1)
   - "All ways to reach the target in the minimum number of moves" → BFS for distances, then
     backtracking for enumeration (6.2's two-phase pattern)
   - "Is the target reachable at all" → any traversal (BFS or DFS), Subtopic 1.4/2's connectivity tools
   - "What's the best order to resolve dependencies" → topological sort (Subtopic 7)

### Boundary transitions table — is your implicit graph well-posed?

| Decision point | Condition | Action |
|---|---|---|
| Is the state space provably finite (or tightly boundable)? | yes | proceed with standard BFS/DFS, guaranteed termination |
| Is the state space provably finite? | no / unclear | add explicit bounds/pruning (as in 1.1's `bound` parameter) before running any traversal |
| Does the query need only distance, or full path enumeration? | distance only | single BFS pass suffices, visited-set discipline applies without modification |
| Does the query need full path enumeration? | yes | two-phase pattern (6.2): finalize distances first, then a separate backtracking pass that does NOT rely on single parent pointers |
| Is $\delta(u)$ itself expensive to compute naively? | yes (e.g. Word Ladder's $O(26L)$ substitution generation) | invest in a precomputation structure (pattern buckets) analogous to picking the right representation in Subtopic 1.2 |

### The delta
The non-obvious insight tying this subtopic together: **"implicit graph" problems are not a
different category of algorithm from everything in Subtopics 1–5** — they are the *same* BFS/DFS
machinery, with the *only* new work being (a) defining $\delta$ correctly and efficiently, and (b)
correctly matching the query type to the traversal pattern (single-path vs. all-shortest-paths,
exactly as sharpened by the 6.2 deep dive). Once $\delta$ is defined, every proof, invariant, and
failure mode from Subtopics 2 and 5 (visited-set discipline, layer invariant, GRAY-tracking for
directed cycles if the state graph happens to be directed) applies completely unchanged.


````c++
// ================= UNIFIED MENTAL MODEL — Subtopic 6 =================
#include <iostream>
#include <vector>
#include <string>
#include <unordered_map>
#include <queue>
using namespace std;

// --- Template: implicit graph BFS for shortest distance (6.1 pattern, generalizes 1.1's example) ---
// delta(state) generates neighbors lazily -- swap this out for any state-space problem.
template<typename State, typename DeltaFn>
unordered_map<State,int> implicitBFSDistances(State start, DeltaFn delta) {
    unordered_map<State,int> dist;
    queue<State> q;
    dist[start] = 0;
    q.push(start);
    while (!q.empty()) {
        State u = q.front(); q.pop();
        for (State v : delta(u)) {
            if (dist.count(v)) continue;         // visited-set discipline: finalize distance once
            dist[v] = dist[u] + 1;
            q.push(v);
        }
    }
    return dist;
}

// --- Template: two-phase pattern for ALL shortest paths (6.2 pattern) ---
// Phase 1 uses implicitBFSDistances() above (or equivalent) to get dist[].
// Phase 2 backtracks from target, recomputing valid predecessors via delta-inverse or a direct check --
// NEVER via a single stored parent pointer, which silently drops equally-short alternate paths.

int main() {
    // Demonstrate with the Subtopic 1.1 min-ops delta, confirming the template generalizes.
    auto delta = [](int u) { return vector<int>{u+1, u-1, u*2}; };
    auto dist = implicitBFSDistances<int>(5, delta);
    cout << "Unified template sanity check: dist[8] from 5 -> " << dist[8] << " (Expected: 3, matches Subtopic 1.1's example)\n";
    return 0;
}
````

### Decision Tree — Subtopic 6

```
Can the problem be phrased as "states + legal moves between states"?
│
├── NO  → not a graph-modelable problem as stated (revisit the formulation, Subtopic 1.1)
│
└── YES → define delta(state) = the transition function (the implicit edges)
     │
     Is delta(u) expensive to compute naively (e.g. many candidate substitutions per state)?
     ├── YES → invest in a precomputed lookup structure (pattern buckets, 6.1) analogous to
     │         picking the right representation (Subtopic 1.2)
     └── NO  → generate neighbors directly on demand, as in Subtopic 1.1's example
     │
     What does the query need?
     ├── "Shortest distance / min moves only" → single BFS pass, standard visited-set discipline (6.1)
     ├── "ALL shortest paths, not just one"    → two-phase: BFS for dist[] first, THEN a separate
     │                                            backtracking pass that recomputes valid predecessors
     │                                            per state -- do NOT reuse a single-parent-pointer
     │                                            BFS template here (6.2 deep dive)
     └── "Is it reachable at all / dependency ordering" → plain traversal (Subtopic 1/2) or
                                                            topological sort (Subtopic 7)
     │
     Is the state space provably finite?
     ├── YES → proceed as normal
     └── NO  → add explicit bounds/pruning BEFORE running BFS/DFS (Subtopic 1.1's `bound` parameter)
```


### Complexity Summary — Subtopic 6

| Pattern | Time | Space | When to use | Key invariant | Failure mode if misused |
|---|---|---|---|---|---|
| Word Ladder I (naive delta) | $O(V \cdot 26L)$ | $O(V)$ | Small dictionaries, simplicity over speed | BFS layer invariant, unweighted implicit graph | fine for small inputs, wasteful at scale |
| Word Ladder I (pattern buckets) | $O(V \cdot L)$ to build buckets, $O(V \cdot L)$ BFS | $O(V \cdot L)$ | Larger dictionaries where naive substitution generation dominates runtime | same BFS proof, cheaper $\delta(u)$ evaluation | forgetting to rebuild/reuse buckets correctly per query wastes the precomputation investment |
| Word Ladder I (bidirectional BFS) | $O(b^{d/2})$ vs. $O(b^d)$ naive | $O(b^{d/2})$ | Long shortest-path distances, high branching factor $b$ | meeting-in-the-middle still respects both frontiers' layer invariants | forgetting to alternate expanding the SMALLER frontier loses the balanced-tree speedup |
| Word Ladder II, Phase 1 (BFS distances) | $O(V \cdot L)$ (with buckets) | $O(V)$ | Computing dist[] for every word, needed before any path reconstruction | distance finalized once per word (visited-set discipline preserved) | none, if buckets/delta correctly defined |
| Word Ladder II, Phase 2 (backtracking) | $O(P \cdot V \cdot L)$ worst case, where $P$ = total path-reconstruction work | $O(\text{total output path length})$ | Enumerating ALL shortest paths | only step to layer $\text{dist}-1$ neighbors — guarantees shortest-path-only exploration | using a single stored parent pointer instead silently drops all but one shortest path per word |
