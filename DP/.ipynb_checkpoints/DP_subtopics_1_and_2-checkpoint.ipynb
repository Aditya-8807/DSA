# Subtopic 1 — Introduction to DP

## 1.1 What makes a problem a DP problem?

A problem admits a DP solution **iff** it has both:

1. **Optimal substructure** — the optimal answer for an instance can be expressed as a function of optimal answers to strictly smaller instances of the *same* problem.
2. **Overlapping subproblems** — the naive recursion revisits the same subproblem many times. (If subproblems are *disjoint*, you have divide-and-conquer, not DP.)

We prove both on Fibonacci, the canonical entry point.

---

## 1.2 Fibonacci — proving both properties

**Definition.** $F_0 = 0,\; F_1 = 1,\; F_n = F_{n-1} + F_{n-2}$ for $n \geq 2$.

### Invariant (state definition)

$$dp[i] = \text{the } i\text{-th Fibonacci number}$$

### Recurrence

$$
dp[i] =
\begin{cases}
i & \text{if } i \in \{0, 1\} \\
dp[i-1] + dp[i-2] & \text{if } i \geq 2
\end{cases}
$$

### Optimal substructure (proof)

To compute $F_n$ we *must* know $F_{n-1}$ and $F_{n-2}$. Both are independent instances of the same problem on strictly smaller inputs. Substituting any non-optimal (i.e., wrong) value for $F_{n-1}$ or $F_{n-2}$ breaks the answer. ✓

### Overlapping subproblems (proof on the call tree)

Consider naive recursion `fib(5)`:

```
                       fib(5)
                      /      \
                  fib(4)      fib(3)
                  /    \      /    \
              fib(3) fib(2) fib(2) fib(1)
              /  \    / \    / \
         fib(2) f(1) f(1) f(0) f(1) f(0)
         /  \
      f(1)  f(0)
```

`fib(3)` appears **twice**, `fib(2)` appears **three** times, `fib(1)` and `fib(0)` even more. The number of leaves grows as $\Theta(\varphi^n)$ where $\varphi = (1+\sqrt 5)/2$. With memoization each subproblem is computed once → $\Theta(n)$. ✓

---

## 1.3 Memoization vs Tabulation — the duality

| Aspect | Memoization (top-down) | Tabulation (bottom-up) |
|---|---|---|
| Direction | Recurse from $n$ toward base | Iterate from base toward $n$ |
| Computes only | States reachable from $n$ | All states up to $n$ |
| Stack | Uses call stack (risk of overflow at $n \approx 10^5$) | No recursion |
| Easier when | Recurrence is hard to topologically order | Recurrence has clear iteration order |
| Constants | Higher (recursion + hashing/array lookup) | Lower (tight loop) |
| Space optimization | Harder to reason about | Direct — drop unneeded rows |

**Rule of thumb.** Develop with memoization (closest to the recurrence — fewer bugs). Ship with tabulation (faster, no stack risk, enables space optimization).

---

## 1.4 Identifying the state

> **State = the minimum information needed to describe a subproblem uniquely.**

Test: if two recursive calls would receive the same state arguments, they must produce the same answer. If they don't, your state is under-specified — add a dimension.

For Fibonacci, the state is just $i$. For House Robber II we will see that the state needs an extra "skipped first?" flag, encoded by running the DP twice.

---

## 1.5 Identifying the transition

> **Transition = how the answer at state $s$ combines answers at strictly smaller states.**

Ask: *"What is the last decision I made to arrive at state $s$?"* Enumerate all possible last decisions; each yields a smaller state. Combine them with the appropriate operator (next section).

---

## 1.6 The counting / optimization / existence trichotomy

The **shape** of the recurrence changes with the question:

| Question type | Operator | Base case | Example |
|---|---|---|---|
| **Optimization** (min/max) | $\min$ / $\max$ | $\infty$ / $-\infty$ or a known optimum | Min path sum, LIS, Knapsack value |
| **Counting** (how many ways) | $+$ (sum) | $1$ for the empty/trivial way, $0$ otherwise | Climbing Stairs ways, Coin Change 2 |
| **Existence** (yes/no) | $\lor$ (OR) | `true` for trivial, `false` otherwise | Subset Sum, Word Break |

**Same state, same transitions — different combinator.** This is the single most useful generalization in DP.

---

## 1.7 DP on a DAG — the unifying view

Build a graph $G = (V, E)$ where:
- $V$ = the set of all states
- $(u \to v) \in E$ iff state $v$ appears in the recurrence of state $u$ (i.e., $u$ depends on $v$)

Because the recurrence only references *strictly smaller* states, $G$ is a **DAG**. Then:

- **Optimization DP** = shortest/longest path on the DAG
- **Counting DP** = number of paths on the DAG
- **Existence DP** = reachability on the DAG

Topological order of the DAG = the iteration order of tabulation. This unification will be referenced repeatedly in later subtopics.

## 1.8 Fibonacci — four implementations

We implement four versions to make the duality concrete:
1. **Naive recursion** — exponential, just to confirm the overlap is real.
2. **Top-down memoization** — $O(n)$ time, $O(n)$ space (table + stack).
3. **Bottom-up tabulation** — $O(n)$ time, $O(n)$ space, no stack.
4. **Space-optimized** — $O(n)$ time, $O(1)$ space.

### Why $O(1)$ space optimization is valid

The recurrence $dp[i] = dp[i-1] + dp[i-2]$ depends **only on the previous two states**. Once $dp[i]$ is computed, $dp[i-2]$ is never referenced again. We only need to keep a sliding window of size 2. This is the prototype of every constant-space optimization in this notebook.

````c++
#include <iostream>
#include <vector>
#include <cstdint>
#include <climits>     // INT_MAX
#include <cstdlib>     // abs
#include <algorithm>   // min, max
using namespace std;

// ─── Version 1: Naive recursion ──────────────────────────────────────────────
// Exponential time. Included to make the overlap real.
uint64_t fib_naive(int i) {
    if (i < 2) return i;                              // base: F0=0, F1=1
    return fib_naive(i - 1) + fib_naive(i - 2);       // exactly the recurrence
}

// ─── Version 2: Top-down memoization ─────────────────────────────────────────
// dp[i] = i-th Fibonacci number, or -1 if not yet computed (sentinel).
uint64_t fib_memo_helper(int i, vector<int64_t>& dp) {
    if (i < 2) return i;                              // base case
    if (dp[i] != -1) return dp[i];                    // already solved → reuse
    return dp[i] = fib_memo_helper(i-1, dp)           // store before returning
                 + fib_memo_helper(i-2, dp);          // memoization invariant
}
uint64_t fib_memo(int n) {
    vector<int64_t> dp(n + 1, -1);                    // -1 = "uncomputed"
    return fib_memo_helper(n, dp);
}

// ─── Version 3: Bottom-up tabulation ─────────────────────────────────────────
// Fill dp[0], dp[1], ..., dp[n] in increasing index order (topo order of DAG).
uint64_t fib_tab(int n) {
    if (n < 2) return n;
    vector<uint64_t> dp(n + 1);
    dp[0] = 0; dp[1] = 1;                             // base cases written explicitly
    for (int i = 2; i <= n; ++i)
        dp[i] = dp[i-1] + dp[i-2];                    // direct recurrence
    return dp[n];
}

// ─── Version 4: Space-optimized ──────────────────────────────────────────────
// Only dp[i-1] and dp[i-2] are ever referenced; keep a sliding window of 2.
uint64_t fib_opt(int n) {
    if (n < 2) return n;
    uint64_t prev2 = 0, prev1 = 1;                    // represent dp[i-2], dp[i-1]
    for (int i = 2; i <= n; ++i) {
        uint64_t cur = prev1 + prev2;                 // dp[i]
        prev2 = prev1;                                // shift window forward
        prev1 = cur;
    }
    return prev1;                                     // prev1 now holds dp[n]
}
````

````c++
// ─── Tests: Fibonacci ────────────────────────────────────────────────────────
auto test_fib = []() {
    struct T { int n; uint64_t expected; };
    vector<T> ts = {
        {0, 0}, {1, 1}, {2, 1}, {5, 5},
        {10, 55}, {30, 832040}, {50, 12586269025ULL}
    };
    for (auto& t : ts) {
        uint64_t a = fib_opt(t.n);
        cout << "n=" << t.n << " -> " << a
             << " (Expected: " << t.expected << ")"
             << (a == t.expected ? " ok" : " FAIL") << "\n";
    }
    // Boundary: cross-check memo vs tab vs opt on small n
    cout << "Cross-check at n=20: memo=" << fib_memo(20)
         << " tab=" << fib_tab(20) << " opt=" << fib_opt(20) << "\n";
};
test_fib();
````

# Subtopic 2 — 1D DP

All problems here share a common structure: **the state is a single index $i$**, and the recurrence references a small constant number of strictly smaller indices. They are all path-counting / max-weight problems on a 1D DAG.

## 2.1 Climbing Stairs

> You are climbing a staircase of $n$ steps. Each move you can climb either 1 or 2 steps. How many distinct ways can you reach step $n$?

### State (invariant)

$$dp[i] = \text{number of distinct ways to reach step } i \text{ starting from step } 0.$$

### Recurrence

To reach step $i$, the **last** move was either a 1-step (from $i-1$) or a 2-step (from $i-2$). These two cases partition all ways into disjoint classes, so by the sum rule:

$$
dp[i] =
\begin{cases}
1 & i = 0 \quad \text{(empty path)} \\
1 & i = 1 \quad \text{(single 1-step)} \\
dp[i-1] + dp[i-2] & i \geq 2
\end{cases}
$$

### Base cases — why?

- $dp[0] = 1$ — there is exactly **one** way to be at the start: do nothing. (If we instead set $dp[0] = 0$, then $dp[2] = dp[1] + dp[0] = 1 + 0 = 1$, which is wrong — there are 2 ways: 1+1 or 2.)
- $dp[1] = 1$ — the single 1-step.

### Boundary transitions table

| $i$ | Came from $i-1$ via 1-step | Came from $i-2$ via 2-step | $dp[i]$ |
|---|---|---|---|
| 0 | — | — | 1 (empty) |
| 1 | $dp[0]=1$ | invalid ($i-2<0$) | 1 |
| 2 | $dp[1]=1$ | $dp[0]=1$ | 2 |
| 3 | $dp[2]=2$ | $dp[1]=1$ | 3 |
| 4 | $dp[3]=3$ | $dp[2]=2$ | 5 |

### Why it works

- **Optimal substructure:** every path to $i$ ends with a 1-step or a 2-step; the prefix is itself a valid path to $i-1$ or $i-2$.
- **Overlapping subproblems:** same as Fibonacci — exponential repetition without memoization.

### Δ from Fibonacci

Recurrence is *identical* to Fibonacci, but the **base case is shifted**: $dp[0] = 1$ here (counting the empty path) vs $F_0 = 0$. The trichotomy is the reason: Climbing Stairs is a *counting* problem, Fibonacci is a *definition*. The shape of the recurrence is the same DAG; the question we ask of it differs.

### Complexity

- Time: $\Theta(n)$
- Space: $\Theta(n)$ → optimized to $\Theta(1)$ (only $dp[i-1], dp[i-2]$ needed).

````c++
// Climbing Stairs — three flavors

// Top-down memoization
int climb_memo_h(int i, vector<int>& dp) {
    if (i <= 1) return 1;                             // dp[0]=dp[1]=1
    if (dp[i] != -1) return dp[i];
    return dp[i] = climb_memo_h(i-1, dp) + climb_memo_h(i-2, dp);
}
int climbStairs_memo(int n) {
    vector<int> dp(n + 1, -1);
    return climb_memo_h(n, dp);
}

// Bottom-up tabulation
int climbStairs_tab(int n) {
    if (n <= 1) return 1;
    vector<int> dp(n + 1);
    dp[0] = dp[1] = 1;                                // base cases (see note above)
    for (int i = 2; i <= n; ++i)
        dp[i] = dp[i-1] + dp[i-2];
    return dp[n];
}

// Space-optimized: only the last two states are referenced
int climbStairs_opt(int n) {
    if (n <= 1) return 1;
    int prev2 = 1, prev1 = 1;                         // dp[0], dp[1]
    for (int i = 2; i <= n; ++i) {
        int cur = prev1 + prev2;                      // dp[i]
        prev2 = prev1;                                // slide window
        prev1 = cur;
    }
    return prev1;
}
````

````c++
// Tests: Climbing Stairs
auto test_climb = []() {
    struct T { int n; int expected; };
    vector<T> ts = {
        {0, 1},   // empty path: 1 way
        {1, 1},   // single 1-step
        {2, 2},   // (1+1), (2)
        {3, 3},   // (1+1+1), (1+2), (2+1)
        {4, 5},
        {5, 8},
        {10, 89}
    };
    for (auto& t : ts) {
        int a = climbStairs_opt(t.n);
        cout << "n=" << t.n << " -> " << a
             << " (Expected: " << t.expected << ")"
             << (a == t.expected ? " ok" : " FAIL") << "\n";
    }
};
test_climb();
````

## 2.2 Frog Jump (1 or 2 steps with cost)

> A frog is on stair $0$ and wants to reach stair $n-1$. From stair $i$, it can jump to stair $i+1$ or stair $i+2$. The energy cost of a jump from $i$ to $j$ is $|h[i] - h[j]|$. Minimize total energy.

### State (invariant)

$$dp[i] = \text{minimum total energy to reach stair } i \text{ starting from stair } 0.$$

### Recurrence

The last jump landed on $i$ from either $i-1$ or $i-2$:

$$
dp[i] = \min\bigl(\, dp[i-1] + |h[i] - h[i-1]|,\; dp[i-2] + |h[i] - h[i-2]| \,\bigr) \quad \text{for } i \geq 2
$$

### Base cases — why?

- $dp[0] = 0$ — already at stair 0, no energy spent.
- $dp[1] = |h[1] - h[0]|$ — only one way to reach stair 1 (one 1-jump).

### Boundary transitions table

| $i$ | From $i-1$ (cost) | From $i-2$ (cost) | $dp[i]$ |
|---|---|---|---|
| 0 | — | — | 0 |
| 1 | $dp[0] + \lvert h[1]-h[0]\rvert$ | invalid | only the $i-1$ option |
| $\geq 2$ | $dp[i-1] + \lvert h[i]-h[i-1]\rvert$ | $dp[i-2] + \lvert h[i]-h[i-2]\rvert$ | $\min$ of both |

### Δ from Climbing Stairs

Same DAG, **different combinator**: counting ($+$) → optimization ($\min$). The transition edges now carry **weights** ($|h[i] - h[j]|$), so this is the shortest-path-on-a-DAG view in its purest form.

### Why it works

- **Optimal substructure:** the optimal path to $i$ must use an optimal path to whichever predecessor ($i-1$ or $i-2$) it came from. Otherwise we could swap in a better prefix and get a strictly smaller total — contradiction.
- **Overlapping subproblems:** every stair $j < i$ is reached by multiple recursive expansions.

### Complexity

- Time: $\Theta(n)$
- Space: $\Theta(n)$ → optimized to $\Theta(1)$.

````c++
// Frog Jump — bottom-up and space-optimized
int frogJump_tab(const vector<int>& h) {
    int n = (int)h.size();
    if (n == 1) return 0;
    vector<int> dp(n);
    dp[0] = 0;                                        // base: at start
    dp[1] = abs(h[1] - h[0]);                         // base: only one predecessor
    for (int i = 2; i < n; ++i) {
        int one = dp[i-1] + abs(h[i] - h[i-1]);       // came from i-1
        int two = dp[i-2] + abs(h[i] - h[i-2]);       // came from i-2
        dp[i] = min(one, two);
    }
    return dp[n-1];
}

// Space-optimized: only dp[i-1], dp[i-2] are referenced
int frogJump_opt(const vector<int>& h) {
    int n = (int)h.size();
    if (n == 1) return 0;
    int prev2 = 0;                                    // dp[0]
    int prev1 = abs(h[1] - h[0]);                     // dp[1]
    for (int i = 2; i < n; ++i) {
        int one = prev1 + abs(h[i] - h[i-1]);
        int two = prev2 + abs(h[i] - h[i-2]);
        int cur = min(one, two);                      // dp[i]
        prev2 = prev1;
        prev1 = cur;
    }
    return prev1;
}
````

````c++
// Tests: Frog Jump
auto test_frog = []() {
    struct T { vector<int> h; int expected; };
    vector<T> ts = {
        {{10}, 0},                                    // single stair
        {{10, 20}, 10},                               // forced single jump
        {{10, 30, 40, 20}, 30},                       // 10→30→20: |20|+|10|=30
        {{30, 10, 60, 10, 60, 50}, 40},               // classic Striver example
        {{7, 7, 7, 7}, 0},                            // all-same: zero cost
        {{0, 0, 0, 0, 0}, 0}                          // boundary: all zeros
    };
    for (auto& t : ts) {
        int a = frogJump_opt(t.h);
        cout << "h=[";
        for (size_t k = 0; k < t.h.size(); ++k)
            cout << t.h[k] << (k+1<t.h.size() ? "," : "");
        cout << "] -> " << a << " (Expected: " << t.expected << ")"
             << (a == t.expected ? " ok" : " FAIL") << "\n";
    }
};
test_frog();
````

## 2.3 Frog Jump with K Distance

> Same as before, but the frog can jump to any of $i+1, i+2, \ldots, i+k$.

### State (invariant)

$$dp[i] = \text{minimum total energy to reach stair } i \text{ from stair } 0.$$

### Recurrence

The last jump landed on $i$ from any $j \in \{i-1, i-2, \ldots, i-k\}$ with $j \geq 0$:

$$
dp[i] = \min_{\substack{1 \leq s \leq k \\ i - s \geq 0}} \bigl(\, dp[i - s] + |h[i] - h[i - s]| \,\bigr) \quad \text{for } i \geq 1
$$

with $dp[0] = 0$.

### Δ from Frog Jump (k = 2)

The hard-coded two-branch recurrence becomes a **loop over $s = 1 \ldots k$**. This is the prototype for every "variable number of transitions" recurrence (Unbounded Knapsack, Coin Change, etc. — to be seen in later subtopics).

### Boundary transitions table

| Condition | Effect on the loop |
|---|---|
| $i - s < 0$ | Skip this $s$ (jump would come from before stair 0) |
| $i - s \geq 0$ | Candidate cost: $dp[i-s] + \lvert h[i] - h[i-s]\rvert$ |

### Complexity

- Time: $\Theta(n \cdot k)$ — outer loop over $i$, inner loop over $s$.
- Space: $\Theta(n)$.

### Why no $O(1)$ space optimization?

The recurrence depends on the last $k$ values, not the last 2. To remove space we would need a sliding window of size $k$, which is still $\Theta(k)$ space. The $O(1)$ trick only works when the dependency window is a **fixed small constant**.

````c++
// Frog Jump with K Distance
int frogJumpK(int k, const vector<int>& h) {
    int n = (int)h.size();
    if (n == 1) return 0;
    const int INF = INT_MAX;
    vector<int> dp(n, INF);
    dp[0] = 0;                                        // base: at start
    for (int i = 1; i < n; ++i) {
        for (int s = 1; s <= k; ++s) {                // try all jump sizes 1..k
            int j = i - s;
            if (j < 0) break;                         // jumps from before 0 invalid
            if (dp[j] == INF) continue;               // unreachable predecessor
            int cost = dp[j] + abs(h[i] - h[j]);
            if (cost < dp[i]) dp[i] = cost;           // take the best last jump
        }
    }
    return dp[n-1];
}
````

````c++
// Tests: Frog Jump K
auto test_frogK = []() {
    struct T { int k; vector<int> h; int expected; };
    vector<T> ts = {
        {1, {10, 20}, 10},                            // forced k=1
        {2, {10, 30, 40, 20}, 30},                    // matches k=2 case
        {3, {10, 30, 40, 50, 20}, 30},                // 10→30→20? no — try paths
        {4, {40, 10, 20, 70, 80, 10, 20, 70, 80, 60}, 40},
        {2, {7, 7, 7, 7}, 0},                         // all-same: zero
        {3, {30}, 0},                                 // single stair
        {5, {10, 5, 20, 0, 15}, 5}                    // any k≥n−1 lets one big jump
    };
    for (auto& t : ts) {
        int a = frogJumpK(t.k, t.h);
        cout << "k=" << t.k << " h=[";
        for (size_t i = 0; i < t.h.size(); ++i)
            cout << t.h[i] << (i+1<t.h.size() ? "," : "");
        cout << "] -> " << a << " (Expected: " << t.expected << ")"
             << (a == t.expected ? " ok" : " FAIL") << "\n";
    }
};
test_frogK();
````

## 2.4 Maximum Sum of Non-Adjacent Elements (House Robber I)

> Given an array $a[0 \ldots n-1]$ of non-negative integers, pick a subset such that **no two picked indices are adjacent**, maximizing the sum.

### State (invariant)

$$dp[i] = \text{maximum sum achievable using elements from } a[0..i] \text{ subject to the no-two-adjacent rule.}$$

### Recurrence

At index $i$ we either **pick** $a[i]$ or **skip** it:

- **Pick:** then $a[i-1]$ cannot be picked, so we add $a[i]$ to the best answer up to $i-2$: $a[i] + dp[i-2]$.
- **Skip:** the best answer up to $i$ is the best answer up to $i-1$: $dp[i-1]$.

$$
dp[i] =
\begin{cases}
a[0] & i = 0 \\
\max(a[0], a[1]) & i = 1 \\
\max\bigl(a[i] + dp[i-2],\; dp[i-1]\bigr) & i \geq 2
\end{cases}
$$

### Base cases — why?

- $dp[0] = a[0]$ — only one element, best is itself.
- $dp[1] = \max(a[0], a[1])$ — two elements, adjacent → can pick at most one, pick the larger.

### Boundary transitions table

| Decision | Recurrence contribution | Why it's valid |
|---|---|---|
| Pick $a[i]$ | $a[i] + dp[i-2]$ | Skipping $i-1$ keeps non-adjacency |
| Skip $a[i]$ | $dp[i-1]$ | $dp[i-1]$ already optimal under the constraint |

### Δ from Frog Jump

Frog Jump *had to* land on every stair eventually (you must reach stair $n-1$). Here you may *skip* indices — the "skip" transition is what makes it a subset problem. Despite that, the dependency window is still just $\{i-1, i-2\}$, so $O(1)$ space is possible.

### Why it works

- **Optimal substructure:** an optimal non-adjacent subset of $a[0..i]$ either uses $a[i]$ (then the remainder is an optimal subset of $a[0..i-2]$) or doesn't (then it's an optimal subset of $a[0..i-1]$). Exchange argument: if either prefix were suboptimal, we could swap it in and improve, contradiction.
- **Overlapping subproblems:** the same $dp[j]$ is referenced by both $dp[j+1]$ and $dp[j+2]$.

### Complexity

- Time: $\Theta(n)$
- Space: $\Theta(n)$ → optimized to $\Theta(1)$.

````c++
// House Robber I — bottom-up and space-optimized

int rob_tab(const vector<int>& a) {
    int n = (int)a.size();
    if (n == 0) return 0;
    if (n == 1) return a[0];
    vector<int> dp(n);
    dp[0] = a[0];                                     // base
    dp[1] = max(a[0], a[1]);                          // base: pick larger of first two
    for (int i = 2; i < n; ++i) {
        int pick = a[i] + dp[i-2];                    // pick a[i] → skip a[i-1]
        int skip = dp[i-1];                           // skip a[i] → carry dp[i-1]
        dp[i] = max(pick, skip);
    }
    return dp[n-1];
}

// Space-optimized: dependency is only on dp[i-1], dp[i-2]
int rob_opt(const vector<int>& a) {
    int n = (int)a.size();
    if (n == 0) return 0;
    if (n == 1) return a[0];
    int prev2 = a[0];                                 // dp[0]
    int prev1 = max(a[0], a[1]);                      // dp[1]
    for (int i = 2; i < n; ++i) {
        int pick = a[i] + prev2;                      // a[i] + dp[i-2]
        int skip = prev1;                             // dp[i-1]
        int cur  = max(pick, skip);                   // dp[i]
        prev2 = prev1;
        prev1 = cur;
    }
    return prev1;
}
````

````c++
// Tests: House Robber I
auto test_rob = []() {
    struct T { vector<int> a; int expected; };
    vector<T> ts = {
        {{}, 0},                                      // empty
        {{5}, 5},                                     // single
        {{2, 7}, 7},                                  // two, pick larger
        {{1, 2, 3, 1}, 4},                            // pick 1+3
        {{2, 7, 9, 3, 1}, 12},                        // pick 2+9+1
        {{2, 1, 4, 5, 3, 1, 1, 3}, 12},               // pick 2+4+3+3
        {{0, 0, 0, 0}, 0},                            // all zeros
        {{100, 1, 1, 100}, 200}                       // 100 + 100
    };
    for (auto& t : ts) {
        int a = rob_opt(t.a);
        cout << "a=[";
        for (size_t k = 0; k < t.a.size(); ++k)
            cout << t.a[k] << (k+1<t.a.size() ? "," : "");
        cout << "] -> " << a << " (Expected: " << t.expected << ")"
             << (a == t.expected ? " ok" : " FAIL") << "\n";
    }
};
test_rob();
````

## 2.5 House Robber II (Circular)

> Same as House Robber I, but the houses are arranged in a **circle**: $a[0]$ and $a[n-1]$ are now adjacent.

### Δ from House Robber I

A single 1D state $dp[i]$ no longer captures the constraint. The wrap-around adjacency creates a coupling: whether we pick $a[0]$ affects whether $a[n-1]$ is eligible. Adding a "did we pick the first?" boolean would work but doubles the state. There is a cleaner trick.

### The two-independent-runs reduction

**Claim.** Let $\mathrm{run}(L, R)$ = House Robber I applied to the contiguous sub-array $a[L..R]$. Then for the circular problem on $a[0..n-1]$:

$$\text{answer} = \max\bigl(\,\mathrm{run}(0, n-2),\; \mathrm{run}(1, n-1)\,\bigr)$$

### Proof of correctness

Every valid circular selection $S$ satisfies *at least one* of:
1. $0 \notin S$ — then $S \subseteq \{1, \ldots, n-1\}$. No wrap-around adjacency can fire because $0$ is excluded. So the linear non-adjacency on $a[1..n-1]$ already captures the constraint. $\mathrm{run}(1, n-1)$ covers this case optimally.
2. $n-1 \notin S$ — then $S \subseteq \{0, \ldots, n-2\}$. Same argument: $\mathrm{run}(0, n-2)$ covers it.

**Are cases 1 and 2 jointly exhaustive?** Yes. The only way to violate both is $0 \in S \land n-1 \in S$, but $0$ and $n-1$ are adjacent in the circle — that selection is *infeasible*. So every **feasible** selection falls in case 1 or case 2 (possibly both, if neither endpoint is picked).

**Does the $\max$ miss anything?** Every feasible $S$ is counted by at least one run, and each run only counts feasible selections (it is the linear non-adjacency optimum on a sub-array), so $\max(\mathrm{run}_1, \mathrm{run}_2)$ ≥ value of every feasible $S$. And the optimum is itself feasible, so it is achieved by one of the runs. ✓

### Boundary transitions table

| $n$ | Handling |
|---|---|
| 0 | return 0 |
| 1 | return $a[0]$ (no adjacency to worry about) |
| 2 | return $\max(a[0], a[1])$ — circular adjacency forces picking at most one |
| $\geq 3$ | $\max(\mathrm{run}(0, n-2),\; \mathrm{run}(1, n-1))$ |

The $n \leq 2$ cases need special handling because $\mathrm{run}(0, n-2)$ would be an empty range when $n = 1$ and the two runs overlap weirdly when $n = 2$.

### Complexity

- Time: $\Theta(n)$ — two linear passes.
- Space: $\Theta(1)$ if each run is space-optimized.

````c++
// House Robber II — run the O(1)-space linear robber on two sub-ranges

// Linear robber on a[L..R] (inclusive), space-optimized.
// Treats the range as a fresh 1D problem.
int rob_linear(const vector<int>& a, int L, int R) {
    if (L > R) return 0;                              // empty range
    if (L == R) return a[L];                          // single element
    int prev2 = a[L];                                 // dp[L]
    int prev1 = max(a[L], a[L+1]);                    // dp[L+1]
    for (int i = L + 2; i <= R; ++i) {
        int pick = a[i] + prev2;                      // pick → skip a[i-1]
        int skip = prev1;                             // skip → carry
        int cur  = max(pick, skip);
        prev2 = prev1;
        prev1 = cur;
    }
    return prev1;
}

int rob2(const vector<int>& a) {
    int n = (int)a.size();
    if (n == 0) return 0;
    if (n == 1) return a[0];                          // no adjacency
    if (n == 2) return max(a[0], a[1]);               // circle of 2: pick one
    int excludeLast  = rob_linear(a, 0, n - 2);       // case: n-1 not in S
    int excludeFirst = rob_linear(a, 1, n - 1);       // case: 0 not in S
    return max(excludeLast, excludeFirst);
}
````

````c++
// Tests: House Robber II
auto test_rob2 = []() {
    struct T { vector<int> a; int expected; };
    vector<T> ts = {
        {{}, 0},                                      // empty
        {{5}, 5},                                     // single
        {{2, 3}, 3},                                  // two
        {{2, 3, 2}, 3},                               // circle of 3: pick middle
        {{1, 2, 3, 1}, 4},                            // same as linear here: 2+1? no, 1+3=4
        {{1, 2, 3}, 3},                               // pick max single
        {{200, 3, 140, 20, 10}, 340},                 // 200+140=340 (excl last)
        {{0, 0, 0}, 0},                               // all zeros
        {{100, 1, 1, 100}, 101}                       // circle: can't pick both 100s
    };
    for (auto& t : ts) {
        int a = rob2(t.a);
        cout << "a=[";
        for (size_t k = 0; k < t.a.size(); ++k)
            cout << t.a[k] << (k+1<t.a.size() ? "," : "");
        cout << "] -> " << a << " (Expected: " << t.expected << ")"
             << (a == t.expected ? " ok" : " FAIL") << "\n";
    }
};
test_rob2();
````

# Unified Mental Model — Subtopics 1 & 2

All five problems share a single skeleton. Differences are localized to (a) the **combinator** (sum/min/max), (b) the **transition set**, (c) the **base case**.

## The skeleton in code form

````c++
// ─── DP skeleton for every 1D problem in this notebook ──────────────────────
//
// for i from base+1 to n:
//     dp[i] = COMBINE over (s : transition_from_i) of  EDGE_COST(i, s) + dp[s]
//
// COMBINE ∈ { +, min, max }      ← chosen by the trichotomy
// transition_from_i               ← the set of predecessor states
// EDGE_COST                       ← 0 for counting, |h[i]-h[s]| for frog, -a[i] for robber, ...

// Fibonacci / Climbing Stairs  (counting, edges to {i-1, i-2}, edge cost 0)
//     dp[i] = dp[i-1] + dp[i-2]

// Frog Jump (optimization, edges to {i-1, i-2}, edge cost |h[i]-h[s]|)
//     dp[i] = min(dp[i-1] + |h[i]-h[i-1]|, dp[i-2] + |h[i]-h[i-2]|)

// Frog Jump K (optimization, edges to {i-1, ..., i-k}, edge cost |h[i]-h[s]|)
//     dp[i] = min over s in 1..k of  dp[i-s] + |h[i]-h[i-s]|

// House Robber I (optimization, two "decisions" per i: pick or skip)
//     dp[i] = max(a[i] + dp[i-2],   // pick
//                  dp[i-1])          // skip

// House Robber II (House Robber I run twice on sub-arrays, take max)
//     ans = max(rob_linear(a, 0, n-2), rob_linear(a, 1, n-1))
````

## Decision tree — which 1D pattern to apply

Read these in order. The first match is the pattern.

```
Q1. Is the answer "how many ways"?
    └─ YES → COUNTING DP (sum combinator). Base case = 1 for the empty/trivial path.
    └─ NO  → continue.

Q2. Is the answer "minimum/maximum cost (or value)"?
    └─ YES → OPTIMIZATION DP (min/max combinator). Base case = identity element (0 or ±∞).
    └─ NO  → continue.

Q3. Is the answer "does there exist a valid …"?
    └─ YES → EXISTENCE DP (OR combinator). Base case = true for trivial, false otherwise.

Q4. From state i, what predecessor states does the recurrence reference?
    ├─ {i-1, i-2}                          → 2-step pattern   → O(1) space achievable
    ├─ {i-1, i-2, ..., i-k} with fixed k   → k-step pattern   → O(k) space achievable
    ├─ at most one of {i-1, i-2}, decided
       by a per-i pick/skip decision        → House-Robber-shape → O(1) space achievable
    └─ all of {0, ..., i-1}                 → LIS-shape         → O(n) space (Subtopic 7)

Q5. Is the array circular (a[0] adjacent to a[n-1])?
    └─ YES → run the linear DP twice on a[0..n-2] and a[1..n-1], take max.

Q6. Are the transition edges weighted (e.g., |h[i] - h[j]|)?
    └─ YES → write the recurrence as "previous-state value + edge cost", combine with min/max.
    └─ NO  → just sum the previous-state values (counting case).
```

## Complexity summary

| Problem | Time | Space | Optimized Space | Key Insight |
|---|---|---|---|---|
| Fibonacci | $\Theta(n)$ | $\Theta(n)$ | $\Theta(1)$ | Window of 2 — prototype of all $O(1)$ optimizations |
| Climbing Stairs | $\Theta(n)$ | $\Theta(n)$ | $\Theta(1)$ | Same DAG as Fibonacci; counting combinator + shifted base case |
| Frog Jump | $\Theta(n)$ | $\Theta(n)$ | $\Theta(1)$ | Shortest path on a 1D weighted DAG; $\min$ combinator |
| Frog Jump K | $\Theta(nk)$ | $\Theta(n)$ | $\Theta(k)$ | Loop over $s \in 1..k$ inside; window of size $k$ |
| House Robber I | $\Theta(n)$ | $\Theta(n)$ | $\Theta(1)$ | Pick/skip decision at every $i$; subset (not path) DP |
| House Robber II | $\Theta(n)$ | $\Theta(n)$ | $\Theta(1)$ | Two runs reduce circular to linear; case-split on which endpoint to exclude |

---

## Three takeaways to carry forward

1. **State first, recurrence second, combinator third.** Pin down what $dp[i]$ *means* in English before writing any math. Pin down the combinator (sum/min/max/OR) before writing any code.
2. **The $O(1)$ space optimization is valid iff the recurrence references a fixed constant window of previous states.** This will repeat in 2D DP (column-by-column reduction) and in 0/1 Knapsack (two-row reduction).
3. **Every DP in this notebook is shortest/longest/count path on a DAG of states.** The 1D problems use a chain DAG; later subtopics use grids, lattices, and interval DAGs.
