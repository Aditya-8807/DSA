# Sliding Window

Sliding Window works when the **search space has a monotonic feasibility structure**.

Specifically, the window $[l, r]$ over a sequence satisfies:

- A **constraint function** $f([l,r])$ is monotone in window size
- Expanding the window (increasing $r$) makes the constraint *harder to satisfy* (or easier — depends on the variant)
- Shrinking the window (increasing $l$) makes it *easier* (or harder)

This monotonicity is what lets you avoid $O(n^2)$ brute force.

Sliding Window finds the **optimal window** under a constraint — either the longest valid window, the shortest valid window, or counts all valid windows.

## The Fundamental Invariant

---

### The Two-Pointer Geometry

At any moment, the window is defined by two indices:
$$[l, r] \quad 0 \leq l \leq r < n$$

The **window state** is some aggregate $S$ of $arr[l..r]$ — a sum, a frequency map, a count of distinct elements, etc.

The core invariant (for the *fixed constraint* variant):
$$\text{Invariant}: S([l, r]) \text{ satisfies the constraint} \quad \forall \text{ iterations}$$

For the *find optimal* variant:
$$\text{Invariant}: l \text{ is the leftmost index such that } [l, r] \text{ is valid}$$

---

### Why $O(n)$? — The Amortized Argument

Naively, two nested pointers feel like $O(n^2)$. The key insight:

> Each pointer ($l$ and $r$) moves **strictly right**, at most $n$ times total.

Therefore: total pointer movements $\leq 2n \Rightarrow O(n)$ amortized.

**Formal bound:** Let $\text{mov}(r)$ = total right-moves of $r$, $\text{mov}(l)$ = total right-moves of $l$.
$$\text{mov}(r) \leq n, \quad \text{mov}(l) \leq n$$
$$\Rightarrow \text{Total operations} = O(n)$$

This breaks if $l$ ever moves **left** — which is why sliding window only works with monotonic constraints.

## Category 1 — Fixed-Size Window

---

### Problem Shape

Window size $k$ is fixed. Slide it across the array, compute something at each position.

### Invariant
$$\text{window} = [i - k + 1,\; i] \quad \text{as } i \text{ sweeps } [0, n)$$

State update at step $i$:
$$S_i = S_{i-1} - arr[i - k] + arr[i]$$

This is the **sliding aggregate** — add the new right element, remove the old left element. $O(1)$ per step.

### Boundary Condition

| Phase | Condition | Action |
|---|---|---|
| Warmup | $i < k - 1$ | Add `arr[i]` to state, do NOT record answer |
| Steady | $i \geq k - 1$ | Add `arr[i]`, remove `arr[i-k]`, record answer |

Alternatively: process the first window $[0, k-1]$ separately, then slide.

### The Monotonic Deque Upgrade

For **window max/min**, a naive approach is $O(nk)$. The fix: maintain a **monotonic deque** that stores *indices* of elements in decreasing order (for max).

**Deque Invariant:** `deque.front()` always holds the index of the maximum element in the current window.

Operations:
- **Add `arr[i]`:** Pop from back while `arr[deque.back()] <= arr[i]` (they're dominated — can never be the max)
- **Expire old elements:** Pop from front while `deque.front() <= i - k`
- **Query:** `arr[deque.front()]`

Each element is pushed/popped at most once $\Rightarrow O(n)$ total.

````c++
#include <iostream>
#include <vector>
#include <deque>
#include <numeric>
#include <climits>

using namespace std;
````

````c++
// ─── Fixed Window: Maximum Sum Subarray of size k ────────────────────────
// Invariant: window sum S = sum(arr[i-k+1 .. i]) at each step i >= k-1
// State update: S += arr[i] - arr[i-k]  →  O(1) per step
int maxSumFixedWindow(const vector<int>& arr, int k) {
    int n = (int)arr.size();
    if (n < k) return -1;

    // Warmup: compute sum of first window [0, k-1]
    int windowSum = 0;
    for (int i = 0; i < k; i++) windowSum += arr[i];

    int maxSum = windowSum;

    // Slide: at each step, add arr[i], remove arr[i-k]
    for (int i = k; i < n; i++) {
        windowSum += arr[i] - arr[i - k]; // O(1) slide
        maxSum = max(maxSum, windowSum);
    }
    return maxSum;
}

// ─── Fixed Window: Sliding Window Maximum (monotonic deque) ───────────────
// Deque invariant: indices in deque are in DECREASING order of arr values.
// deque.front() = index of max in current window.
// Each index pushed and popped at most once → O(n) total.
vector<int> slidingWindowMaximum(const vector<int>& arr, int k) {
    int n = (int)arr.size();
    vector<int> result;
    deque<int> dq; // stores INDICES, not values

    for (int i = 0; i < n; i++) {
        // 1. Expire: remove indices outside current window [i-k+1, i]
        while (!dq.empty() && dq.front() <= i - k)
            dq.pop_front();

        // 2. Maintain decreasing order: pop dominated elements from back
        //    arr[dq.back()] <= arr[i] means dq.back() can NEVER be max → discard
        while (!dq.empty() && arr[dq.back()] <= arr[i])
            dq.pop_back();

        dq.push_back(i);

        // 3. Record answer once first full window is formed
        if (i >= k - 1)
            result.push_back(arr[dq.front()]);
    }
    return result;
}
````

````c++
// Test Fixed Window
vector<int> arr1 = {2, 1, 5, 1, 3, 2};
cout << "Max sum (k=3): " << maxSumFixedWindow(arr1, 3) << " (Expected: 9)\n"; // [5,1,3]

vector<int> arr2 = {1, 3, -1, -3, 5, 3, 6, 7};
vector<int> maxWin = slidingWindowMaximum(arr2, 3);
cout << "Sliding max:   ";
for (int v : maxWin) cout << v << " "; // 3 3 5 5 6 7
cout << "(Expected: 3 3 5 5 6 7)\n";
````

## Category 2 — Variable-Size Window: Longest Valid

---

### Problem Shape

Find the **longest** subarray satisfying a constraint. The constraint becomes harder to satisfy as the window grows.

**Examples:** Longest subarray with sum $\leq k$, longest substring with at most $k$ distinct characters, longest substring without repeating characters.

### The Monotonicity Argument

Define $\text{valid}(l, r) = 1$ if $[l,r]$ satisfies the constraint, else $0$.

For *longest valid* problems:
$$\text{valid}(l, r) = 1 \Rightarrow \text{valid}(l', r) = 1 \quad \forall l' > l$$

i.e., shrinking a valid window keeps it valid. This is the **monotonicity** that makes the two-pointer work.

### The Skeleton

```
l = 0, state = empty
for r in [0, n):
    add arr[r] to state          // expand right — always
    while state is INVALID:      // shrink left until valid
        remove arr[l] from state
        l++
    // Invariant: [l, r] is now valid (or empty)
    ans = max(ans, r - l + 1)    // record best
```

### Boundary Transitions

| Event | Action | Why |
|---|---|---|
| Expand (`r++`) | Add `arr[r]` to state | Try extending the window |
| Constraint violated | Enter `while` loop | Window is invalid; must shrink |
| Shrink (`l++`) | Remove `arr[l]` from state | Eliminate leftmost element |
| `while` exits | `[l, r]` is valid | Restore invariant |
| Record | `ans = max(ans, r - l + 1)` | Current window is the best seen so far |

**The delta — `while` vs `if`:**
- Use `while`: when violating one element could require removing *multiple* from the left (e.g., distinct character count > k)
- Use `if`: only when adding one element breaks the constraint by exactly one unit (rare)

Always default to `while` — it's correct in all cases.

````c++
// ─── Longest subarray with sum <= k ──────────────────────────────────────
// State: running sum. Monotone: larger window → larger sum → harder to satisfy.
// while sum > k: shrink from left
int longestSubarraySumAtMostK(const vector<int>& arr, int k) {
    int n = (int)arr.size();
    int l = 0, sum = 0, ans = 0;
    for (int r = 0; r < n; r++) {
        sum += arr[r];                     // expand
        while (sum > k && l <= r) {        // shrink until valid
            sum -= arr[l];
            l++;
        }
        ans = max(ans, r - l + 1);         // [l,r] is valid
    }
    return ans;
}

// ─── Longest substring with at most k distinct characters ─────────────────
// State: frequency map. Constraint: distinct count <= k.
// while distinct > k: remove arr[l], shrink
int longestSubstringKDistinct(const string& s, int k) {
    int n = (int)s.size();
    int freq[128] = {}; // ASCII frequency map
    int distinct = 0, l = 0, ans = 0;
    for (int r = 0; r < n; r++) {
        if (freq[(int)s[r]]++ == 0) distinct++; // new character
        while (distinct > k) {                   // shrink
            if (--freq[(int)s[l]] == 0) distinct--;
            l++;
        }
        ans = max(ans, r - l + 1);
    }
    return ans;
}

// ─── Longest substring without repeating characters ───────────────────────
// Special case of above with k = 26 (all distinct), but we detect repeat directly.
// State: last seen index. Constraint: no char appears twice.
// When duplicate found: jump l to max(l, lastSeen[c] + 1) — O(1) shrink
int longestUniqueSubstring(const string& s) {
    int n = (int)s.size();
    int lastSeen[128];
    fill(lastSeen, lastSeen + 128, -1);
    int l = 0, ans = 0;
    for (int r = 0; r < n; r++) {
        int c = (int)s[r];
        // If s[r] was seen inside current window [l,r], jump l past it
        if (lastSeen[c] >= l) l = lastSeen[c] + 1;
        lastSeen[c] = r;
        ans = max(ans, r - l + 1);
    }
    return ans;
}
````

````c++
// Test Variable Window — Longest Valid
vector<int> a = {1, 2, 1, 0, 1, 1, 0};
cout << "Longest sum<=3:  " << longestSubarraySumAtMostK(a, 3) << " (Expected: 4)\n"; // [1,0,1,1]

string s1 = "eceba";
cout << "K-distinct (k=2): " << longestSubstringKDistinct(s1, 2) << " (Expected: 3)\n"; // ece

string s2 = "abcabcbb";
cout << "No repeat:        " << longestUniqueSubstring(s2) << " (Expected: 3)\n"; // abc

string s3 = "pwwkew";
cout << "No repeat:        " << longestUniqueSubstring(s3) << " (Expected: 3)\n"; // wke
````

## Category 3 — Variable-Size Window: Shortest Valid

---

### Problem Shape

Find the **shortest** subarray satisfying a constraint. The constraint becomes *easier* to satisfy as the window grows — so you expand until valid, then shrink aggressively.

**Examples:** Minimum window substring, smallest subarray with sum $\geq k$.

### The Flipped Monotonicity

$$\text{valid}(l, r) = 1 \Rightarrow \text{valid}(l, r') = 1 \quad \forall r' > r$$

i.e., growing a valid window stays valid. So once $[l,r]$ is valid: record it, then **shrink from left** to find a potentially shorter valid window.

### The Skeleton

```
l = 0, state = empty, ans = INT_MAX
for r in [0, n):
    add arr[r] to state              // expand right
    while state is VALID:            // shrink left while still valid
        ans = min(ans, r - l + 1)    // record BEFORE shrinking
        remove arr[l] from state
        l++
```

### Contrast with Longest Valid

| | Longest Valid | Shortest Valid |
|---|---|---|
| `while` condition | `state INVALID` | `state VALID` |
| Record answer | After `while` (window valid) | Inside `while` (before shrink) |
| `ans` init | `0` | `INT_MAX` |
| Monotonicity | Larger window → harder to satisfy | Larger window → easier to satisfy |

**The critical order:** In shortest valid, record `ans` **before** removing `arr[l]`. If you remove first, you record an invalid window.

````c++
// ─── Smallest subarray with sum >= k ─────────────────────────────────────
// State: running sum. Valid when sum >= k.
// Once valid: record, then shrink left to find shorter valid windows.
int minSubarraySumAtLeastK(const vector<int>& arr, int k) {
    int n = (int)arr.size();
    int l = 0, sum = 0, ans = INT_MAX;
    for (int r = 0; r < n; r++) {
        sum += arr[r];                          // expand
        while (sum >= k) {                      // shrink while still valid
            ans = min(ans, r - l + 1);          // record FIRST (window is valid now)
            sum -= arr[l];
            l++;
        }
    }
    return (ans == INT_MAX) ? 0 : ans;
}

// ─── Minimum Window Substring ─────────────────────────────────────────────
// State: frequency map + count of characters still "needed".
// Valid when need == 0 (all chars of t are covered).
// Classic two-map approach.
string minWindowSubstring(const string& s, const string& t) {
    int freq[128] = {}; // how many of each char t requires
    for (char c : t) freq[(int)c]++;

    int need = (int)t.size(); // chars still needed in window
    int l = 0, bestL = -1, bestLen = INT_MAX;

    for (int r = 0; r < (int)s.size(); r++) {
        // Expand: include s[r]
        if (--freq[(int)s[r]] >= 0) need--; // freq[c] was > 0 → it was needed → one less needed

        while (need == 0) {                  // window covers all of t
            if (r - l + 1 < bestLen) {       // record
                bestLen = r - l + 1;
                bestL = l;
            }
            // Shrink: remove s[l]
            if (++freq[(int)s[l]] > 0) need++; // freq[c] goes back positive → we need it again
            l++;
        }
    }
    return (bestL == -1) ? "" : s.substr(bestL, bestLen);
}
````

````c++
// Test Variable Window — Shortest Valid
vector<int> b = {2, 3, 1, 2, 4, 3};
cout << "Min subarray sum>=7: " << minSubarraySumAtLeastK(b, 7) << " (Expected: 2)\n"; // [4,3]

cout << "Min window (ADOBECODEBANC, ABC): "
     << minWindowSubstring("ADOBECODEBANC", "ABC") << " (Expected: BANC)\n";

cout << "Min window (a, a): "
     << minWindowSubstring("a", "a") << " (Expected: a)\n";
````

## Category 4 — Count of Valid Windows (At Most K → Exactly K)

---

### The Reduction

Counting subarrays with **exactly** $k$ of something is harder to do directly. The standard reduction:

$$\text{exactly}(k) = \text{atMost}(k) - \text{atMost}(k-1)$$

This works because $\text{atMost}(k)$ is a quantity you *can* compute with a sliding window.

### Count with atMost(k) — The Counting Skeleton

For each valid window $[l, r]$, **all subarrays ending at $r$ with left endpoint in $[l, r]$ are valid**. Count = $r - l + 1$.

```
l = 0, count = 0, ans = 0
for r in [0, n):
    add arr[r] to state
    while state exceeds k:         // shrink
        remove arr[l]
        l++
    ans += r - l + 1               // count subarrays ending at r: [l,r],[l+1,r],...,[r,r]
```

**Why $r - l + 1$?** After shrinking, every left endpoint from $l$ to $r$ produces a valid subarray ending at $r$. That's exactly $r - l + 1$ subarrays.

### Example: Subarrays with at most k odd numbers

Then **exactly k odd numbers** = atMost(k) - atMost(k-1).

````c++
// ─── Count subarrays with at most k odd numbers ───────────────────────────
// State: count of odd numbers in window. Constraint: count <= k.
// For each r, all windows [l..r],[l+1..r],...,[r..r] are valid → add r-l+1.
long long countAtMostKOdd(const vector<int>& arr, int k) {
    int n = (int)arr.size();
    int l = 0, odds = 0;
    long long ans = 0;
    for (int r = 0; r < n; r++) {
        if (arr[r] % 2 != 0) odds++;         // expand: new odd?
        while (odds > k) {                    // shrink until valid
            if (arr[l] % 2 != 0) odds--;
            l++;
        }
        ans += (r - l + 1);                   // all subarrays ending at r are valid
    }
    return ans;
}

// ─── Count subarrays with EXACTLY k odd numbers ───────────────────────────
// Reduction: exactly(k) = atMost(k) - atMost(k-1)
long long countExactlyKOdd(const vector<int>& arr, int k) {
    if (k == 0) return countAtMostKOdd(arr, 0);
    return countAtMostKOdd(arr, k) - countAtMostKOdd(arr, k - 1);
}
````

````c++
// Test Counting
vector<int> c = {1, 1, 2, 1, 1};
cout << "Exactly 3 odd subarrays: " << countExactlyKOdd(c, 3) << " (Expected: 2)\n";
// [1,1,2,1] and [1,2,1,1]

vector<int> d = {2, 4, 6};
cout << "Exactly 1 odd subarrays: " << countExactlyKOdd(d, 1) << " (Expected: 0)\n";
````

## The Unified Mental Model

---

```
FIXED window (size k)
    → slide aggregate: S += arr[r] - arr[r-k]
    → for max/min: monotonic deque, O(n)

VARIABLE window — LONGEST valid
    → expand always, shrink while INVALID
    → record AFTER while loop
    → ans = max(ans, r - l + 1)

VARIABLE window — SHORTEST valid
    → expand always, shrink while VALID
    → record INSIDE while loop, BEFORE shrink
    → ans = min(ans, r - l + 1)

COUNT — at most K
    → shrink while INVALID (like longest)
    → ans += r - l + 1  (count all valid subarrays ending at r)

COUNT — exactly K
    → exactly(K) = atMost(K) - atMost(K-1)
```

---

## Key Implementation Notes

- **`while` not `if`** in the shrink step — always. One expansion can cause multiple violations (e.g., frequency map).
- **State must be $O(1)$ or $O(\Sigma)$ to update** — if updating state on add/remove is $O(k)$, the total complexity degrades to $O(nk)$.
- **The monotonic deque** stores *indices*, not values — necessary to check expiry.
- **Minimum window substring:** The `need` counter is the key — decrement only when `freq[c]` was positive (char was required), increment only when it goes positive again (char is now missing).
- **Exactly K reduction** works when the `atMost` function is well-defined and monotone in $k$.
- **Left pointer never goes back** — if your problem requires $l$ to decrease, sliding window does not apply.

## Complexity Summary

| Variant | Time | Space | Key State |
|---|---|---|---|
| Fixed window sum | $O(n)$ | $O(1)$ | Running sum |
| Fixed window max/min | $O(n)$ | $O(k)$ | Monotonic deque |
| Longest valid | $O(n)$ | $O(\Sigma)$ | Freq map or counter |
| Shortest valid | $O(n)$ | $O(\Sigma)$ | Freq map + `need` |
| Count at most K | $O(n)$ | $O(\Sigma)$ | Freq map or counter |
| Count exactly K | $O(n)$ | $O(\Sigma)$ | Two `atMost` calls |

**Amortized argument (formal):**
Let $T_r$ = total right-steps of $r$ = $n$. Let $T_l$ = total right-steps of $l \leq n$.
Inner `while` body executes at most $T_l$ times total across all outer loop iterations.
$$\Rightarrow \text{Total work} = T_r + T_l \leq 2n = O(n)$$
