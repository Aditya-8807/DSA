## 8.1 Shortest Path in Undirected Graph with Unit Weights (BFS Baseline)

### State definition
$\text{dist}[v]$ = minimum number of edges from a given source $s$ to $v$ — identical to Subtopic
2.1's BFS distance, restated here as the entry point into this subtopic's broader shortest-path
landscape.

### Invariant
Subtopic 2.1's layer invariant, unchanged: BFS dequeues vertices in non-decreasing distance order,
which is optimal precisely because every edge has identical weight 1.

### Why it works
Already proven in full in Subtopic 2.1. Restated here as the **baseline this entire subtopic
generalizes away from**: every algorithm that follows (8.2–8.6) exists specifically to handle a
situation where BFS's core assumption — *all edges cost the same* — no longer holds.

### Boundary transitions table

| Decision point | Condition | Action |
|---|---|---|
| All edges have identical (unit) cost | true | BFS is optimal and sufficient — no need for anything heavier |
| Edges have varying non-negative costs | — | BFS breaks (see 8.3's Dijkstra); need priority-based relaxation |
| Edges can be negative | — | Dijkstra breaks too (see 8.3's failure proof); need Bellman-Ford (8.5) |
| Graph is a DAG with varying costs | — | topo-sort + relaxation is strictly better than Dijkstra (8.2) |

### The delta
The non-obvious insight framing the whole subtopic: **every shortest-path algorithm here is BFS
with its core assumption relaxed by exactly one degree** — DAG shortest path relaxes "no ordering
assumption" into "we have one, for free, via topo sort"; Dijkstra relaxes "unit weight" into
"non-negative weight, priority-ordered"; Bellman-Ford relaxes "non-negative" into "no negative
cycle"; Floyd-Warshall relaxes "single source" into "all sources." Recognizing *which* assumption a
problem violates is the entire skill this subtopic teaches.


````c++
// Shortest Path, Undirected Unit Weights: BFS baseline, restated as the entry point for Subtopic 8
#include <iostream>
#include <vector>
#include <queue>
using namespace std;
#define vi vector<int>
#define vvi vector<vector<int>>

vi shortestPathUnitWeights(int V, vvi& adj, int src) {
    vi dist(V, -1);
    queue<int> q;
    dist[src] = 0;
    q.push(src);
    while (!q.empty()) {
        int u = q.front(); q.pop();
        for (int v : adj[u]) if (dist[v] == -1) { dist[v] = dist[u] + 1; q.push(v); }
    }
    return dist;
}
````

````c++
// Test cell: input -> actual (Expected: X)

{
    vvi adj(5);
    for (int i=0;i<4;i++){adj[i].push_back(i+1); adj[i+1].push_back(i);}
    vi dist = shortestPathUnitWeights(5, adj, 0);
    cout << "shortestPathUnitWeights(path, from 0) dist[4] -> " << dist[4] << " (Expected: 4)\n";
}
{
    vvi adj(1);
    vi dist = shortestPathUnitWeights(1, adj, 0);
    cout << "shortestPathUnitWeights(single node) dist[0] -> " << dist[0] << " (Expected: 0)\n";
}
{
    vvi adj(3); adj[0]={1}; adj[1]={0};   // 2 unreachable
    vi dist = shortestPathUnitWeights(3, adj, 0);
    cout << "shortestPathUnitWeights(disconnected) dist[2] -> " << dist[2] << " (Expected: -1)\n";
}
````

## 8.2 Shortest Path in DAG (Topo Sort + Relaxation)

### State definition
$\text{dist}[v]$ = minimum total edge weight from source $s$ to $v$, over a **weighted DAG**
(weights may be arbitrary, including negative — this algorithm does not require non-negativity).

### Invariant
Process vertices in **topological order** (Subtopic 7). Relax every outgoing edge of $u$
immediately when $u$ is processed:
$$
\text{dist}[v] = \min\big(\text{dist}[v],\ \text{dist}[u] + w(u,v)\big) \quad \text{for each edge } (u,v)
$$
The invariant: **by the time $u$ is processed, $\text{dist}[u]$ is already final** — because every
predecessor of $u$ (every vertex with an edge into $u$) appears *earlier* in topological order
(that is the definition of a valid topo order, Subtopic 7.1), and has therefore already relaxed its
edge into $u$.

### Why it works — proof that a single topo-order pass suffices (no negative-weight caveat needed)

**Claim:** processing vertices in topological order and relaxing each vertex's outgoing edges
exactly once produces correct shortest distances, even with negative edge weights.

*Proof (by induction on topo-order position):* the source $s$ has $\text{dist}[s]=0$, trivially
correct (no earlier vertex can contribute, and this is the empty-path base case). Suppose every
vertex before $u$ in topo order has final, correct distance. Every edge $(p, u)$ into $u$ has $p$
appearing before $u$ in topo order (by the topo-order definition itself) — so by the inductive
hypothesis, $\text{dist}[p]$ is already finalized and correctly relaxed into $u$ by the time $u$ is
reached. Since **every** predecessor of $u$ has already contributed its best possible relaxation
before $u$ is processed, $\text{dist}[u]$ is final and correct the moment $u$ itself is processed —
no predecessor can arrive "later" and improve it further, because topo order guarantees no
predecessor comes after $u$. $\blacksquare$ **Why negative weights are safe here specifically:**
Dijkstra's proof (8.3) crucially relies on processing the globally-nearest unprocessed vertex next
so that no cheaper path can appear later — a guarantee that breaks under negative weights. Here,
correctness instead relies purely on topo order (a structural/ordering guarantee, not a
distance-based one), which is completely unaffected by edge sign.

### Boundary transitions table

| Decision point | Condition | Action |
|---|---|---|
| Vertex unreachable from source | $\text{dist}[v]$ never updated from $\infty$ | remains $\infty$ / unreachable, correctly |
| Edge relaxation | $\text{dist}[u] + w(u,v) < \text{dist}[v]$ | update $\text{dist}[v]$ |
| Vertex before source in topo order | irrelevant to source's reachable subgraph | dist stays $\infty$, no effect |

### The delta
The non-obvious insight: this is the **only** shortest-path algorithm in this subtopic that
handles negative weights in $O(V+E)$ **without needing $|V|-1$ repeated passes** (contrast
Bellman-Ford, 8.5) — the speedup is entirely because a DAG's topo order already provides the
"process in an order that guarantees no re-improvement" guarantee for free, which Bellman-Ford has
to painstakingly re-derive via repeated relaxation on general graphs.


````c++
// Shortest Path in DAG: topo sort (Kahn's) + single-pass relaxation, correct even with negative weights
#include <iostream>
#include <vector>
#include <queue>
#include <climits>
using namespace std;
#define vi vector<int>
#define pii pair<int,int>
#define vpii vector<pair<int,int>>
#define vvpii vector<vector<pii>>

vi shortestPathDAG(int V, vvpii& adj, int src) {
    // Kahn's topo sort (Subtopic 7.2), reused directly.
    vi indeg(V, 0);
    for (int u = 0; u < V; u++) for (auto& [v, w] : adj[u]) indeg[v]++;
    queue<int> q;
    for (int v = 0; v < V; v++) if (indeg[v] == 0) q.push(v);
    vi topoOrder;
    while (!q.empty()) {
        int u = q.front(); q.pop();
        topoOrder.push_back(u);
        for (auto& [v, w] : adj[u]) if (--indeg[v] == 0) q.push(v);
    }

    vi dist(V, INT_MAX);
    dist[src] = 0;             // base case: source reaches itself at cost 0
    for (int u : topoOrder) {
        if (dist[u] == INT_MAX) continue;   // unreachable from src -- nothing to relax
        for (auto& [v, w] : adj[u]) {
            if (dist[u] + w < dist[v]) dist[v] = dist[u] + w;   // relaxation, safe due to topo-order invariant
        }
    }
    return dist;
}
````

````c++
// Test cell: input -> actual (Expected: X)

// Standard weighted DAG
{
    vvpii adj(6);
    adj[0] = {{1,2},{4,1}};
    adj[1] = {{2,3}};
    adj[2] = {{3,6}};
    adj[3] = {{5,1}};
    adj[4] = {{2,2},{5,4}};
    vi dist = shortestPathDAG(6, adj, 0);
    cout << "shortestPathDAG dist[2] -> " << dist[2] << " (Expected: 3, via 0->4->2 costing 1+2=3, cheaper than 0->1->2=5)\n";
}

// Negative edge weight in a DAG -- must still work correctly (proof depended on topo order, not sign)
{
    vvpii adj(3);
    adj[0] = {{1,5}};
    adj[1] = {{2,-3}};
    vi dist = shortestPathDAG(3, adj, 0);
    cout << "shortestPathDAG(negative edge) dist[2] -> " << dist[2] << " (Expected: 2, 5 + (-3))\n";
}

// Unreachable vertex
{
    vvpii adj(3);
    adj[0] = {{1,1}};
    vi dist = shortestPathDAG(3, adj, 0);
    cout << "shortestPathDAG(unreachable) dist[2] -> " << (dist[2] == INT_MAX ? -1 : dist[2]) << " (Expected: -1, meaning unreachable)\n";
}

// Single node
{
    vvpii adj(1);
    vi dist = shortestPathDAG(1, adj, 0);
    cout << "shortestPathDAG(single node) dist[0] -> " << dist[0] << " (Expected: 0)\n";
}
````

## 8.3 Dijkstra's Algorithm (Priority-Queue Version)

### State definition
$\text{dist}[v]$ = shortest total edge weight from source $s$ to $v$, over a graph with
**non-negative** edge weights (this restriction is load-bearing — see the failure proof below).
A vertex is **finalized** once extracted from the priority queue for the first time.

### Invariant — "extracted node has final distance"
At the moment a vertex $u$ is extracted from the priority queue (the min-heap, keyed by current
tentative distance), $\text{dist}[u]$ is **already the true shortest distance** from $s$ to $u$,
and will never be improved again.

### Why it works — formal proof

**Claim:** when $u$ is extracted (popped) from the min-priority-queue, $\text{dist}[u]$ is optimal.

*Proof (by induction on extraction order):* the source $s$ is extracted first with
$\text{dist}[s]=0$, trivially optimal (no shorter path than the empty path exists, assuming
non-negative weights so no path can have negative total cost). Suppose every vertex extracted
before $u$ has correct final distance. Suppose for contradiction some shorter path
$s \rightsquigarrow u$ exists with true cost $< \text{dist}[u]$ (the value used to extract $u$).
That path must cross from the "already finalized" set to the "not yet finalized" set at some edge
$(p, q)$ — where $p$ is finalized (correct dist by inductive hypothesis) and $q$ is the first
not-yet-finalized vertex on the path. Since $p$ is finalized, $q$'s tentative distance was already
updated to at most $\text{dist}[p] + w(p,q)$ when $p$ was processed. Because **all edge weights are
non-negative**, the remaining segment of the path from $q$ to $u$ can only add more cost, so
$\text{dist}[p] + w(p,q) \le$ (true cost of the alleged shorter path) $< \text{dist}[u]$. But then
$q$'s tentative distance is strictly less than $u$'s, meaning the min-priority-queue would have
extracted $q$ before $u$ — contradicting that $u$ was extracted at this step while $q$ remains
unfinalized. Hence no such shorter path exists, and $\text{dist}[u]$ is optimal at extraction.
$\blacksquare$

**Why non-negative weights are essential — concrete counterexample.** Consider $s \to a$ with
weight 10, $s \to b$ with weight 1, $b \to a$ with weight $-100$. Dijkstra extracts $b$ first
(distance 1, smaller than $a$'s tentative 10), finalizing $\text{dist}[b]=1$ — correct so far. It
then extracts $a$ with tentative distance $\min(10, 1 + (-100)) = -99$ *if* the relaxation from $b$
happened before $a$'s extraction — but if $a$ was already extracted (finalized at distance 10)
**before** $b$ was processed (e.g. if edge weights were examined in a different order, or more
generally in any graph shape where a negative edge is discovered too late), the negative edge
$b \to a$ can never retroactively fix $a$'s already-finalized, wrong distance of 10. The core proof
above breaks specifically at "the remaining segment can only add more cost" — with negative
weights, this is false, so a vertex extracted "early" (high preliminary distance) can later be
proven non-optimal by a negative edge discovered afterward, which Dijkstra structurally cannot
revisit.

### Boundary transitions table

| Decision point | Condition | Action |
|---|---|---|
| Vertex popped, already finalized (stale heap entry) | seen before with a better/equal distance | skip — lazy deletion, avoids reprocessing |
| Edge relaxation improves tentative distance | $\text{dist}[u] + w(u,v) < \text{dist}[v]$ | update, push $(new\ dist, v)$ onto the heap |
| Negative edge weight present anywhere | any $w(u,v) < 0$ | Dijkstra's correctness proof breaks — must use Bellman-Ford (8.5) instead |

### The delta
The non-obvious insight: Dijkstra is BFS (Subtopic 2.1) with the FIFO queue replaced by a
min-priority-queue — the *entire* proof structure (extracted-vertex-has-final-value, by
contradiction via "a shorter path must cross the finalized boundary") is identical in shape to
BFS's layer-invariant proof; the only substantive change is that "non-decreasing distance dequeue
order" is now *enforced directly* by the heap instead of emerging naturally from FIFO+unit-weights.
This is exactly why unit-weight BFS is a special case of Dijkstra (a priority queue over
all-equal keys degenerates to FIFO order).


````c++
// Dijkstra's Algorithm: priority-queue version, extracted-vertex-has-final-distance invariant
#include <iostream>
#include <vector>
#include <queue>
#include <climits>
using namespace std;
#define vi vector<int>
#define pii pair<int,int>
#define vpii vector<pair<int,int>>
#define vvpii vector<vector<pii>>

vi dijkstra(int V, vvpii& adj, int src) {
    vi dist(V, INT_MAX);
    // min-heap keyed by (tentative distance, vertex) -- pair comparison naturally orders by distance first
    priority_queue<pii, vpii, greater<pii>> pq;
    dist[src] = 0;
    pq.push({0, src});

    while (!pq.empty()) {
        auto [d, u] = pq.top(); pq.pop();
        if (d > dist[u]) continue;    // stale heap entry (u already finalized with a better distance) -- skip
        for (auto& [v, w] : adj[u]) {
            if (dist[u] + w < dist[v]) {   // relaxation
                dist[v] = dist[u] + w;
                pq.push({dist[v], v});
            }
        }
    }
    return dist;
}
````

````c++
// Test cell: input -> actual (Expected: X)

// Standard weighted graph
{
    vvpii adj(5);
    adj[0] = {{1,4},{2,1}};
    adj[1] = {{0,4},{3,1}};
    adj[2] = {{0,1},{1,2},{3,5}};
    adj[3] = {{1,1},{2,5},{4,3}};
    adj[4] = {{3,3}};
    vi dist = dijkstra(5, adj, 0);
    cout << "dijkstra dist[4] -> " << dist[4] << " (Expected: 7, via 0->2->1->3->4 = 1+2+1+3)\n";
}

// Unreachable vertex
{
    vvpii adj(3);
    adj[0] = {{1,5}};
    vi dist = dijkstra(3, adj, 0);
    cout << "dijkstra(unreachable) dist[2] -> " << (dist[2]==INT_MAX ? -1 : dist[2]) << " (Expected: -1)\n";
}

// Single node
{
    vvpii adj(1);
    vi dist = dijkstra(1, adj, 0);
    cout << "dijkstra(single node) dist[0] -> " << dist[0] << " (Expected: 0)\n";
}

// Graph with a zero-weight edge (boundary of "non-negative")
{
    vvpii adj(3);
    adj[0] = {{1,0},{2,5}};
    adj[1] = {{2,1}};
    vi dist = dijkstra(3, adj, 0);
    cout << "dijkstra(zero-weight edge) dist[2] -> " << dist[2] << " (Expected: 1, via 0->1->2 costing 0+1)\n";
}

// Dense graph (complete graph), all weights equal -- should match BFS result on unit weights
{
    vvpii adj(4);
    for (int i=0;i<4;i++) for (int j=0;j<4;j++) if (i!=j) adj[i].push_back({j,1});
    vi dist = dijkstra(4, adj, 0);
    cout << "dijkstra(K4, unit weights) dist[3] -> " << dist[3] << " (Expected: 1)\n";
}
````

## 8.4 Why Priority Queue Is Used — Comparison With the Naive $O(V^2)$ Array Version

### The naive array version
Instead of a heap, maintain `dist[]` as a plain array, and on each of $V$ iterations, **linearly
scan** all $V$ vertices to find the unfinalized one with the smallest tentative distance.

```
for i in 1..V:
    u = argmin over unfinalized v of dist[v]     // O(V) scan
    finalize u
    for each edge (u, v): relax                   // O(deg(u)) total, sums to O(E) across all iterations
```

### Complexity comparison

| Version | "Find minimum" cost | Total for $V$ extractions | Relaxation cost | Total |
|---|---|---|---|---|
| Naive array | $O(V)$ per extraction | $O(V^2)$ | $O(E)$ total | $O(V^2 + E)$ |
| Binary-heap priority queue | $O(\log V)$ per extraction | $O(V \log V)$ | $O(E \log V)$ (each relaxation may push, $O(\log V)$ per push) | $O((V+E)\log V)$ |

### Why each wins where it does

**Dense graphs** ($E = \Theta(V^2)$): $O(V^2+E) = O(V^2)$ for the array version vs.
$O((V+E)\log V) = O(V^2 \log V)$ for the heap version — the array version is asymptotically
**better** here, because the $\log V$ factor on a large $E$ outweighs the saved factor on the
cheap-relative-to-$E$ minimum-finding step.

**Sparse graphs** ($E = O(V)$): array version is $O(V^2)$ regardless of how few edges exist (the
linear scan for minimum dominates, unavoidably, every single iteration) — while the heap version is
$O((V+E)\log V) = O(V \log V)$, strictly better for large sparse $V$.

### Proof that both compute the same correct distances
Both versions maintain the exact same finalize-then-relax structure and rely on the exact same
non-negative-weight proof from 8.3 (extracted/selected vertex has final distance) — the priority
queue is purely an *implementation* choice for "find the next vertex to finalize efficiently," not
a change to the algorithm's logic or correctness argument. Swapping between them never changes the
answer, only the running time.

### Boundary transitions table

| Decision point | Condition | Choose |
|---|---|---|
| Graph is dense, $E = \Theta(V^2)$ | — | naive $O(V^2)$ array version (heap's $\log V$ overhead isn't worth it) |
| Graph is sparse, $E = O(V \log V)$ or smaller | — | binary-heap priority queue |
| Using a Fibonacci heap (advanced, rarely implemented in practice) | — | $O(E + V\log V)$, theoretically optimal but high constant factors, rarely worth it outside specialized settings |

### The delta
The non-obvious insight: **"always use a priority queue for Dijkstra" is a heuristic, not a law**
— it is the right default because most competitive-programming and interview graphs are sparse, but
the actual decision criterion is a direct comparison of $V^2$ against $(V+E)\log V$ for the specific
graph's density, exactly the same kind of representation trade-off reasoning introduced back in
Subtopic 1.2 for adjacency matrix vs. list.


````c++
// Naive O(V^2) Dijkstra: linear scan for minimum, no heap -- correctness identical to 8.3's heap version
#include <iostream>
#include <vector>
#include <climits>
using namespace std;
#define vi vector<int>
#define vb vector<bool>
#define pii pair<int,int>
#define vpii vector<pair<int,int>>
#define vvpii vector<vector<pii>>

vi dijkstraNaive(int V, vvpii& adj, int src) {
    vi dist(V, INT_MAX);
    vb finalized(V, false);
    dist[src] = 0;

    for (int iter = 0; iter < V; iter++) {
        // O(V) linear scan for the unfinalized vertex with smallest tentative distance
        int u = -1;
        for (int v = 0; v < V; v++) {
            if (!finalized[v] && (u == -1 || dist[v] < dist[u])) u = v;
        }
        if (u == -1 || dist[u] == INT_MAX) break;   // remaining vertices are all unreachable
        finalized[u] = true;                         // same "extracted -> final" invariant as the heap version

        for (auto& [v, w] : adj[u]) {
            if (!finalized[v] && dist[u] + w < dist[v]) dist[v] = dist[u] + w;
        }
    }
    return dist;
}
````

````c++
// Test cell: input -> actual (Expected: X) -- cross-checked directly against 8.3's heap-based dijkstra()

{
    vvpii adj(5);
    adj[0] = {{1,4},{2,1}};
    adj[1] = {{0,4},{3,1}};
    adj[2] = {{0,1},{1,2},{3,5}};
    adj[3] = {{1,1},{2,5},{4,3}};
    adj[4] = {{3,3}};
    vi distNaive = dijkstraNaive(5, adj, 0);
    vi distHeap = dijkstra(5, adj, 0);
    cout << "dijkstraNaive dist[4] -> " << distNaive[4] << " (Expected: 7, matches heap version)\n";
    cout << "Cross-check all vertices match heap version -> ";
    bool allMatch = true;
    for (int i = 0; i < 5; i++) if (distNaive[i] != distHeap[i]) allMatch = false;
    cout << allMatch << " (Expected: 1)\n";
}

// Unreachable vertex, naive version
{
    vvpii adj(3);
    adj[0] = {{1,5}};
    vi dist = dijkstraNaive(3, adj, 0);
    cout << "dijkstraNaive(unreachable) dist[2] -> " << (dist[2]==INT_MAX ? -1 : dist[2]) << " (Expected: -1)\n";
}

// Single node
{
    vvpii adj(1);
    vi dist = dijkstraNaive(1, adj, 0);
    cout << "dijkstraNaive(single node) dist[0] -> " << dist[0] << " (Expected: 0)\n";
}
````

## 8.5 Bellman-Ford

### State definition
$\text{dist}[v]$ = shortest total edge weight from source $s$ to $v$, over a graph that **may
have negative edge weights** (but is checked for negative cycles, which make "shortest path"
undefined).

### Invariant — the core relaxation invariant
$$
\text{After } k \text{ iterations of relaxing every edge}, \quad
\text{dist}[v] = \text{shortest path from } s \text{ to } v \text{ using at most } k \text{ edges.}
$$
This is the precise generalization Dijkstra's proof could not offer under negative weights: instead
of "extracted vertex is final" (which relies on non-negative weights to know no cheaper path can
appear later), Bellman-Ford makes no ordering assumption at all — it simply relaxes **every** edge,
**every** iteration, guaranteeing that after $k$ rounds, all paths using $\le k$ edges have been
considered.

### Why it works — proof by induction on path edge-count

**Claim:** after $k$ full passes (each pass relaxes every edge once), $\text{dist}[v]$ equals the
true shortest-path cost among all paths from $s$ to $v$ using at most $k$ edges.

*Proof (induction on $k$):* **Base case** $k=0$: $\text{dist}[s]=0$ (path using 0 edges), all
others $\infty$ (no 0-edge path reaches them) — correct. **Inductive step:** assume after $k-1$
passes, $\text{dist}[v]$ is correct for all shortest paths using $\le k-1$ edges. Consider the true
shortest path to $v$ using $\le k$ edges; let $u$ be the second-to-last vertex on it (so the path
is $s \rightsquigarrow u \to v$, using $\le k-1$ edges to reach $u$). By the inductive hypothesis,
$\text{dist}[u]$ is already correct (at most $\le k-1$ edges) *before* pass $k$ begins. Pass $k$
relaxes every edge including $(u,v)$, so $\text{dist}[v]$ is updated to at most
$\text{dist}[u] + w(u,v)$, which by construction equals the true shortest $\le k$-edge path cost.
Since relaxation never *increases* a distance, and this is a valid upper bound achieved by pass
$k$, $\text{dist}[v]$ is correct after pass $k$. $\blacksquare$

**Why $|V|-1$ passes suffice:** any simple shortest path (no repeated vertices, since repeating a
vertex could only add cost, assuming no negative cycle — see below) has at most $|V|-1$ edges. So
after $|V|-1$ passes, every shortest path (which by definition is simple, given no negative cycle)
has been fully accounted for.

### Negative cycle detection via one extra relaxation round

**Claim:** if a **$V$-th** relaxation pass (one more than $|V|-1$) still finds an edge that can be
relaxed (i.e. $\text{dist}[u] + w(u,v) < \text{dist}[v]$ for some edge), then $G$ contains a
negative-weight cycle reachable from the source.

*Proof:* by the $|V|-1$-passes-suffice argument, every *simple* shortest path has been fully
captured after $|V|-1$ passes. If a further improvement is still found on pass $V$, the
"path" achieving it must be using more than $|V|-1$ edges without being simple — meaning it
revisits some vertex, i.e. it traverses a cycle. Since revisiting the cycle is only useful (strictly
decreases the distance) if the cycle's total weight is negative, this can only happen if a
negative-weight cycle exists on some path from $s$. $\blacksquare$ (If no negative cycle exists,
$|V|-1$ passes are provably sufficient, so a $V$-th pass finding further improvement is only
possible in the negative-cycle case.)

### Boundary transitions table

| Decision point | Condition | Action |
|---|---|---|
| Edge relaxable during pass $k \le V-1$ | $\text{dist}[u]+w(u,v) < \text{dist}[v]$ | update — this is a genuine shorter path using $\le k$ edges |
| Edge still relaxable on pass $V$ | improvement found beyond $|V|-1$ passes | negative cycle detected — "shortest path" is undefined (can be driven to $-\infty$) |
| Vertex unreachable from source | $\text{dist}[v]$ stays $\infty$ through all passes | correctly remains unreachable |

### The delta
The non-obvious insight: Bellman-Ford's $O(VE)$ cost, compared to Dijkstra's $O((V+E)\log V)$, is
the **direct price of dropping the non-negative-weight assumption** — Dijkstra can safely finalize
one vertex per step because non-negative weights guarantee no future edge can undercut an already-
extracted vertex; Bellman-Ford, admitting negative weights, cannot make that guarantee for any
single vertex early, so it must uniformly re-relax **every** edge, **every** round, until a
provable fixed point ($|V|-1$ rounds) is reached.


````c++
// Bellman-Ford: relax every edge V-1 times, then one more pass to detect negative cycles
#include <iostream>
#include <vector>
#include <climits>
using namespace std;
#define vi vector<int>

struct Edge { int u, v, w; };

// bellmanFord: returns {dist, hasNegativeCycle}.
pair<vi,bool> bellmanFord(int V, vector<Edge>& edges, int src) {
    vi dist(V, INT_MAX);
    dist[src] = 0;

    // |V|-1 passes: after pass k, dist[v] is correct for shortest paths using <= k edges (proof above).
    for (int k = 1; k <= V - 1; k++) {
        for (auto& e : edges) {
            if (dist[e.u] != INT_MAX && dist[e.u] + e.w < dist[e.v]) {
                dist[e.v] = dist[e.u] + e.w;
            }
        }
    }

    // One extra pass: any further relaxation implies a negative cycle reachable from src.
    bool hasNegCycle = false;
    for (auto& e : edges) {
        if (dist[e.u] != INT_MAX && dist[e.u] + e.w < dist[e.v]) {
            hasNegCycle = true;
            break;
        }
    }

    return {dist, hasNegCycle};
}
````

````c++
// Test cell: input -> actual (Expected: X)

// Standard case with a negative edge but no negative cycle
{
    vector<Edge> edges = {{0,1,4},{0,2,1},{2,1,-2},{1,3,3},{2,3,5}};
    auto [dist, hasNeg] = bellmanFord(4, edges, 0);
    cout << "bellmanFord dist[1] -> " << dist[1] << " (Expected: -1, via 0->2->1 costing 1+(-2))\n";
    cout << "bellmanFord hasNegCycle -> " << hasNeg << " (Expected: 0)\n";
}

// Negative cycle present: 0->1 (1), 1->2 (-3), 2->1 (1) forms a cycle of weight -2
{
    vector<Edge> edges = {{0,1,1},{1,2,-3},{2,1,1}};
    auto [dist, hasNeg] = bellmanFord(3, edges, 0);
    cout << "bellmanFord(negative cycle) hasNegCycle -> " << hasNeg << " (Expected: 1)\n";
}

// Unreachable vertex
{
    vector<Edge> edges = {{0,1,5}};
    auto [dist, hasNeg] = bellmanFord(3, edges, 0);
    cout << "bellmanFord(unreachable) dist[2] -> " << (dist[2]==INT_MAX ? -1 : dist[2]) << " (Expected: -1, meaning unreachable)\n";
}

// Single node, no edges
{
    vector<Edge> edges = {};
    auto [dist, hasNeg] = bellmanFord(1, edges, 0);
    cout << "bellmanFord(single node) dist[0] -> " << dist[0] << " (Expected: 0)\n";
}

// Negative edge NOT reachable from source -- must not be falsely flagged as a negative cycle
{
    vector<Edge> edges = {{1,2,-5},{2,1,-5}};   // negative cycle exists, but NOT reachable from src=0
    auto [dist, hasNeg] = bellmanFord(3, edges, 0);
    cout << "bellmanFord(unreachable negative cycle) hasNegCycle -> " << hasNeg << " (Expected: 0, cycle not reachable from source)\n";
}
````

## 8.6 Floyd-Warshall (All-Pairs, DP Framing)

### State definition — dynamic programming formulation
$$
\text{dist}[i][j][k] = \text{shortest path from } i \text{ to } j \text{ using only intermediate vertices from } \{1, \ldots, k\}
$$
This is genuinely a new kind of state compared to every single-source algorithm so far: the "layer"
dimension $k$ is not a distance or a graph vertex directly reached, but a constraint on **which
vertices are allowed to be used as intermediate stepping stones**.

### Invariant — the recurrence
$$
\text{dist}[i][j][k] = \min\Big(\text{dist}[i][j][k-1],\ \ \text{dist}[i][k][k-1] + \text{dist}[k][j][k-1]\Big)
$$
Base case: $\text{dist}[i][j][0] = w(i,j)$ if edge $(i,j)$ exists, $0$ if $i=j$, $\infty$
otherwise — this is "no intermediate vertices allowed," i.e. only the direct edge (or staying put).

### Why it works — proof of the recurrence

**Claim:** the recurrence above correctly computes $\text{dist}[i][j][k]$ for every $k$ from
$0$ to $V$.

*Proof (by induction on $k$):* **Base case** $k=0$ is correct by definition (no intermediates
allowed, so only the direct edge or self-loop applies). **Inductive step:** assume
$\text{dist}[\cdot][\cdot][k-1]$ is correct for all pairs. For $\text{dist}[i][j][k]$, consider the
true shortest $i \to j$ path using intermediates from $\{1,\ldots,k\}$. Either it uses vertex $k$ as
an intermediate, or it doesn't:
- **Doesn't use $k$:** the path only uses intermediates from $\{1,\ldots,k-1\}$, so its cost is
  exactly $\text{dist}[i][j][k-1]$ by the inductive hypothesis.
- **Does use $k$:** since a shortest path never revisits a vertex (assuming no negative cycles),
  $k$ appears exactly once, splitting the path into $i \rightsquigarrow k$ and $k \rightsquigarrow
  j$, **each** using only intermediates from $\{1,\ldots,k-1\}$ (since $k$ itself only appears once,
  at the split point). By the inductive hypothesis, these two sub-paths cost exactly
  $\text{dist}[i][k][k-1]$ and $\text{dist}[k][j][k-1]$ respectively.

Taking the minimum of these two cases gives exactly the recurrence, and since these are the only
two possibilities (either $k$ is used or it isn't), the recurrence is exhaustive and correct.
$\blacksquare$

### Space optimization — dropping the $k$ dimension

**Claim:** the algorithm can be implemented with a 2D array `dist[i][j]`, updated in place as $k$
increases from $1$ to $V$, without needing to store all three dimensions.

*Proof this preserves correctness:* when computing layer $k$'s values, the recurrence only ever
reads $\text{dist}[i][k][k-1]$, $\text{dist}[k][j][k-1]$, and $\text{dist}[i][j][k-1]$ — **none of
these three ever have $k$ as a strict "in-progress-being-overwritten" cell during layer $k$'s
computation**, because $\text{dist}[k][k][k-1] $ (the one cell that could be ambiguous) equals
$\text{dist}[k][k][k]$ trivially (using $k$ itself as an intermediate to reach itself can never
help — it would mean traversing a cycle through $k$, which cannot improve a non-negative — or even
just cycle-free-optimal — shortest path). Therefore overwriting `dist[i][j]` in place, row/column
$k$ included, during the processing of layer $k$, reads only already-finalized-for-this-layer
values, and the loop order `for k, for i, for j` is safe. $\blacksquare$

### Boundary transitions table

| Decision point | Condition | Action |
|---|---|---|
| No edge between $i,j$ directly | — | initialize to $\infty$, not $0$ (must not be confused with a real zero-weight edge) |
| $i = j$ | self-distance | initialize to $0$ (a vertex reaches itself with an empty, zero-cost path) |
| Negative cycle present | $\text{dist}[i][i] < 0$ for some $i$ after the full DP | shortest paths through that cycle are undefined — must detect and report separately |

### The delta
The non-obvious insight: Floyd-Warshall is not "Dijkstra/Bellman-Ford run $V$ times" (though that
would also solve all-pairs shortest paths, at cost $O(V \cdot (V+E)\log V)$ or $O(V^2 E)$
respectively) — it is a **genuinely different algorithmic structure**, a DP over "allowed
intermediate vertex sets" rather than a relaxation-based single-source method repeated. Its
$O(V^3)$ cost is independent of $E$ entirely, which makes it the right choice specifically when the
graph is dense enough that $V^3 \lesssim V \cdot E \log V$, or when all-pairs output is needed and
the implementation simplicity of a triple nested loop is valuable.


````c++
// Floyd-Warshall: all-pairs shortest paths via the "allowed intermediate vertices" DP
#include <iostream>
#include <vector>
#include <climits>
using namespace std;
#define vi vector<int>
#define vvi vector<vector<int>>

const int INF = INT_MAX / 2;   // half of INT_MAX to avoid overflow when summing two INF-ish values

vvi floydWarshall(int V, vvi& weightMatrix) {
    // weightMatrix[i][j] = edge weight if exists, INF if no direct edge, 0 if i==j (base case, k=0 layer)
    vvi dist = weightMatrix;   // copy: dist[i][j] represents dist[i][j][0] initially (base case)

    for (int k = 0; k < V; k++) {          // layer k: allow vertex k as an intermediate
        for (int i = 0; i < V; i++) {
            for (int j = 0; j < V; j++) {
                if (dist[i][k] + dist[k][j] < dist[i][j]) {
                    dist[i][j] = dist[i][k] + dist[k][j];   // recurrence: use k as intermediate, or don't (implicit via min)
                }
            }
        }
    }
    return dist;
}
````

````c++
// Test cell: input -> actual (Expected: X)

// Standard 4-node weighted graph
{
    vvi w(4, vi(4, INF));
    for (int i=0;i<4;i++) w[i][i]=0;
    w[0][1]=5; w[1][2]=3; w[2][3]=1; w[0][3]=20;
    vvi dist = floydWarshall(4, w);
    cout << "floydWarshall dist[0][3] -> " << dist[0][3] << " (Expected: 9, via 0->1->2->3 = 5+3+1, cheaper than direct 20)\n";
}

// Disconnected pair
{
    vvi w(3, vi(3, INF));
    for (int i=0;i<3;i++) w[i][i]=0;
    w[0][1]=1;
    // vertex 2 isolated
    vvi dist = floydWarshall(3, w);
    cout << "floydWarshall(disconnected) dist[0][2] -> " << (dist[0][2]>=INF ? -1 : dist[0][2]) << " (Expected: -1, unreachable)\n";
}

// Single node
{
    vvi w(1, vi(1, 0));
    vvi dist = floydWarshall(1, w);
    cout << "floydWarshall(single node) dist[0][0] -> " << dist[0][0] << " (Expected: 0)\n";
}

// Negative edges but no negative cycle
{
    vvi w(3, vi(3, INF));
    for (int i=0;i<3;i++) w[i][i]=0;
    w[0][1]=4; w[1][2]=-2; w[0][2]=10;
    vvi dist = floydWarshall(3, w);
    cout << "floydWarshall(negative edge) dist[0][2] -> " << dist[0][2] << " (Expected: 2, via 0->1->2 = 4+(-2))\n";
}

// Fully connected small graph, verify triangle inequality holds post-computation
{
    vvi w(3, vi(3, INF));
    for (int i=0;i<3;i++) w[i][i]=0;
    w[0][1]=2; w[1][0]=2; w[1][2]=2; w[2][1]=2; w[0][2]=10; w[2][0]=10;
    vvi dist = floydWarshall(3, w);
    cout << "floydWarshall dist[0][2] -> " << dist[0][2] << " (Expected: 4, via 0->1->2, cheaper than direct 10)\n";
}
````

## 8.7 Applications

Six applications, each stressing a different modification of the core algorithms above. Ordered
from "smallest tweak" to "most structurally different."


### 8.7.1 Path with Minimum Effort (Dijkstra With Modified Relaxation)

**State definition.** Given a grid of heights, the "effort" of a path is the **maximum** absolute
height difference between any two *consecutive* cells on that path (not the sum). Find a path from
top-left to bottom-right minimizing this maximum.
$\text{effort}[v]$ = the minimum possible "path max-edge-weight" from source to $v$, over all
paths.

**Invariant.** Same extracted-vertex-has-final-value invariant as Dijkstra (8.3), but the
relaxation rule is changed from *addition* to *max*:
$$
\text{effort}[v] = \min\big(\text{effort}[v],\ \max(\text{effort}[u],\ |h(u)-h(v)|)\big)
$$

**Why it works.** Dijkstra's correctness proof (8.3) never actually used "+" specifically — it
used only that combining a finalized prefix cost with one more edge produces a **monotonically
non-decreasing** result (so a smaller finalized value can never be beaten later by a path that
"pays" the same or more at every step). $\max(\text{effort}[u], w)$ is monotonically non-decreasing
in both its arguments exactly as addition-with-non-negative-weights is — so the entire extraction
proof carries over unchanged, with $\max$ substituted for $+$ throughout.

**The delta.** This is Dijkstra's proof shown to generalize beyond "+" to **any monotonic
combining operator** — a fact worth recognizing broadly: any "minimize the worst single step,"
"minimize the bottleneck," or "minimize the max" path problem is Dijkstra with the relaxation
operator swapped, not a new algorithm.


````c++
// Path with Minimum Effort: Dijkstra with max() relaxation instead of +
#include <iostream>
#include <vector>
#include <queue>
#include <climits>
#include <cmath>
using namespace std;
#define vi vector<int>
#define vvi vector<vector<int>>

int minimumEffortPath(vvi& heights) {
    int R = heights.size(), C = heights[0].size();
    vvi effort(R, vi(C, INT_MAX));
    priority_queue<tuple<int,int,int>, vector<tuple<int,int,int>>, greater<>> pq;   // (effort, r, c)
    effort[0][0] = 0;
    pq.push({0, 0, 0});
    int dr[4]={-1,1,0,0}, dc[4]={0,0,-1,1};

    while (!pq.empty()) {
        auto [e, r, c] = pq.top(); pq.pop();
        if (e > effort[r][c]) continue;        // stale entry, same lazy-deletion pattern as 8.3
        if (r == R-1 && c == C-1) return e;      // destination finalized -- optimal by the extraction invariant
        for (int d = 0; d < 4; d++) {
            int nr = r+dr[d], nc = c+dc[d];
            if (nr<0||nr>=R||nc<0||nc>=C) continue;
            int newEffort = max(e, abs(heights[nr][nc] - heights[r][c]));   // MAX relaxation, not +
            if (newEffort < effort[nr][nc]) {
                effort[nr][nc] = newEffort;
                pq.push({newEffort, nr, nc});
            }
        }
    }
    return effort[R-1][C-1];
}
````

````c++
// Test cell: input -> actual (Expected: X)

{
    vvi h = {{1,2,2},{3,8,2},{5,3,5}};
    cout << "minimumEffortPath(standard) -> " << minimumEffortPath(h) << " (Expected: 2)\n";
}
{
    vvi h = {{1,2,3},{3,8,4},{5,3,5}};
    cout << "minimumEffortPath(variant) -> " << minimumEffortPath(h) << " (Expected: 1)\n";
}
{
    vvi h = {{1,2,1,1,1},{1,2,1,2,1},{1,2,1,2,1},{1,2,1,2,1},{1,1,1,2,1}};
    cout << "minimumEffortPath(flat-ish path exists) -> " << minimumEffortPath(h) << " (Expected: 0)\n";
}
{
    vvi h = {{5}};
    cout << "minimumEffortPath(single cell) -> " << minimumEffortPath(h) << " (Expected: 0)\n";
}
````

### 8.7.2 Cheapest Flights Within K Stops

**State definition.** Given flights (directed, weighted edges) and a stop limit $K$, find the
cheapest price from `src` to `dst` using **at most $K$ stops** (i.e. at most $K+1$ edges).

**Invariant — why plain Dijkstra fails here, and Bellman-Ford-style layering is the fix.** Dijkstra
(8.3) finalizes vertices by *globally* smallest distance, with no notion of "how many edges were
used to get there" — but this problem's constraint is fundamentally about edge-count, not just
total cost, so a smaller-cost-but-more-stops path might be found and finalized first, incorrectly
blocking a valid larger-cost-but-fewer-stops alternative from ever being considered relative to the
stop budget. The fix: run **Bellman-Ford-style relaxation restricted to exactly $K+1$ rounds**,
where round $r$ only uses paths of at most $r$ edges — directly reusing 8.5's "after $k$ passes,
dist correct for $\le k$-edge paths" invariant, just stopped early at $K+1$ instead of running to
$|V|-1$.

**Why it works.** Direct corollary of Bellman-Ford's inductive proof (8.5): the invariant "after
$k$ passes, dist[v] is correct for shortest paths using $\le k$ edges" is exactly what's needed —
stopping at $k = K+1$ passes gives exactly "shortest path using at most $K+1$ edges (at most $K$
stops)," which is precisely the problem's constraint. **Subtlety:** each pass must relax using the
*previous* pass's distance array (not update in place), or a single pass could "cheat" by chaining
multiple edge relaxations together within one round, silently violating the stop-count limit.

**The delta.** This problem exists specifically to test whether the stop-limit constraint is
recognized as an edge-count constraint, mapping directly onto Bellman-Ford's layer invariant rather
than Dijkstra's cost-only invariant — a wrong-but-plausible Dijkstra-based solution will pass many
test cases and fail exactly on inputs where the K-stops constraint and the raw cheapest path
disagree.


````c++
// Cheapest Flights Within K Stops: Bellman-Ford-style, exactly K+1 relaxation rounds using a PREVIOUS snapshot
#include <iostream>
#include <vector>
#include <climits>
using namespace std;
#define vi vector<int>
#define pii pair<int,int>

int findCheapestPrice(int n, vector<vector<int>>& flights, int src, int dst, int K) {
    vi dist(n, INT_MAX);
    dist[src] = 0;

    for (int round = 0; round <= K; round++) {          // K+1 edges allowed = K stops
        vi newDist = dist;                                // snapshot: relax using PREVIOUS round's values only
        for (auto& f : flights) {
            int u = f[0], v = f[1], w = f[2];
            if (dist[u] != INT_MAX && dist[u] + w < newDist[v]) {
                newDist[v] = dist[u] + w;                  // reads dist[u] (previous round), writes newDist[v]
            }
        }
        dist = newDist;                                     // commit this round's results together
    }

    return dist[dst] == INT_MAX ? -1 : dist[dst];
}
````

````c++
// Test cell: input -> actual (Expected: X)

{
    vector<vector<int>> flights = {{0,1,100},{1,2,100},{0,2,500}};
    cout << "findCheapestPrice(K=1) -> " << findCheapestPrice(3, flights, 0, 2, 1) << " (Expected: 200, via 0->1->2 using exactly 1 stop)\n";
}
{
    vector<vector<int>> flights = {{0,1,100},{1,2,100},{0,2,500}};
    cout << "findCheapestPrice(K=0) -> " << findCheapestPrice(3, flights, 0, 2, 0) << " (Expected: 500, direct flight only, 0 stops)\n";
}
{
    vector<vector<int>> flights = {{0,1,100}};
    cout << "findCheapestPrice(unreachable) -> " << findCheapestPrice(3, flights, 0, 2, 5) << " (Expected: -1)\n";
}
{
    vector<vector<int>> flights = {};
    cout << "findCheapestPrice(src==dst, no edges) -> " << findCheapestPrice(1, flights, 0, 0, 0) << " (Expected: 0)\n";
}
````

### 8.7.3 Network Delay Time

**State definition.** Given a directed weighted graph (signal travel times) and a source node,
find the time for a signal sent from the source to reach **all** nodes, or $-1$ if some node is
unreachable. Answer $= \max_v \text{dist}[v]$ (the last node to receive the signal determines total
elapsed time) — structurally identical to Rotten Oranges' "answer = max layer" (Subtopic 4.2), now
with weighted edges instead of unit time steps.

**Invariant.** Plain Dijkstra (8.3), unmodified — this problem is a direct, unmodified application.

**Why it works.** Direct corollary of 8.3's proof. The only new content is the answer-extraction
rule (`max` over all finalized distances, and detecting any still-$\infty$ distance as
"unreachable"), which mirrors Subtopic 4.2's Rotten Oranges pattern exactly, just with Dijkstra
substituted for multi-source BFS since edges are now weighted.

**The delta.** This problem is included specifically to demonstrate that the "answer = max over all
finalized distances, -1 if any unreachable" pattern from Subtopic 4.2 is **not** tied to BFS/unit
weights — it generalizes cleanly to any single-source shortest-path algorithm, because the
underlying reasoning (all nodes must be reached; the slowest one determines total elapsed time) has
nothing to do with how individual distances are computed.


````c++
// Network Delay Time: plain Dijkstra, answer = max distance (or -1 if any node unreachable)
#include <iostream>
#include <vector>
#include <queue>
#include <climits>
using namespace std;
#define vi vector<int>
#define pii pair<int,int>
#define vpii vector<pair<int,int>>
#define vvpii vector<vector<pii>>

int networkDelayTime(vector<vector<int>>& times, int n, int k) {
    vvpii adj(n+1);   // nodes are 1-indexed per typical problem convention
    for (auto& t : times) adj[t[0]].push_back({t[1], t[2]});

    vi dist(n+1, INT_MAX);
    priority_queue<pii, vpii, greater<pii>> pq;
    dist[k] = 0;
    pq.push({0, k});

    while (!pq.empty()) {
        auto [d, u] = pq.top(); pq.pop();
        if (d > dist[u]) continue;
        for (auto& [v, w] : adj[u]) {
            if (dist[u] + w < dist[v]) { dist[v] = dist[u] + w; pq.push({dist[v], v}); }
        }
    }

    int maxDist = 0;
    for (int v = 1; v <= n; v++) {
        if (dist[v] == INT_MAX) return -1;       // some node unreachable -- signal never fully propagates
        maxDist = max(maxDist, dist[v]);
    }
    return maxDist;
}
````

````c++
// Test cell: input -> actual (Expected: X)

{
    vector<vector<int>> times = {{2,1,1},{2,3,1},{3,4,1}};
    cout << "networkDelayTime(standard) -> " << networkDelayTime(times, 4, 2) << " (Expected: 2)\n";
}
{
    vector<vector<int>> times = {{1,2,1}};
    cout << "networkDelayTime(unreachable node) -> " << networkDelayTime(times, 2, 2) << " (Expected: -1)\n";
}
{
    vector<vector<int>> times = {{1,2,1}};
    cout << "networkDelayTime(single edge) -> " << networkDelayTime(times, 2, 1) << " (Expected: 1)\n";
}
{
    vector<vector<int>> times = {};
    cout << "networkDelayTime(single node, no edges) -> " << networkDelayTime(times, 1, 1) << " (Expected: 0)\n";
}
````

### 8.7.4 Number of Ways to Arrive at Destination

**State definition.** Given a weighted undirected graph, count the number of **distinct shortest
paths** from node 0 to node $n-1$ (modulo $10^9+7$).
$\text{ways}[v]$ = number of distinct shortest paths from source to $v$ (not just the distance).

**Invariant — the augmented Dijkstra invariant.** Alongside $\text{dist}[v]$, maintain
$\text{ways}[v]$, updated during relaxation:
- If a **strictly shorter** path to $v$ is found via $u$: $\text{dist}[v] \leftarrow
  \text{dist}[u]+w(u,v)$, and $\text{ways}[v] \leftarrow \text{ways}[u]$ (reset — the old count is
  no longer relevant to the new shortest distance).
- If a path to $v$ via $u$ **ties** the current shortest distance: $\text{ways}[v] \mathrel{+}=
  \text{ways}[u]$ (accumulate — this is an *additional* shortest path, not a replacement).

**Why it works.** By 8.3's extraction invariant, when $u$ is finalized, $\text{dist}[u]$ and (by
this augmented invariant, proven by the same induction) $\text{ways}[u]$ are both final and
correct. Every shortest path to $v$ must arrive via some finalized predecessor $u$ on the shortest-
path DAG (the subgraph of edges $(u,v)$ with $\text{dist}[u]+w(u,v)=\text{dist}[v]$) — summing
$\text{ways}[u]$ over all such predecessors, by the multiplication/addition principle of counting
paths through a DAG, gives the total path count. This is structurally the same "count paths through
a DAG via predecessor summation" idea as Subtopic 7's topological-order-based computations, applied
here on the *shortest-path DAG* Dijkstra implicitly builds.

**The delta.** This is Dijkstra augmented with a *second* piece of state that must obey its own
consistent update rule (reset-on-strictly-shorter, accumulate-on-tie) — a common pattern for
"count/reconstruct optimal solutions" problems layered on top of a pure optimization algorithm,
directly analogous to Word Ladder II's two-phase pattern (Subtopic 6.2), but folded into a single
Dijkstra pass here instead of a separate backtracking phase, because Dijkstra's finalize-in-order
processing already provides the correct dependency order for free.


````c++
// Number of Ways to Arrive at Destination: Dijkstra augmented with a path-count accumulator
#include <iostream>
#include <vector>
#include <queue>
#include <climits>
using namespace std;
#define vi vector<int>
#define vll vector<long long>
#define pii pair<long long,int>
#define vpii vector<pair<int,int>>
#define vvpii vector<vector<pair<int,int>>>

const long long MOD = 1e9 + 7;

int countPaths(int n, vector<vector<int>>& roads) {
    vvpii adj(n);
    for (auto& r : roads) {
        adj[r[0]].push_back({r[1], r[2]});
        adj[r[1]].push_back({r[0], r[2]});   // undirected
    }

    vll dist(n, LLONG_MAX);
    vll ways(n, 0);
    priority_queue<pii, vector<pii>, greater<pii>> pq;
    dist[0] = 0; ways[0] = 1;                 // base case: exactly one (empty) way to reach the source itself
    pq.push({0, 0});

    while (!pq.empty()) {
        auto [d, u] = pq.top(); pq.pop();
        if (d > dist[u]) continue;
        for (auto& [v, w] : adj[u]) {
            long long nd = dist[u] + w;
            if (nd < dist[v]) {
                dist[v] = nd;
                ways[v] = ways[u];             // strictly shorter path found: RESET the count
                pq.push({nd, v});
            } else if (nd == dist[v]) {
                ways[v] = (ways[v] + ways[u]) % MOD;   // tied shortest path: ACCUMULATE
            }
        }
    }
    return (int)(ways[n-1] % MOD);
}
````

````c++
// Test cell: input -> actual (Expected: X)

{
    vector<vector<int>> roads = {{0,6,7},{0,1,2},{1,2,3},{1,3,3},{6,3,3},{3,5,1},{6,5,1},{2,5,1},{0,4,5},{4,6,2}};
    cout << "countPaths(standard) -> " << countPaths(7, roads) << " (Expected: 4)\n";
}
{
    vector<vector<int>> roads = {{0,1,1}};
    cout << "countPaths(single edge) -> " << countPaths(2, roads) << " (Expected: 1)\n";
}
{
    vector<vector<int>> roads = {};
    cout << "countPaths(single node, src==dst) -> " << countPaths(1, roads) << " (Expected: 1)\n";
}
{
    // Two equally-short parallel-ish paths: 0-1-3 and 0-2-3, both length 2
    vector<vector<int>> roads = {{0,1,1},{1,3,1},{0,2,1},{2,3,1}};
    cout << "countPaths(two tied shortest paths) -> " << countPaths(4, roads) << " (Expected: 2)\n";
}
````

### 8.7.5 Minimum Multiplications to Reach End

**State definition.** Given a start value, an end value, and an array of multipliers, find the
minimum number of multiplication operations (each multiplying the current value by one array
element, then taking mod $100000$) to transform `start` into `end`. This is a direct return to
Subtopic 1.1's and Subtopic 6's implicit-graph framing: $V = \{0, \ldots, 99999\}$ (the space of
values mod $100000$), $\delta(u) = \{(u \times m) \bmod 100000 : m \in \text{multipliers}\}$.

**Invariant.** Exactly Subtopic 2.1's / Subtopic 6.1's BFS layer invariant — every operation
(edge) costs 1, so unweighted BFS gives the minimum operation count.

**Why the mod-$100000$ bound is what makes this a *well-posed* implicit graph.** Recall Subtopic
1.1's boundary table: an implicit graph's state space must be provably finite/bounded for BFS to be
guaranteed to terminate correctly and efficiently. Without the problem's mod-$100000$ constraint,
the reachable value set would be unbounded (values could grow arbitrarily via repeated
multiplication) — the mod operation is precisely what collapses the state space down to a fixed,
finite $|V| = 100000$, making this BFS-solvable in bounded time at all.

**The delta.** This problem is the clearest possible callback to Subtopic 1.1's original framing
example (the $x \to x{+}1, x{-}1, 2x$ minimum-operations problem) — same shape entirely, with
$\delta$ swapped for "multiply by each array element, mod $100000$." Recognizing "this is BFS on an
implicit graph over values mod $N$" immediately, rather than reaching for some numeric/DP technique,
is the entire skill being tested.


````c++
// Minimum Multiplications to Reach End: BFS on the implicit graph of values mod 100000
#include <iostream>
#include <vector>
#include <queue>
using namespace std;
#define vi vector<int>

const int MODV = 100000;

int minimumMultiplications(vector<int>& arr, int start, int end) {
    if (start == end) return 0;               // base case: already there, zero operations
    vi dist(MODV, -1);
    queue<int> q;
    dist[start] = 0;
    q.push(start);

    while (!q.empty()) {
        int u = q.front(); q.pop();
        for (int m : arr) {                    // delta(u): multiply by each array element, mod 100000
            int v = (int)(((long long)u * m) % MODV);
            if (dist[v] != -1) continue;        // visited-set discipline, finite state space guaranteed by mod
            dist[v] = dist[u] + 1;
            if (v == end) return dist[v];        // early exit: BFS layer invariant guarantees optimality
            q.push(v);
        }
    }
    return -1;   // end unreachable within the bounded state space
}
````

### 8.7.6 Find the City With Smallest Number of Neighbors at Threshold Distance

**State definition.** Given a weighted graph of cities, and a distance threshold, for every city
count how many *other* cities are reachable within the threshold distance. Return the city with the
**fewest** such reachable neighbors (tie-break: the city with the largest id).

**Invariant.** This is an all-pairs shortest-path query — needs $\text{dist}[i][j]$ for **every**
pair $(i,j)$, not just from one source. This is exactly the signal (per Subtopic 8's decision tree)
to reach for **Floyd-Warshall (8.6)** rather than running Dijkstra/Bellman-Ford $V$ separate times.

**Why it works.** Direct application of 8.6's DP: after computing the full $\text{dist}$ matrix,
for each city $i$, count $|\{j \neq i : \text{dist}[i][j] \le \text{threshold}\}|$ — a simple
$O(V^2)$ scan over the already-correct all-pairs matrix. No new proof is needed; this is purely a
post-processing query over Floyd-Warshall's output.

**The delta.** This problem is included specifically as the natural terminus of "how many sources
does this problem need" (the running question since Subtopic 4's decision tree) — the answer here
is unambiguously "all of them, for every pair," which is the exact condition under which
Floyd-Warshall's $O(V^3)$, source-count-independent cost becomes preferable to repeating a
single-source algorithm $V$ times.


````c++
// Find the City With Smallest Number of Neighbors at Threshold Distance: Floyd-Warshall + a counting scan
#include <iostream>
#include <vector>
#include <climits>
using namespace std;
#define vi vector<int>
#define vvi vector<vector<int>>

const int INF8 = INT_MAX / 2;

int findTheCity(int n, vector<vector<int>>& edges, int distanceThreshold) {
    vvi dist(n, vi(n, INF8));
    for (int i = 0; i < n; i++) dist[i][i] = 0;
    for (auto& e : edges) {
        dist[e[0]][e[1]] = min(dist[e[0]][e[1]], e[2]);
        dist[e[1]][e[0]] = min(dist[e[1]][e[0]], e[2]);   // undirected
    }

    // Floyd-Warshall (8.6), reused directly.
    for (int k = 0; k < n; k++)
        for (int i = 0; i < n; i++)
            for (int j = 0; j < n; j++)
                if (dist[i][k] + dist[k][j] < dist[i][j]) dist[i][j] = dist[i][k] + dist[k][j];

    int bestCity = -1, bestCount = INT_MAX;
    for (int i = 0; i < n; i++) {
        int count = 0;
        for (int j = 0; j < n; j++) if (j != i && dist[i][j] <= distanceThreshold) count++;
        if (count <= bestCount) { bestCount = count; bestCity = i; }   // <= keeps the LARGEST id on ties
    }
    return bestCity;
}
````

````c++
// Test cell: input -> actual (Expected: X)

{
    vector<vector<int>> edges = {{0,1,3},{1,2,1},{1,3,4},{2,3,1}};
    cout << "findTheCity(standard) -> " << findTheCity(4, edges, 4) << " (Expected: 3)\n";
}
{
    vector<vector<int>> edges = {{0,1,2},{0,4,8},{1,2,3},{1,4,2},{2,3,1},{3,4,1}};
    cout << "findTheCity(variant) -> " << findTheCity(5, edges, 2) << " (Expected: 0)\n";
}
{
    vector<vector<int>> edges = {{0,1,1}};
    cout << "findTheCity(two nodes) -> " << findTheCity(2, edges, 1) << " (Expected: 1, tie-break to larger id)\n";
}
````

````c++
// ================= UNIFIED MENTAL MODEL — Subtopic 8 =================
#include <iostream>
#include <vector>
#include <queue>
#include <climits>
using namespace std;
#define vi vector<int>
#define pii pair<int,int>
#define vpii vector<pair<int,int>>
#define vvpii vector<vector<pii>>

// --- Template: Dijkstra, the central pattern -- swap the relaxation operator for variants (8.7.1) ---
vi dijkstraTemplate(int V, vvpii& adj, int src) {
    vi dist(V, INT_MAX);
    priority_queue<pii, vpii, greater<pii>> pq;
    dist[src] = 0; pq.push({0, src});
    while (!pq.empty()) {
        auto [d, u] = pq.top(); pq.pop();
        if (d > dist[u]) continue;
        for (auto& [v, w] : adj[u]) {
            int nd = dist[u] + w;                 // swap this line for max(), etc. -- proof carries over (8.7.1)
            if (nd < dist[v]) { dist[v] = nd; pq.push({nd, v}); }
        }
    }
    return dist;
}

// --- Template: Bellman-Ford, layer-bounded variant for "at most K edges" constraints (8.7.2) ---
vi bellmanFordKLayerTemplate(int V, vector<array<int,3>>& edges, int src, int maxEdges) {
    vi dist(V, INT_MAX);
    dist[src] = 0;
    for (int round = 0; round <= maxEdges; round++) {
        vi newDist = dist;                         // snapshot -- do not update in place (8.7.2's subtlety)
        for (auto& e : edges) {
            int u = e[0], v = e[1], w = e[2];
            if (dist[u] != INT_MAX && dist[u] + w < newDist[v]) newDist[v] = dist[u] + w;
        }
        dist = newDist;
    }
    return dist;
}

// --- Template: Floyd-Warshall, all-pairs (8.6) ---
vector<vi> floydWarshallTemplate(int V, vector<vi>& weight) {
    vector<vi> dist = weight;
    for (int k = 0; k < V; k++)
        for (int i = 0; i < V; i++)
            for (int j = 0; j < V; j++)
                if (dist[i][k] + dist[k][j] < dist[i][j]) dist[i][j] = dist[i][k] + dist[k][j];
    return dist;
}

int main() {
    vvpii adj(3); adj[0]={{1,1}}; adj[1]={{2,1}};
    vi dist = dijkstraTemplate(3, adj, 0);
    cout << "Unified template sanity check: dist[2] -> " << dist[2] << " (Expected: 2)\n";
    return 0;
}
````

### Decision Tree — Subtopic 8

```
How many sources does the query actually need?
│
├── Exactly ONE source, need distance to some/all targets
│    │
│    Are ALL edge weights equal (unit weight)?
│    ├── YES → BFS (8.1) -- O(V+E), simplest, no heap needed
│    │
│    └── NO, weights vary -- are all weights NON-NEGATIVE?
│         ├── YES → is the graph a DAG?
│         │        ├── YES → topo sort + relaxation (8.2) -- O(V+E), handles negative weights too,
│         │        │         strictly better than Dijkstra when applicable
│         │        └── NO  → Dijkstra (8.3) -- O((V+E) log V) with a heap, or O(V^2+E) naive array
│         │                  version if the graph is dense (8.4's density crossover)
│         │
│         └── NO, some weights are negative
│                  → Bellman-Ford (8.5) -- O(V*E), ALSO detects negative cycles via one extra pass
│                    (unless the graph is a DAG, in which case 8.2 is still valid and faster)
│
└── ALL PAIRS needed (every source to every target)
     → Floyd-Warshall (8.6) -- O(V^3), independent of E, best when V is small/dense or
       all-pairs output is genuinely required


Is the relaxation rule actually addition, or something else (max, a modified constraint)?
├── Addition of edge weights → standard Dijkstra/Bellman-Ford/DAG relaxation as above
├── MAX of edge weights along the path (bottleneck path) → Dijkstra with max() relaxation (8.7.1),
│    same extraction proof, since max() is monotonic just like +
└── Constrained by edge-COUNT (e.g. "at most K stops") → Bellman-Ford, layer-bounded to K+1
     rounds using a snapshot-per-round (8.7.2) -- plain Dijkstra structurally cannot enforce this


Does the query need something BEYOND distance (path count, city with fewest neighbors, etc.)?
├── Count of shortest paths → augment Dijkstra with a second array, reset-on-strictly-shorter,
│    accumulate-on-tie (8.7.4)
├── All-pairs threshold counting → Floyd-Warshall, then a simple O(V^2) post-processing scan (8.7.6)
└── The "graph" isn't given directly (state-space search) → identify delta() and V (Subtopic 1.1/6),
     confirm the state space is finite/bounded, then apply plain unweighted BFS if all transitions
     cost equally (8.7.5)
```


### Complexity Summary — Subtopic 8

| Algorithm | Time | Space | When to use | Key invariant | Failure mode if misused |
|---|---|---|---|---|---|
| BFS, unit weights (8.1) | $O(V+E)$ | $O(V)$ | All edges cost the same | non-decreasing distance dequeue order | using it on weighted graphs silently ignores weight differences |
| DAG shortest path (8.2) | $O(V+E)$ | $O(V)$ | Graph is a DAG, weights arbitrary (incl. negative) | topo order guarantees all predecessors finalized first | running on a non-DAG (cyclic) graph — topo sort itself is undefined/fails |
| Dijkstra, heap (8.3) | $O((V+E)\log V)$ | $O(V+E)$ | Non-negative weights, sparse-ish graph | extracted vertex has final distance (needs non-negative weights) | negative edges silently produce wrong (too-large) finalized distances |
| Dijkstra, naive array (8.4) | $O(V^2+E)$ | $O(V+E)$ | Non-negative weights, dense graph | same as heap version, different minimum-finding cost | using this on sparse, large-V graphs wastes the $O(V^2)$ factor unnecessarily |
| Bellman-Ford (8.5) | $O(VE)$ | $O(V)$ | Negative weights allowed, need negative-cycle detection | after pass $k$, dist correct for $\le k$-edge paths | stopping before $|V|-1$ passes on a graph that needs them gives an under-relaxed, wrong answer |
| Floyd-Warshall (8.6) | $O(V^3)$ | $O(V^2)$ | All-pairs shortest paths, dense graphs, or V is small | dist[i][j][k] recurrence over allowed intermediates | using it for single-source queries wastes $O(V^3)$ vs. a single Dijkstra/Bellman-Ford run |
| Path with Minimum Effort (8.7.1) | $O((V+E)\log V)$ | $O(V+E)$ | "Minimize the max edge on the path" queries | max() relaxation is monotonic, same extraction proof as Dijkstra | using + instead of max silently solves the wrong objective |
| Cheapest Flights K Stops (8.7.2) | $O(K \cdot E)$ | $O(V)$ | Edge-count-bounded shortest path | per-round snapshot prevents over-counting edges within one round | updating dist in place instead of via a snapshot silently allows more than K+1 edges in one round |
| Network Delay Time (8.7.3) | $O((V+E)\log V)$ | $O(V+E)$ | Single-source, need max distance / full-coverage time | plain Dijkstra + max-of-all-finalized-distances | forgetting to check for any still-unreachable vertex before taking the max |
| Number of Ways to Arrive (8.7.4) | $O((V+E)\log V)$ | $O(V+E)$ | Need count of shortest paths, not just distance | reset-on-strictly-shorter, accumulate-on-tie | resetting on ties (instead of accumulating) undercounts; accumulating on strictly-shorter (instead of resetting) overcounts |
| Min Multiplications to Reach End (8.7.5) | $O(100000 \cdot \|arr\|)$ | $O(100000)$ | Implicit graph, all transitions cost 1, state space is bounded via mod | plain BFS layer invariant, state space finiteness enforced by the mod constraint | forgetting the mod bound (or getting it wrong) makes the state space effectively unbounded |
| Smallest Neighbors at Threshold (8.7.6) | $O(V^3)$ | $O(V^2)$ | All-pairs distances needed for a per-city count | Floyd-Warshall's dist matrix, then a direct $O(V^2)$ counting scan | running single-source shortest path $V$ times instead is correct but usually slower and more code |
