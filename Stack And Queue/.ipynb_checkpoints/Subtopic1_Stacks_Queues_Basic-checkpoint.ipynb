# Subtopic 1 — Stacks & Queues: Basic Implementation

**Problems:** Stack via Array, Queue via Array, Stack via Queue, Queue via Stack, Stack via Linked List, Queue via Linked List, Balanced Parentheses, Min Stack.

This notebook follows one discipline: **the invariant is stated before any code**. Every data structure here is a contract about what is always true between operations. Get the invariant right and the code writes itself; get it wrong and no amount of patching saves you.

## 1. Array-based Stack

### Invariant
At all times, a single index `top` points to the slot **above** the last valid element.

$$\text{valid elements occupy indices } [0,\ \text{top}) \qquad \text{top} \in [0, \text{capacity}]$$

- `top == 0` $\iff$ empty
- `top == capacity` $\iff$ full
- The element pushed most recently lives at index `top - 1` (the **LIFO** end).

### Why it is trivial
A stack only ever touches **one end**. Both `push` and `pop` mutate `top` by $\pm 1$ and touch exactly one array slot. No element ever moves. Hence:

$$\text{push}, \text{pop}, \text{top}, \text{empty} = O(1) \text{ worst case.}$$

There is no circular arithmetic because the live region always starts at index $0$.

### Boundary transitions
| State | `push(x)` | `pop()` |
|---|---|---|
| empty (`top=0`) | `a[0]=x; top=1` | error / underflow |
| mid (`0<top<cap`) | `a[top]=x; top++` | `top--; return a[top]` |
| full (`top=cap`) | error / overflow | `top--; return a[top]` |

### Delta vs Queue
The stack's live region is **anchored at 0**, so it never wraps. The queue's live region *drifts* (both ends move), which is exactly why the queue — and not the stack — needs modular arithmetic.

````c++
#include <iostream>
#include <stdexcept>
using namespace std;

template <typename T>
class ArrayStack {
    T* a;
    int cap;
    int top;            // INVARIANT: live elements are a[0 .. top-1]
public:
    ArrayStack(int capacity) : cap(capacity), top(0) { a = new T[cap]; }
    ~ArrayStack() { delete[] a; }

    bool empty() const { return top == 0; }       // top==0  <=>  no elements
    bool full()  const { return top == cap; }     // top==cap <=> array exhausted
    int  size()  const { return top; }

    void push(const T& x) {
        if (full()) throw overflow_error("stack overflow");
        a[top++] = x;          // write THEN advance: a[top] is the free slot
    }
    T pop() {
        if (empty()) throw underflow_error("stack underflow");
        return a[--top];       // retreat THEN read: --top is the last live slot
    }
    T peek() const {
        if (empty()) throw underflow_error("stack empty");
        return a[top - 1];     // top-1 is the most recently pushed element
    }
};
````

````c++
// Tests: input -> actual (Expected: X)
{
    ArrayStack<int> s(3);
    cout << "empty -> " << s.empty() << " (Expected: 1)\n";
    s.push(10); s.push(20); s.push(30);
    cout << "peek after 10,20,30 -> " << s.peek() << " (Expected: 30)\n";
    cout << "full -> " << s.full() << " (Expected: 1)\n";
    cout << "pop -> " << s.pop() << " (Expected: 30)\n";
    cout << "pop -> " << s.pop() << " (Expected: 20)\n";
    cout << "pop -> " << s.pop() << " (Expected: 10)\n";
    cout << "empty -> " << s.empty() << " (Expected: 1)\n";
    try { s.pop(); } catch (exception& e) { cout << "underflow caught (Expected)\n"; }

    ArrayStack<int> one(1);
    one.push(7);
    cout << "single peek -> " << one.peek() << " (Expected: 7)\n";
    try { one.push(9); } catch (exception& e) { cout << "overflow caught (Expected)\n"; }
}
````

## 2. Array-based Queue (Circular Buffer)

### The naive trap
If you dequeue by shifting every remaining element left by one, each dequeue is $O(n)$. Over $n$ dequeues that is $O(n^2)$. Unacceptable. The fix is to **let both ends drift** and wrap around the array — a circular buffer.

### Invariant
Two indices and a count:
- `front` — index of the oldest element (the next to be dequeued)
- `rear`  — index of the **next free slot** (where the next enqueue lands)
- `count` — number of live elements

The live region is the half-open arc, taken **modulo capacity**:

$$\text{indices } \{\, (\text{front} + k) \bmod C : 0 \le k < \text{count} \,\}$$

After every operation: `rear == (front + count) mod C`.

### Why `count` (the delta)
With wraparound, `front == rear` is **ambiguous** — it means empty *and* full. Two clean resolutions exist:
1. Keep an explicit `count` (used here) — simplest, no wasted slot.
2. Sacrifice one slot: `full` $\iff$ `(rear+1) mod C == front`; `empty` $\iff$ `front == rear`.

We use `count`. Then:

$$\text{empty} \iff \text{count}=0, \qquad \text{full} \iff \text{count}=C.$$

### Why modular arithmetic gives $O(1)$
$(\text{idx}+1) \bmod C$ advances an index and wraps it past the high end back to $0$ in constant time — no element ever physically moves. Hence enqueue and dequeue are $O(1)$ worst case.

### Boundary transitions
| State | `enqueue(x)` | `dequeue()` |
|---|---|---|
| empty (`count=0`) | `a[rear]=x; rear=(rear+1)%C; count++` | error |
| mid | same as above | `r=a[front]; front=(front+1)%C; count--` |
| full (`count=C`) | error | as mid |
| wrap edge (`rear=C-1`) | writes slot `C-1`, then `rear` wraps to `0` | symmetric for `front` |

````c++
#include <iostream>
#include <stdexcept>
using namespace std;

template <typename T>
class ArrayQueue {
    T* a;
    int cap, front, rear, count;
    // INVARIANT: live = { a[(front+k)%cap] : 0<=k<count };  rear==(front+count)%cap
public:
    ArrayQueue(int capacity)
        : cap(capacity), front(0), rear(0), count(0) { a = new T[cap]; }
    ~ArrayQueue() { delete[] a; }

    bool empty() const { return count == 0; }
    bool full()  const { return count == cap; }
    int  size()  const { return count; }

    void enqueue(const T& x) {
        if (full()) throw overflow_error("queue full");
        a[rear] = x;                 // rear is ALWAYS the free slot
        rear = (rear + 1) % cap;     // wrap: C-1 -> 0 in O(1), no shifting
        count++;
    }
    T dequeue() {
        if (empty()) throw underflow_error("queue empty");
        T r = a[front];              // front is the oldest live element
        front = (front + 1) % cap;   // advance the drifting head, wrapping
        count--;
        return r;
    }
    T peek() const {
        if (empty()) throw underflow_error("queue empty");
        return a[front];
    }
};
````

````c++
// Tests: input -> actual (Expected: X)
{
    ArrayQueue<int> q(3);
    cout << "empty -> " << q.empty() << " (Expected: 1)\n";
    q.enqueue(1); q.enqueue(2); q.enqueue(3);
    cout << "full -> " << q.full() << " (Expected: 1)\n";
    cout << "deq -> " << q.dequeue() << " (Expected: 1)\n";  // free a slot
    q.enqueue(4);                                            // forces wraparound
    cout << "deq -> " << q.dequeue() << " (Expected: 2)\n";
    cout << "deq -> " << q.dequeue() << " (Expected: 3)\n";
    cout << "deq -> " << q.dequeue() << " (Expected: 4)\n";  // the wrapped element
    cout << "empty -> " << q.empty() << " (Expected: 1)\n";
    try { q.dequeue(); } catch (exception& e) { cout << "underflow caught (Expected)\n"; }

    ArrayQueue<int> one(1);
    one.enqueue(99);
    cout << "single peek -> " << one.peek() << " (Expected: 99)\n";
    try { one.enqueue(0); } catch (exception& e) { cout << "overflow caught (Expected)\n"; }
}
````

## 3. Stack using Queue(s)

A queue is FIFO; a stack is LIFO. To impose LIFO on FIFO machinery we must, at some moment, **reverse the access order**. The only design choice is *when* to pay for that reversal — on `push` or on `pop`. This gives two dual approaches.

### LIFO invariant (must hold for both)
After every operation, dequeuing the queue front-to-back must yield elements in **reverse insertion order** (newest first). `pop()` simply removes the front.

---

### Approach A — Push-heavy ($O(n)$ push, $O(1)$ pop), single queue
On each push, enqueue the new element, then **rotate** the older $n-1$ elements behind it. This keeps the newest element permanently at the front.

$$\text{push: } 1 \text{ enqueue} + (n-1) \text{ rotations} = O(n); \qquad \text{pop/top} = O(1).$$

**Correctness:** by induction, before the push the front is the newest of the old elements. After enqueuing $x$ (now at back) and rotating $n-1$ items around, $x$ sits at front and every older element trails it in reverse-insertion order. Invariant preserved. $\square$

---

### Approach B — Pop-heavy ($O(1)$ push, $O(n)$ pop), two queues
Push = enqueue into `q1` ($O(1)$). Pop = move all but the last from `q1` to `q2`, the lone survivor is the answer, then swap roles.

$$\text{push} = O(1); \qquad \text{pop/top: move } n-1 \text{ then dequeue } 1 = O(n).$$

**Correctness:** `q1` holds elements in insertion order. The last-inserted element is the last one remaining after draining $n-1$ — exactly the LIFO target. $\square$

### Boundary transitions (Approach A)
| State | `push(x)` |
|---|---|
| empty | `q.enq(x)` — no rotation needed |
| size $k$ | `q.enq(x)`; rotate $k$ times so $x$ returns to front |

````c++
#include <iostream>
#include <queue>
#include <stdexcept>
using namespace std;

// Approach A: push-heavy, ONE queue. Newest element kept at front.
template <typename T>
class StackViaQueueA {
    queue<T> q;
public:
    void push(const T& x) {
        q.push(x);                       // x goes to the back
        int k = q.size() - 1;            // rotate the OLD elements around x
        while (k--) { q.push(q.front()); q.pop(); }  // back element cycles to front each step
        // after k rotations, x has bubbled to the front => newest-first order
    }
    T pop() {
        if (q.empty()) throw underflow_error("empty");
        T r = q.front(); q.pop();        // front is newest by invariant => O(1)
        return r;
    }
    T top() { if (q.empty()) throw underflow_error("empty"); return q.front(); }
    bool empty() const { return q.empty(); }
};

// Approach B: pop-heavy, TWO queues.
template <typename T>
class StackViaQueueB {
    queue<T> q1, q2;
public:
    void push(const T& x) { q1.push(x); }   // pure O(1)
    T pop() {
        if (q1.empty()) throw underflow_error("empty");
        while (q1.size() > 1) { q2.push(q1.front()); q1.pop(); } // drain all but last
        T r = q1.front(); q1.pop();          // the lone survivor = last inserted = LIFO
        swap(q1, q2);                         // q2 becomes the working queue
        return r;
    }
    bool empty() const { return q1.empty(); }
};
````

````c++
// Tests: input -> actual (Expected: X)
{
    StackViaQueueA<int> A;
    A.push(1); A.push(2); A.push(3);
    cout << "A top -> " << A.top() << " (Expected: 3)\n";
    cout << "A pop -> " << A.pop() << " (Expected: 3)\n";
    cout << "A pop -> " << A.pop() << " (Expected: 2)\n";
    A.push(4);
    cout << "A pop -> " << A.pop() << " (Expected: 4)\n";
    cout << "A pop -> " << A.pop() << " (Expected: 1)\n";

    StackViaQueueB<int> B;
    B.push(5); B.push(6); B.push(7);
    cout << "B pop -> " << B.pop() << " (Expected: 7)\n";
    cout << "B pop -> " << B.pop() << " (Expected: 6)\n";
    B.push(8);
    cout << "B pop -> " << B.pop() << " (Expected: 8)\n";
    cout << "B pop -> " << B.pop() << " (Expected: 5)\n";

    StackViaQueueA<int> one; one.push(42);
    cout << "single A pop -> " << one.pop() << " (Expected: 42)\n";
}
````

## 4. Queue using Stacks — the Two-Stack Trick

### Design
- `in`  — receives every `enqueue` (push onto `in`).
- `out` — serves every `dequeue` (pop from `out`).
- **Transfer rule:** when `out` is empty and we need to serve, pour *all* of `in` into `out`. Popping a stack into another stack reverses order, so the oldest element lands on top of `out` — FIFO restored.

### Invariant
Reading `out` top-to-bottom, then `in` **bottom-to-top**, yields the elements in queue order (oldest first). Equivalently: `out` holds a reversed prefix ready to serve; `in` holds the newer suffix in arrival order.

### Amortized $O(1)$ — potential method
Let the potential be the size of the input stack:

$$\Phi = |\text{in}| \ge 0, \qquad \Phi_0 = 0.$$

Amortized cost $\hat{c} = c + \Delta\Phi$:

- **enqueue:** one push ($c=1$), $\Delta\Phi = +1 \Rightarrow \hat{c} = 2 = O(1)$.
- **dequeue, `out` non-empty:** one pop ($c=1$), $\Delta\Phi = 0 \Rightarrow \hat{c} = 1$.
- **dequeue, `out` empty, $k$ items in `in`:** $k$ pops + $k$ pushes + final pop, $c = 2k+1$; but $\Delta\Phi = -k$, so $\hat{c} = 2k+1-k = k+1$… 

The clean statement: **each element is moved across at most once** — pushed to `in` once, transferred to `out` once, popped from `out` once: $\le 3$ touches over its lifetime. Total work for any sequence of $m$ operations is $O(m)$, hence $O(1)$ amortized per operation. The $\Phi$ accounting confirms the expensive transfer is fully prepaid by the cheap enqueues that built up `in`. $\square$

### Boundary transitions
| State | `enqueue(x)` | `dequeue()` |
|---|---|---|
| both empty | `in.push(x)` | error |
| `out` non-empty | `in.push(x)` | `out.pop()` — $O(1)$ |
| `out` empty, `in` non-empty | `in.push(x)` | transfer all `in`→`out`, then `out.pop()` |

````c++
#include <iostream>
#include <stack>
#include <stdexcept>
using namespace std;

template <typename T>
class QueueViaStacks {
    stack<T> in, out;        // INVARIANT: out (top->bottom) then in (bottom->top) = queue order
    void shift() {           // called ONLY when out is empty
        while (!in.empty()) { out.push(in.top()); in.pop(); }  // reversal restores FIFO
    }
public:
    void enqueue(const T& x) { in.push(x); }   // always O(1)

    T dequeue() {
        if (out.empty()) shift();              // amortized: each elem shifted at most once
        if (out.empty()) throw underflow_error("empty");
        T r = out.top(); out.pop();
        return r;
    }
    T peek() {
        if (out.empty()) shift();
        if (out.empty()) throw underflow_error("empty");
        return out.top();
    }
    bool empty() const { return in.empty() && out.empty(); }
};
````

````c++
// Tests: input -> actual (Expected: X)
{
    QueueViaStacks<int> q;
    cout << "empty -> " << q.empty() << " (Expected: 1)\n";
    q.enqueue(1); q.enqueue(2); q.enqueue(3);
    cout << "deq -> " << q.dequeue() << " (Expected: 1)\n"; // triggers transfer
    q.enqueue(4);                                           // mixes new into 'in'
    cout << "peek -> " << q.peek()  << " (Expected: 2)\n";
    cout << "deq -> " << q.dequeue() << " (Expected: 2)\n";
    cout << "deq -> " << q.dequeue() << " (Expected: 3)\n";
    cout << "deq -> " << q.dequeue() << " (Expected: 4)\n"; // second transfer
    cout << "empty -> " << q.empty() << " (Expected: 1)\n";
    try { q.dequeue(); } catch (exception& e) { cout << "underflow caught (Expected)\n"; }

    QueueViaStacks<int> one; one.enqueue(7);
    cout << "single deq -> " << one.dequeue() << " (Expected: 7)\n";
}
````

## 5. Linked-List Stack

### Invariant
`head` points to the most recently pushed node; each node's `next` points to the element pushed just before it. The chain read from `head` is exactly the stack top-to-bottom. `head == nullptr` $\iff$ empty.

### Why it is natural
**Prepend = push.** Both `push` and `pop` only touch `head`:

$$\text{push: new node} \to \text{next} = \text{head};\ \text{head} = \text{new node} \quad (O(1))$$
$$\text{pop: } \text{head} = \text{head} \to \text{next} \quad (O(1))$$

No tail pointer is needed — a stack never touches its far end. This is the cleanest possible LL structure: the natural insertion point (the front) coincides with the only end a stack uses.

### Boundary transitions
| State | `push(x)` | `pop()` |
|---|---|---|
| empty | `head = node(x)` | error |
| non-empty | `node->next = head; head = node` | save `head->val`; `head = head->next`; free |

````c++
#include <iostream>
#include <stdexcept>
using namespace std;

template <typename T>
class LinkedStack {
    struct Node { T val; Node* next; Node(const T& v, Node* n): val(v), next(n) {} };
    Node* head = nullptr;     // INVARIANT: head = most recent; chain = top..bottom
    int   n    = 0;
public:
    ~LinkedStack() { while (head) pop(); }

    void push(const T& x) {
        head = new Node(x, head);   // prepend: new node points at old top, becomes top
        n++;
    }
    T pop() {
        if (!head) throw underflow_error("empty");
        Node* old = head;
        T r = old->val;
        head = head->next;          // detach top; second node becomes new top
        delete old;
        n--;
        return r;
    }
    T peek() const { if (!head) throw underflow_error("empty"); return head->val; }
    bool empty() const { return head == nullptr; }
    int  size()  const { return n; }
};
````

````c++
// Tests: input -> actual (Expected: X)
{
    LinkedStack<int> s;
    cout << "empty -> " << s.empty() << " (Expected: 1)\n";
    s.push(1); s.push(2); s.push(3);
    cout << "size -> " << s.size() << " (Expected: 3)\n";
    cout << "peek -> " << s.peek() << " (Expected: 3)\n";
    cout << "pop -> " << s.pop() << " (Expected: 3)\n";
    cout << "pop -> " << s.pop() << " (Expected: 2)\n";
    cout << "pop -> " << s.pop() << " (Expected: 1)\n";
    cout << "empty -> " << s.empty() << " (Expected: 1)\n";
    try { s.pop(); } catch (exception& e) { cout << "underflow caught (Expected)\n"; }
}
````

## 6. Linked-List Queue — Why a Tail Pointer

### The problem
A queue inserts at one end (rear) and removes at the other (front). With only a `head` pointer, enqueue must walk the entire list to find the last node — $O(n)$. **Keeping a `tail` pointer makes enqueue $O(1)$.**

### Invariant
- `head` — oldest element (dequeue end).
- `tail` — newest element (enqueue end); `tail->next == nullptr`.
- Empty $\iff$ `head == nullptr` $\iff$ `tail == nullptr` (both must be kept in sync).

$$\text{enqueue: append after tail, advance tail} = O(1)$$
$$\text{dequeue: detach head, advance head} = O(1)$$

### The single-pointer edge cases
Without care, the empty $\leftrightarrow$ one-element transitions are error-prone: enqueue into empty must set **both** `head` and `tail`; dequeue of the last element must reset **both** to `nullptr`.

### The dummy-head trick (delta)
A permanent **sentinel** node sits before the real front. Then `head` is never null, so:
- enqueue always does `tail->next = node; tail = node` — never a special "was empty" branch.
- the real queue starts at `sentinel->next`.

This trades one wasted node for the elimination of all empty/non-empty branching. Below we show the clean two-pointer version; the sentinel variant is noted in comments.

### Boundary transitions
| State | `enqueue(x)` | `dequeue()` |
|---|---|---|
| empty | `head = tail = node` | error |
| one elem | `tail->next = node; tail = node` | detach head; set `head = tail = nullptr` |
| many | `tail->next = node; tail = node` | `head = head->next` |

````c++
#include <iostream>
#include <stdexcept>
using namespace std;

template <typename T>
class LinkedQueue {
    struct Node { T val; Node* next; Node(const T& v): val(v), next(nullptr) {} };
    Node* head = nullptr;   // oldest (dequeue here)
    Node* tail = nullptr;   // newest (enqueue here); tail->next == nullptr
    int   n    = 0;
    // INVARIANT: head==nullptr  <=>  tail==nullptr  <=>  empty
public:
    ~LinkedQueue() { while (head) dequeue(); }

    void enqueue(const T& x) {
        Node* node = new Node(x);
        if (!tail) head = tail = node;     // empty -> both pointers point at the lone node
        else { tail->next = node; tail = node; } // append after tail, advance tail (O(1))
        n++;
    }
    T dequeue() {
        if (!head) throw underflow_error("empty");
        Node* old = head;
        T r = old->val;
        head = head->next;                 // advance front
        if (!head) tail = nullptr;         // removed the LAST element -> reset tail too
        delete old;
        n--;
        return r;
    }
    T peek() const { if (!head) throw underflow_error("empty"); return head->val; }
    bool empty() const { return head == nullptr; }
    int  size()  const { return n; }
};
````

````c++
// Tests: input -> actual (Expected: X)
{
    LinkedQueue<int> q;
    cout << "empty -> " << q.empty() << " (Expected: 1)\n";
    q.enqueue(1);
    cout << "single peek -> " << q.peek() << " (Expected: 1)\n";
    cout << "single deq -> " << q.dequeue() << " (Expected: 1)\n"; // empties -> tail must reset
    cout << "empty -> " << q.empty() << " (Expected: 1)\n";
    q.enqueue(2); q.enqueue(3); q.enqueue(4);  // re-enqueue after emptying
    cout << "size -> " << q.size() << " (Expected: 3)\n";
    cout << "deq -> " << q.dequeue() << " (Expected: 2)\n";
    cout << "deq -> " << q.dequeue() << " (Expected: 3)\n";
    cout << "deq -> " << q.dequeue() << " (Expected: 4)\n";
    try { q.dequeue(); } catch (exception& e) { cout << "underflow caught (Expected)\n"; }
}
````

## 7. Balanced Parentheses

### Invariant
At every point during the scan, the stack contains **exactly the currently-unmatched opening brackets, in order of appearance**, with the most recent on top. A closing bracket is valid iff it matches the top — because brackets must close in **last-opened-first-closed** (LIFO) order.

### Algorithm
Scan left to right:
- opening bracket → push it.
- closing bracket → stack must be non-empty **and** its top must be the matching opener; else reject. Pop on match.
- end of string → balanced iff stack is empty (no unmatched openers remain).

### Correctness (both directions)
**Soundness** (accepts $\Rightarrow$ balanced): every pop pairs a closer with the nearest unmatched opener, and the proper-nesting LIFO order is exactly what the stack enforces. Ending empty means every opener was paired. 

**Completeness** (balanced $\Rightarrow$ accepts): in a balanced string, every closer's matching opener is the most recent unmatched one (proper nesting), which is precisely the stack top; so no valid string is ever rejected. $\square$

### Boundary transitions
| Char | Stack empty | Stack top matches | Stack top mismatches |
|---|---|---|---|
| opener `([{` | push | push | push |
| closer `)]}` | **reject** (nothing to close) | pop, continue | **reject** |
| end of input | **accept** | n/a | n/a |

### Tricky inputs
- `""` → stack empty at end → **balanced** (vacuously true).
- `]` → closer with empty stack → reject.
- `([)]` → at `)`, top is `[` not `(` → reject (interleaving is not nesting).
- `([])` → fully matches → balanced.

````c++
#include <iostream>
#include <stack>
#include <string>
using namespace std;

bool isBalanced(const string& s) {
    stack<char> st;   // INVARIANT: st holds unmatched openers, in order, newest on top
    for (char c : s) {
        if (c == '(' || c == '[' || c == '{') {
            st.push(c);                       // record an unmatched opener
        } else if (c == ')' || c == ']' || c == '}') {
            if (st.empty()) return false;     // closer with nothing to match
            char t = st.top();
            // top must be the SAME bracket family as this closer
            bool ok = (c == ')' && t == '(') ||
                      (c == ']' && t == '[') ||
                      (c == '}' && t == '{');
            if (!ok) return false;            // wrong nesting order, e.g. ([)]
            st.pop();                         // matched -> remove the opener
        }
        // non-bracket chars (if any) are ignored
    }
    return st.empty();                        // balanced iff no opener left unmatched
}
````

````c++
// Tests: input -> actual (Expected: X)
{
    cout << "\"\"      -> " << isBalanced("")       << " (Expected: 1)\n";
    cout << "()      -> " << isBalanced("()")      << " (Expected: 1)\n";
    cout << "([])    -> " << isBalanced("([])")    << " (Expected: 1)\n";
    cout << "([{}])  -> " << isBalanced("([{}])")  << " (Expected: 1)\n";
    cout << "]       -> " << isBalanced("]")       << " (Expected: 0)\n";
    cout << "(       -> " << isBalanced("(")       << " (Expected: 0)\n";
    cout << "([)]    -> " << isBalanced("([)]")    << " (Expected: 0)\n";
    cout << "{[(])}  -> " << isBalanced("{[(])}")  << " (Expected: 0)\n";
    cout << "(((     -> " << isBalanced("(((")     << " (Expected: 0)\n";
}
````

## 8. Min Stack — `getMin()` in $O(1)$

Goal: `push`, `pop`, `top`, `getMin` all $O(1)$. The challenge: after a pop, the minimum may change, and we cannot rescan ($O(n)$). The minimum must be **recoverable per element**.

---

### Approach 1 — Parallel min-stack
Keep a second stack `mins` where `mins.top()` is the minimum of all elements currently in the main stack. On push, push `min(x, mins.top())`. On pop, pop both.

- **Invariant:** `mins[k]` = minimum of `main[0..k]`. Thus `mins.top()` is always the global current min.
- **Cost:** $O(1)$ all ops, **$O(n)$ extra space** (one min entry per element).

---

### Approach 2 — Single stack, delta encoding ($O(1)$ extra space)
Store **encoded** values in one stack and track `currentMin` in a single variable.

**Push encoding.** To push value `v`:
- if stack empty: `currentMin = v`, push `v`.
- else if `v >= currentMin`: push `v` unchanged (a "normal" entry, always $\ge$ current min).
- else (`v < currentMin`, new minimum): push the **delta**

$$\text{encoded} = 2v - \text{currentMin} \quad(\text{i.e. } v - (\text{currentMin}-v) = v + (v-\text{currentMin}))$$

then set `currentMin = v`. Since `v < currentMin`, we have `encoded = 2v - currentMin < v` — strictly **below** the value, a flag that "this slot rewrote the min".

**Recovering the value on `top()`:**
- if `encoded >= currentMin`: it is a normal entry, true value `= encoded`.
- if `encoded <  currentMin`: it was a min-rewrite, true value `= currentMin` (the new min equals the pushed value).

*Proof value is recoverable:* normal entries are stored verbatim. For a rewrite entry, by construction the pushed value **is** the current min, so returning `currentMin` returns `v`. $\square$

**Recovering the previous min on `pop()`:** when popping a rewrite entry (`encoded < currentMin`), restore

$$\text{prevMin} = 2\cdot\text{currentMin} - \text{encoded}.$$

*Proof:* at push time `encoded = 2v - prevMin` and `currentMin` was set to `v`. So `2·currentMin - encoded = 2v - (2v - prevMin) = prevMin`. The old minimum is exactly reconstructed. $\square$

**Why it works with one stack & $O(1)$ overhead per element:** each slot stores a single number; min history is folded into the deltas of rewrite entries and unwound on pop. No parallel structure — vs Approach 1's $O(n)$ second stack.

> **Overflow caveat (the delta):** `2v - currentMin` can overflow fixed-width integers when values straddle the type range. Use a wider type (`long long`) for the stack, or prefer Approach 1 / Approach 3 if inputs span the full `int` range. This is the practical price of the clever encoding.

---

### Approach 3 — Stack of pairs `(value, minSoFar)`
Each entry carries its own snapshot of the min at insertion time. Trivially correct, $O(1)$ everything, $O(n)$ space — the most readable, interview-safe option.

### Boundary transitions (Approach 2 push)
| Condition | Action | currentMin after |
|---|---|---|
| empty | push `v` | `v` |
| `v >= currentMin` | push `v` (normal) | unchanged |
| `v <  currentMin` | push `2v - currentMin` (rewrite) | `v` |

````c++
#include <iostream>
#include <stack>
#include <vector>
#include <utility>
#include <stdexcept>
#include <algorithm>
using namespace std;

// ---- Approach 1: parallel min-stack ----
class MinStackParallel {
    stack<int> s, mins;   // mins.top() == min of all in s
public:
    void push(int x) { s.push(x); mins.push(mins.empty() ? x : min(x, mins.top())); }
    void pop() { if (s.empty()) throw underflow_error("empty"); s.pop(); mins.pop(); }
    int  top() { if (s.empty()) throw underflow_error("empty"); return s.top(); }
    int  getMin() { if (mins.empty()) throw underflow_error("empty"); return mins.top(); }
    bool empty() const { return s.empty(); }
};

// ---- Approach 2: single stack, delta encoding, O(1) extra space ----
class MinStackEncoded {
    stack<long long> s;   // stores encoded values (long long to blunt overflow)
    long long curMin = 0;
public:
    void push(int x) {
        long long v = x;
        if (s.empty()) { curMin = v; s.push(v); }    // first element: stored verbatim
        else if (v >= curMin) { s.push(v); }         // normal entry: >= current min
        else { s.push(2*v - curMin); curMin = v; }   // rewrite: encode delta, update min
    }
    void pop() {
        if (s.empty()) throw underflow_error("empty");
        long long e = s.top(); s.pop();
        if (e < curMin) curMin = 2*curMin - e;       // unwind: restore previous min
    }
    int top() {
        if (s.empty()) throw underflow_error("empty");
        long long e = s.top();
        return (int)(e >= curMin ? e : curMin);      // rewrite slot -> true value is curMin
    }
    int getMin() { if (s.empty()) throw underflow_error("empty"); return (int)curMin; }
    bool empty() const { return s.empty(); }
};

// ---- Approach 3: stack of pairs ----
class MinStackPairs {
    stack<pair<int,int>> s;   // (value, minSoFar)
public:
    void push(int x) { int m = s.empty() ? x : min(x, s.top().second); s.push({x, m}); }
    void pop() { if (s.empty()) throw underflow_error("empty"); s.pop(); }
    int  top() { if (s.empty()) throw underflow_error("empty"); return s.top().first; }
    int  getMin() { if (s.empty()) throw underflow_error("empty"); return s.top().second; }
    bool empty() const { return s.empty(); }
};
````

````c++
// Tests: input -> actual (Expected: X) -- run all three for parity
{
    auto run = [](auto& ms, const string& name) {
        ms.push(5); ms.push(3); ms.push(7); ms.push(2);
        cout << name << " min after 5,3,7,2 -> " << ms.getMin() << " (Expected: 2)\n";
        cout << name << " top -> " << ms.top() << " (Expected: 2)\n";
        ms.pop();   // remove 2 -> min back to 3
        cout << name << " min after pop -> " << ms.getMin() << " (Expected: 3)\n";
        ms.pop();   // remove 7 -> min still 3
        cout << name << " min after pop -> " << ms.getMin() << " (Expected: 3)\n";
        ms.pop();   // remove 3 -> min back to 5
        cout << name << " min after pop -> " << ms.getMin() << " (Expected: 5)\n";
        cout << name << " top -> " << ms.top() << " (Expected: 5)\n";
    };
    MinStackParallel a; run(a, "P1");
    MinStackEncoded  b; run(b, "P2");
    MinStackPairs    c; run(c, "P3");

    // all-same values
    MinStackEncoded same; same.push(4); same.push(4); same.push(4);
    cout << "all-same min -> " << same.getMin() << " (Expected: 4)\n";
    same.pop();
    cout << "all-same min after pop -> " << same.getMin() << " (Expected: 4)\n";

    // single element
    MinStackEncoded one; one.push(9);
    cout << "single min -> " << one.getMin() << " (Expected: 9)\n";
    cout << "single top -> " << one.top() << " (Expected: 9)\n";
}
````

## Unified Mental Model

Every structure here is a statement about **which end you are allowed to touch**, plus the bookkeeping that keeps that end $O(1)$.

```text
                 single end (LIFO)              two ends (FIFO)
                 ---------------                ----------------
   array         top index, anchored at 0       front/rear + count, MODULAR (wrap)
   linked list   head pointer only              head + TAIL pointer

   cross-build:  Stack via Queue  -> reverse-on-push OR reverse-on-pop
                 Queue via Stack  -> two stacks, reverse on transfer (amortized O(1))

   augmentation: Balanced ()  -> stack = unmatched openers (LIFO matching)
                 Min Stack    -> carry the min per element:
                                  parallel stack | delta encoding | pair snapshot
```

The single unifying idea: **a stack reverses order; a queue preserves it.** Whenever you must convert one to the other, you pay exactly one reversal — the only freedom is *when* you pay it, and amortization lets you spread that payment so each operation is $O(1)$ on average.

## Decision Tree — Which Pattern?

```text
Need a container with O(1) ends?
│
├─ Access only the most-recent element?            ─────────────► STACK
│     ├─ fixed capacity, no allocation churn?      → array stack (top index)
│     └─ unbounded / pointer-based notes?          → linked stack (head only)
│
├─ Access oldest first, insert newest last?        ─────────────► QUEUE
│     ├─ fixed capacity?                            → circular array (front/rear/count, mod)
│     └─ unbounded?                                 → linked queue (head + TAIL)
│
├─ Only have queues but need LIFO?                 ─────────────► STACK via QUEUE
│     ├─ many pushes, few pops?                     → pop-heavy (O(1) push)
│     └─ many pops, few pushes?                     → push-heavy (O(1) pop)
│
├─ Only have stacks but need FIFO?                 ─────────────► QUEUE via 2 STACKS (amortized O(1))
│
├─ Must validate nesting / matching?               ─────────────► STACK of openers
│
└─ Need running min/max alongside LIFO?            ─────────────► MIN STACK
      ├─ readability / safety first?                → pairs or parallel stack
      └─ O(1) extra space, values fit safely?       → delta encoding
```

## Complexity Summary

| Operation | Time | Space | Amortized | Key Invariant |
|---|---|---|---|---|
| Array Stack push/pop/top | $O(1)$ | $O(n)$ | — | live = `a[0..top)`, anchored at 0 |
| Array Queue enq/deq | $O(1)$ | $O(n)$ | — | `rear=(front+count) mod C`; wrap is free |
| Stack via Queue (push-heavy) | push $O(n)$, pop $O(1)$ | $O(n)$ | — | newest kept at queue front |
| Stack via Queue (pop-heavy) | push $O(1)$, pop $O(n)$ | $O(n)$ | — | last-inserted survives the drain |
| Queue via 2 Stacks enq/deq | $O(n)$ worst | $O(n)$ | $O(1)$ | each elem crosses `in`→`out` once; $\Phi=\lvert\text{in}\rvert$ |
| Linked Stack push/pop | $O(1)$ | $O(n)$ | — | `head` = top; prepend = push |
| Linked Queue enq/deq | $O(1)$ | $O(n)$ | — | `tail->next=null`; sync both on empty |
| Balanced Parentheses | $O(n)$ | $O(n)$ | — | stack = unmatched openers, LIFO |
| Min Stack (parallel/pairs) | $O(1)$ | $O(n)$ | — | min snapshot stored per element |
| Min Stack (delta encoding) | $O(1)$ | $O(1)$ extra | — | rewrite entry `< curMin`; `prevMin=2·curMin−enc` |
