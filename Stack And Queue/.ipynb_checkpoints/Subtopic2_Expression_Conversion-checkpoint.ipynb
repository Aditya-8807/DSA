# Subtopic 2 — Expression Conversion

**Problems:** Infix→Postfix, Prefix→Infix, Prefix→Postfix, Postfix→Prefix, Postfix→Infix, Infix→Prefix.

Six conversions, but only **two engines** underneath:
1. **Infix→Postfix (Shunting-Yard)** — an *operator* stack resolving precedence.
2. **Linear scan with an *operand* (sub-expression) stack** — used by every Postfix/Prefix conversion.

Everything else is reduction: Infix→Prefix is Infix→Postfix wrapped in two reversals; the four Postfix/Prefix conversions are the same operand-stack loop with different output assembly. Master the two engines and the invariants transfer wholesale.

## The Three Notations

For a binary operator $\circ$ with operands $A, B$:

| Notation | Form | Operator position | Example $a+b\times c$ |
|---|---|---|---|
| **Infix** | $A \circ B$ | between | $a + b \times c$ |
| **Postfix** (RPN, Reverse Polish) | $A\,B\,\circ$ | after | $a\ b\ c \times +$ |
| **Prefix** (Polish) | $\circ\,A\,B$ | before | $+\ a \times b\ c$ |

### Why Postfix and Prefix need no parentheses
Infix is **ambiguous without conventions**: $a+b\times c$ requires a precedence rule (× before +) and sometimes parentheses to override it. The position of the operator in Postfix/Prefix encodes the grouping **structurally** — the operator's place in the string *is* the parse tree, read in a fixed traversal order:

$$\text{Postfix} = \text{post-order traversal}, \qquad \text{Prefix} = \text{pre-order traversal}, \qquad \text{Infix} = \text{in-order traversal}$$

of the expression tree. Post-order and pre-order are **unambiguous for a fixed arity** (each binary operator consumes exactly its two preceding/following sub-results), so no precedence rule and no parentheses are ever required. In-order loses the bracketing, which is why infix alone needs them.

### The structural consequence
- A **Postfix** scan goes **left→right**: when you meet an operator, its two operands are the two most-recently-completed sub-expressions (top of an operand stack).
- A **Prefix** scan goes **right→left**: same logic, mirrored — when you meet an operator, its two operands are the two most-recently-completed sub-expressions to its right.

## Operator Precedence & Associativity

| Operator | Precedence | Associativity |
|---|---|---|
| `^` (power) | 3 (highest) | **right** |
| `*` `/` | 2 | left |
| `+` `-` | 1 (lowest) | left |
| `(` | — (lowest while on stack) | — |

### The `>=` vs `>` rule (the central delta)
In Shunting-Yard, when the incoming operator $o_2$ meets a stack-top operator $o_1$, we must decide: **pop $o_1$ first, or push $o_2$ on top of it?** This is governed entirely by associativity:

- **Left-associative** $o_2$ (like `+`, `*`): pop while $\text{prec}(o_1) \ge \text{prec}(o_2)$.
  Equal precedence means "evaluate the earlier one first" — e.g. $a-b-c = (a-b)-c$ — so we must flush $o_1$ before pushing $o_2$. Hence **$\ge$** (pop on a tie).

- **Right-associative** $o_2$ (like `^`): pop while $\text{prec}(o_1) > \text{prec}(o_2)$.
  Equal precedence means "evaluate the later one first" — e.g. $a\verb|^|b\verb|^|c = a\verb|^|(b\verb|^|c)$ — so we must **keep** $o_1$ on the stack and stack $o_2$ above it. Hence **$>$** (do *not* pop on a tie).

$$\boxed{\text{pop condition} = \begin{cases} \text{prec}(o_1) \ge \text{prec}(o_2) & o_2 \text{ left-associative} \\ \text{prec}(o_1) > \text{prec}(o_2) & o_2 \text{ right-associative} \end{cases}}$$

This single line is the only place associativity enters the algorithm.

````c++
#include <iostream>
#include <stack>
#include <string>
#include <algorithm>
#include <cctype>
using namespace std;

// Shared helpers used across all conversions in this notebook.
int prec(char op) {
    if (op == '^') return 3;
    if (op == '*' || op == '/') return 2;
    if (op == '+' || op == '-') return 1;
    return 0;                       // '(' and non-operators sit below everything
}
bool isRightAssoc(char op) { return op == '^'; }   // only power is right-associative here
bool isOperator(char c)    { return c=='+'||c=='-'||c=='*'||c=='/'||c=='^'; }
bool isOperand(char c)     { return isalnum((unsigned char)c); }  // single-char ids/digits
````

## 1. Infix → Postfix (Shunting-Yard)

### Invariant
The operator stack holds, from bottom to top, operators **awaiting their right operand**, in **non-decreasing effective precedence** within each parenthesis group. An operator stays on the stack exactly as long as no later operator of $\le$ precedence (for left-assoc) forces it out. Output receives operands immediately and operators only once both their operands are guaranteed to have been emitted.

### Boundary transitions
| Token | Action |
|---|---|
| operand | append to output |
| `(` | push (acts as a precedence floor) |
| `)` | pop to output until `(`; discard the `(` |
| operator $o_2$ | while top is an operator and pop-condition (above) holds, pop to output; then push $o_2$ |
| end of input | pop all remaining operators to output |

### Correctness
Each operator is emitted **after** both its operands: its left operand was emitted before the operator was pushed (it appeared earlier in the scan and was either an operand token or flushed out by precedence), and its right operand is emitted before the operator is finally popped (the operator only leaves the stack when a lower/equal operator or `)` or end-of-input signals its right operand is complete). Thus the output is a valid post-order serialization — it evaluates to the same value as the infix input for all valid inputs. $\square$

````c++
string infixToPostfix(const string& s) {
    string out;
    stack<char> ops;   // INVARIANT: operators awaiting their right operand, by precedence
    for (char c : s) {
        if (isspace((unsigned char)c)) continue;
        if (isOperand(c)) {
            out += c;                              // operands flow straight to output
        } else if (c == '(') {
            ops.push(c);                           // precedence floor: nothing pops past it
        } else if (c == ')') {
            while (!ops.empty() && ops.top() != '(') { out += ops.top(); ops.pop(); }
            if (!ops.empty()) ops.pop();           // discard the matching '('
        } else if (isOperator(c)) {
            // pop higher (or equal, if c is left-assoc) precedence operators first
            while (!ops.empty() && ops.top() != '(' &&
                   (prec(ops.top()) > prec(c) ||
                    (prec(ops.top()) == prec(c) && !isRightAssoc(c)))) {
                out += ops.top(); ops.pop();
            }
            ops.push(c);                           // c now waits for its right operand
        }
    }
    while (!ops.empty()) { out += ops.top(); ops.pop(); }  // flush remaining operators
    return out;
}
````

````c++
// Tests: input -> actual (Expected: X)
{
    cout << "a+b*c       -> " << infixToPostfix("a+b*c")       << " (Expected: abc*+)\n";
    cout << "(a+b)*c     -> " << infixToPostfix("(a+b)*c")     << " (Expected: ab+c*)\n";
    cout << "a+b-c       -> " << infixToPostfix("a+b-c")       << " (Expected: ab+c-)\n"; // left-assoc
    cout << "a^b^c       -> " << infixToPostfix("a^b^c")       << " (Expected: abc^^)\n"; // right-assoc
    cout << "(a+b)*(c-d) -> " << infixToPostfix("(a+b)*(c-d)") << " (Expected: ab+cd-*)\n";
    cout << "a+b*c-d/e   -> " << infixToPostfix("a+b*c-d/e")   << " (Expected: abc*+de/-)\n";
    cout << "a           -> " << infixToPostfix("a")           << " (Expected: a)\n"; // single operand
    cout << "((a))       -> " << infixToPostfix("((a))")       << " (Expected: a)\n"; // nested parens
}
````

## 2. Infix → Prefix (reverse · Infix→Postfix · reverse)

### Algorithm
1. **Reverse** the infix string, swapping every `(` with `)` and vice versa.
2. Convert the reversed string to postfix — but with one tweak: to keep `^` (right-assoc) correct under reversal, treat ties using **`>`** for left-assoc operators and **`>=`** for `^`. (Equivalently: flip the tie rule.)
3. **Reverse** the resulting postfix string. The result is the prefix expression.

### Why reversal works
Prefix is the **mirror image** of postfix in a precise sense: reading an expression tree's post-order on the *left-right-mirrored* tree, then reversing the string, yields the pre-order of the original tree. Reversing the input swaps left/right children at every node; converting to postfix serializes that mirrored tree post-order; reversing the output undoes the string-level mirror, leaving a correct pre-order = prefix.

### The tie-rule flip (the subtle delta)
After reversing the string, operator associativity effectively flips for tie-breaking. Left-associative operators (`+ - * /`) must now use strict **`>`** (do not pop on tie), and the right-associative `^` must use **`>=`** (pop on tie). Skipping this flip silently breaks chains like `a-b-c` and `a^b^c`.

### Boundary transitions
Same table as Infix→Postfix, applied to the reversed string with swapped brackets, plus the flipped tie rule. End: reverse the postfix output.

````c++
string infixToPrefix(const string& s) {
    // Step 1: reverse and swap brackets
    string r;
    for (int i = (int)s.size() - 1; i >= 0; --i) {
        char c = s[i];
        if      (c == '(') r += ')';
        else if (c == ')') r += '(';
        else               r += c;
    }
    // Step 2: infix->postfix on reversed string with FLIPPED tie rule
    string out;
    stack<char> ops;
    for (char c : r) {
        if (isspace((unsigned char)c)) continue;
        if (isOperand(c)) { out += c; }
        else if (c == '(') { ops.push(c); }
        else if (c == ')') {
            while (!ops.empty() && ops.top() != '(') { out += ops.top(); ops.pop(); }
            if (!ops.empty()) ops.pop();
        } else if (isOperator(c)) {
            // FLIP: left-assoc uses '>' (no pop on tie); '^' uses '>=' (pop on tie)
            while (!ops.empty() && ops.top() != '(' &&
                   (prec(ops.top()) > prec(c) ||
                    (prec(ops.top()) == prec(c) && isRightAssoc(c)))) {
                out += ops.top(); ops.pop();
            }
            ops.push(c);
        }
    }
    while (!ops.empty()) { out += ops.top(); ops.pop(); }
    // Step 3: reverse postfix -> prefix
    reverse(out.begin(), out.end());
    return out;
}
````

````c++
// Tests: input -> actual (Expected: X)
{
    cout << "a+b*c       -> " << infixToPrefix("a+b*c")       << " (Expected: +a*bc)\n";
    cout << "(a+b)*c     -> " << infixToPrefix("(a+b)*c")     << " (Expected: *+abc)\n";
    cout << "a+b-c       -> " << infixToPrefix("a+b-c")       << " (Expected: -+abc)\n"; // left-assoc
    cout << "a^b^c       -> " << infixToPrefix("a^b^c")       << " (Expected: ^a^bc)\n"; // right-assoc
    cout << "(a+b)*(c-d) -> " << infixToPrefix("(a+b)*(c-d)") << " (Expected: *+ab-cd)\n";
    cout << "a+b*c-d/e   -> " << infixToPrefix("a+b*c-d/e")   << " (Expected: -+a*bc/de)\n";
    cout << "a           -> " << infixToPrefix("a")           << " (Expected: a)\n";
}
````

## Engine 2 — The Operand (Sub-expression) Stack

The remaining four conversions all share one loop. Scan the source; whenever you meet an **operand**, push it (as a finished sub-expression string) onto an operand stack; whenever you meet an **operator**, **pop exactly two** sub-expressions, combine them into a larger sub-expression in the *target* notation, and push the result back.

### Unified invariant
At every step, the operand stack holds completed sub-expressions in the **target notation**, ordered so that the next operator's operands are exactly the top two (with correct left/right orientation). When the scan ends, **exactly one** element remains — the full converted expression.

### Scan direction & operand orientation — the only two knobs
| Source | Scan direction | On operator, operands are |
|---|---|---|
| **Postfix** | left → right | `B = pop()` (top = right), `A = pop()` (next = left) |
| **Prefix** | right → left | `A = pop()` (top = left), `B = pop()` (next = right) |

The asymmetry: in postfix the right operand is completed last, so it is on top; in prefix (scanned right→left) the left operand is completed last. Get this backwards and `-` / `/` / `^` come out reversed.

### 3. Postfix → Infix
Combine as `(A op B)` with parentheses to preserve grouping unconditionally (safe, never wrong; can be minimized but parentheses guarantee correctness).

### 4. Postfix → Prefix
Combine as `op A B`.

### Correctness (both)
By the invariant, when operator `op` is read its two operands `A` (left) and `B` (right) are the completed sub-expressions on top. Wrapping them in the target form yields a sub-expression denoting `A op B`, identical in value to what the postfix encodes. Induction on stack size: each operator reduces the stack by one (pop 2, push 1); a well-formed postfix of $n$ operators and $n+1$ operands terminates with a single element = the whole expression. $\square$

````c++
// Source = POSTFIX, scan LEFT->RIGHT. B is top (right operand), A is next (left).
string postfixToInfix(const string& s) {
    stack<string> st;   // INVARIANT: completed infix sub-expressions
    for (char c : s) {
        if (isspace((unsigned char)c)) continue;
        if (isOperand(c)) { st.push(string(1, c)); }
        else if (isOperator(c)) {
            string B = st.top(); st.pop();      // top completed last = RIGHT operand
            string A = st.top(); st.pop();      // next = LEFT operand
            st.push("(" + A + c + B + ")");     // parenthesize to preserve grouping
        }
    }
    return st.top();                            // lone survivor = full expression
}

string postfixToPrefix(const string& s) {
    stack<string> st;   // INVARIANT: completed prefix sub-expressions
    for (char c : s) {
        if (isspace((unsigned char)c)) continue;
        if (isOperand(c)) { st.push(string(1, c)); }
        else if (isOperator(c)) {
            string B = st.top(); st.pop();      // RIGHT operand
            string A = st.top(); st.pop();      // LEFT operand
            st.push(string(1, c) + A + B);      // prefix: op then left then right
        }
    }
    return st.top();
}
````

````c++
// Tests: input -> actual (Expected: X)
{
    cout << "abc*+   -> infix  " << postfixToInfix("abc*+")   << " (Expected: (a+(b*c)))\n";
    cout << "ab+c*   -> infix  " << postfixToInfix("ab+c*")   << " (Expected: ((a+b)*c))\n";
    cout << "ab+cd-* -> infix  " << postfixToInfix("ab+cd-*") << " (Expected: ((a+b)*(c-d)))\n";
    cout << "abc^^   -> infix  " << postfixToInfix("abc^^")   << " (Expected: (a^(b^c)))\n";

    cout << "abc*+   -> prefix " << postfixToPrefix("abc*+")   << " (Expected: +a*bc)\n";
    cout << "ab+c*   -> prefix " << postfixToPrefix("ab+c*")   << " (Expected: *+abc)\n";
    cout << "ab+cd-* -> prefix " << postfixToPrefix("ab+cd-*") << " (Expected: *+ab-cd)\n";
    cout << "a       -> prefix " << postfixToPrefix("a")       << " (Expected: a)\n"; // single operand
}
````

### 5. Prefix → Infix  &  6. Prefix → Postfix

Same operand-stack loop, but **scan right → left**, and the orientation flips: the element completed last (now on top) is the **left** operand.

- **Prefix → Infix:** combine as `(A op B)`.
- **Prefix → Postfix:** combine as `A B op`.

### Correctness
Scanning right→left, an operator's two operands are the sub-expressions immediately to its right, which are precisely the two most-recently-completed (top two on the stack). With `A = pop()` (top = left) and `B = pop()` (next = right), the combined form denotes `A op B` — value-preserving. Termination identical to the postfix case. $\square$

### Boundary transitions (operand-stack engine, all four)
| Token | Action |
|---|---|
| operand | push single-char sub-expression |
| operator | pop two, combine in target notation with correct L/R, push result |
| end of scan | stack has exactly one element → the answer |

````c++
// Source = PREFIX, scan RIGHT->LEFT. A is top (LEFT operand), B is next (right).
string prefixToInfix(const string& s) {
    stack<string> st;
    for (int i = (int)s.size() - 1; i >= 0; --i) {
        char c = s[i];
        if (isspace((unsigned char)c)) continue;
        if (isOperand(c)) { st.push(string(1, c)); }
        else if (isOperator(c)) {
            string A = st.top(); st.pop();      // top (completed last) = LEFT operand
            string B = st.top(); st.pop();      // next = RIGHT operand
            st.push("(" + A + c + B + ")");
        }
    }
    return st.top();
}

string prefixToPostfix(const string& s) {
    stack<string> st;
    for (int i = (int)s.size() - 1; i >= 0; --i) {
        char c = s[i];
        if (isspace((unsigned char)c)) continue;
        if (isOperand(c)) { st.push(string(1, c)); }
        else if (isOperator(c)) {
            string A = st.top(); st.pop();      // LEFT operand
            string B = st.top(); st.pop();      // RIGHT operand
            st.push(A + B + string(1, c));      // postfix: left right op
        }
    }
    return st.top();
}
````

````c++
// Tests: input -> actual (Expected: X)
{
    cout << "+a*bc    -> infix   " << prefixToInfix("+a*bc")    << " (Expected: (a+(b*c)))\n";
    cout << "*+abc    -> infix   " << prefixToInfix("*+abc")    << " (Expected: ((a+b)*c))\n";
    cout << "*+ab-cd  -> infix   " << prefixToInfix("*+ab-cd")  << " (Expected: ((a+b)*(c-d)))\n";
    cout << "^a^bc    -> infix   " << prefixToInfix("^a^bc")    << " (Expected: (a^(b^c)))\n";

    cout << "+a*bc    -> postfix " << prefixToPostfix("+a*bc")   << " (Expected: abc*+)\n";
    cout << "*+abc    -> postfix " << prefixToPostfix("*+abc")   << " (Expected: ab+c*)\n";
    cout << "*+ab-cd  -> postfix " << prefixToPostfix("*+ab-cd") << " (Expected: ab+cd-*)\n";
    cout << "a        -> postfix " << prefixToPostfix("a")       << " (Expected: a)\n";
}
````

### Round-trip consistency

The strongest correctness check: chaining conversions must return to the start. We verify a few cycles like Infix→Postfix→Prefix→Postfix and Infix→Prefix→Infix→Postfix agree.

````c++
// Tests: round trips -> actual (Expected: match)
{
    string inf = "a+b*c-d/e";
    string post = infixToPostfix(inf);                 // abc*+de/-
    string pre  = infixToPrefix(inf);                  // -+a*bc/de

    cout << "post(inf)               -> " << post << " (Expected: abc*+de/-)\n";
    cout << "pre(inf)                -> " << pre  << " (Expected: -+a*bc/de)\n";
    cout << "post -> prefix          -> " << postfixToPrefix(post) << " (Expected: " << pre << ")\n";
    cout << "pre  -> postfix         -> " << prefixToPostfix(pre)  << " (Expected: " << post << ")\n";
    cout << "post -> infix -> post   -> "
         << infixToPostfix(postfixToInfix(post)) << " (Expected: " << post << ")\n";

    // power chain (right-assoc) round trip
    string p = infixToPostfix("a^b^c");                // abc^^
    cout << "a^b^c post              -> " << p << " (Expected: abc^^)\n";
    cout << "  -> prefix             -> " << postfixToPrefix(p) << " (Expected: ^a^bc)\n";
}
````

## Unified Mental Model

```text
            TWO ENGINES, SIX CONVERSIONS
            ============================

  ENGINE 1: OPERATOR STACK (precedence resolver)
  ----------------------------------------------
    Infix -> Postfix   : scan L->R, pop by precedence, '>=' left / '>' right
    Infix -> Prefix    : reverse+swap()  ->  Engine 1 with FLIPPED tie  ->  reverse

  ENGINE 2: OPERAND STACK (sub-expression builder)
  ------------------------------------------------
    source POSTFIX -> scan L->R : B=pop (right), A=pop (left)
    source PREFIX  -> scan R->L : A=pop (left),  B=pop (right)
       combine into target form, push back; 1 element remains at end

         target form per conversion:
            -> infix   : (A op B)          [+ parentheses]
            -> postfix : A B op
            -> prefix  : op A B
```

**The one idea:** Postfix and Prefix *are* the parse tree (post-order / pre-order). Converting between them is just choosing a traversal and a scan direction. Infix is the odd one out — it needs an operator stack on the way *out* (to add structure) and full parenthesization on the way *in* (to recover structure).

## Decision Tree — Which Conversion Machinery?

```text
Is the SOURCE infix?
│
├─ YES  ── target postfix?  → Engine 1 (operator stack), '>=' left / '>' right
│         target prefix?    → reverse+swap brackets → Engine 1 (FLIP tie) → reverse output
│
└─ NO (source is postfix or prefix) → Engine 2 (operand stack)
          │
          ├─ source POSTFIX → scan LEFT→RIGHT ;  B=pop()=right, A=pop()=left
          └─ source PREFIX  → scan RIGHT→LEFT ;  A=pop()=left,  B=pop()=right
          │
          └─ assemble in target notation:
                infix  → "(" + A + op + B + ")"
                postfix→ A + B + op
                prefix → op + A + B
```

## Complexity Summary

All six conversions are single linear scans with $O(1)$ stack work per token.

| Conversion | Time | Space | Engine | Key Invariant |
|---|---|---|---|---|
| Infix → Postfix | $O(n)$ | $O(n)$ | operator stack | stack = operators awaiting right operand, by precedence; `>=` left / `>` right |
| Infix → Prefix | $O(n)$ | $O(n)$ | reverse + operator stack + reverse | same, with **flipped** tie rule on reversed string |
| Postfix → Infix | $O(n)$ | $O(n)$ | operand stack, L→R | top two = (right, left); combine `(A op B)` |
| Postfix → Prefix | $O(n)$ | $O(n)$ | operand stack, L→R | top two = (right, left); combine `op A B` |
| Prefix → Infix | $O(n)$ | $O(n)$ | operand stack, R→L | top two = (left, right); combine `(A op B)` |
| Prefix → Postfix | $O(n)$ | $O(n)$ | operand stack, R→L | top two = (left, right); combine `A B op` |

> Note on operands: these implementations assume **single-character** operands (the classic interview form). For multi-character tokens, tokenize first and push/emit tokens with separators — the invariants are unchanged.
